## Review: Integration

### Step 1: $f : \mathbb R \to \mathbb R$ on an interval

For an integrable function $f : \mathbb R \to \mathbb R$,  
the definite integral over $[a, b]$ is the limit of signed rectangular sums:
$$ \int_a^b f(x)\,dx
= \lim_{\|\mathcal P\| \to 0}
\sum_j f(x_j^*)\,\Delta x_j. $$
The differential $dx$ records the oriented displacement along the real line.  
Consequently,
$$ \int_b^a f(x)\,dx = -\int_a^b f(x)\,dx. $$

### Step 2: $g : \mathbb R^2 \to \mathbb R^2$ along a curve $\gamma$

Let
$$ \gamma(t) = (x(t), y(t)),\qquad a \le t \le b, $$
be a piecewise continuously differentiable curve,  
and let $g = (P, Q)$ be a vector field.  
The line integral of $g$ along $\gamma$ is
$$ \int_\gamma g \mathbin{\boldsymbol{\cdot}} d\boldsymbol r
= \int_a^b g(\gamma(t)) \mathbin{\boldsymbol{\cdot}} \gamma'(t)\,dt. $$
It measures the tangential accumulation of the field along the curve.

#### Riemann-sum form

For a partition
$$ a = t_0 < t_1 < \cdots < t_n = b $$
and sample points $t_j^* \in [t_{j - 1}, t_j]$, the line integral is
$$ \int_\gamma g \mathbin{\boldsymbol{\cdot}} d\boldsymbol r
= \lim_{\|\mathcal P\| \to 0}
\sum_{j = 1}^n g(\gamma(t_j^*)) \mathbin{\boldsymbol{\cdot}}
\left(\gamma(t_j) - \gamma(t_{j - 1})\right). $$
Each term pairs the field sampled on one small arc  
with that arc's displacement vector.  
Since $\Delta\boldsymbol r_j \approx \gamma'(t_j^*)\Delta t_j$,  
the limit gives the parametrized integral in the next cell.

In coordinates,
$$ \int_\gamma g \mathbin{\boldsymbol{\cdot}} d\boldsymbol r
= \int_\gamma\left(P\,dx + Q\,dy\right)
= \int_a^b
\left(P(x(t), y(t))x'(t) + Q(x(t), y(t))y'(t)\right)dt. $$
An orientation-preserving reparametrization does not change the integral.  
Reversing the direction of $\gamma$ changes its sign.

### Step 3: $h : \mathbb C \to \mathbb C$ along a curve $\gamma$

Let $\gamma : [a, b] \to \mathbb C$ be piecewise continuously differentiable.  
For a continuous function $h : \mathbb C \to \mathbb C$, define
$$ \int_\gamma h(z)\,dz
= \int_a^b h(\gamma(t))\gamma'(t)\,dt. $$
This is formally the same parametrized integral as before,  
but the product is now complex multiplication.  
It is unchanged by orientation-preserving reparametrization  
and changes sign when the curve is reversed.

Write
$$ h(z) = u(x, y) + iv(x, y),\qquad dz = dx + i\,dy. $$
Then
$$ h(z)\,dz
= (u\,dx - v\,dy) + i(v\,dx + u\,dy), $$
so
$$ \int_\gamma h(z)\,dz
= \int_\gamma\left(u\,dx - v\,dy\right)
+ i\int_\gamma\left(v\,dx + u\,dy\right). $$
A complex line integral is therefore two real vector-field line integrals  
carried by the same oriented curve.

### The three constructions

| Integrand | Domain of integration | Definition after parametrization |
|---|---|---|
| $f : \mathbb R \to \mathbb R$ | Interval $[a, b]$ | $\displaystyle\int_a^b f(x)\,dx$ |
| $g : \mathbb R^2 \to \mathbb R^2$ | Curve $\gamma$ in $\mathbb R^2$ | $\displaystyle\int_a^b g(\gamma(t)) \mathbin{\boldsymbol{\cdot}} \gamma'(t)\,dt$ |
| $h : \mathbb C \to \mathbb C$ | Curve $\gamma$ in $\mathbb C$ | $\displaystyle\int_a^b h(\gamma(t))\gamma'(t)\,dt$ |

Green's theorem relates the second type, taken around a closed curve,  
to an ordinary double integral over the enclosed planar region.

## Green's Theorem and the Shoelace Formula

### Green's theorem in the plane

Let $D \subset \mathbb R^2$ be a bounded region whose boundary $C = \partial D$  
consists of finitely many positively oriented, piecewise smooth, simple closed curves.  
Let $P$ and $Q$ have continuous first partial derivatives  
on an open set containing $D$ and its boundary.

Positive orientation means that $D$ stays on the left as we traverse $C$.  
Thus an outer boundary runs counterclockwise; boundaries around holes run clockwise.

### Circulation form

Green's theorem converts a line integral around the boundary  
into a double integral over the region:
$$ \oint_C\left(P\,dx + Q\,dy\right)
= \iint_D\left(\frac{\partial Q}{\partial x}
- \frac{\partial P}{\partial y}\right)\,dA. $$
The integrand on the right is the scalar curl of the planar vector field $(P, Q)$.  
The theorem says that total microscopic rotation inside $D$  
equals macroscopic circulation around $\partial D$.

### Flux form

For a counterclockwise parametrization, the outward normal satisfies  
$\boldsymbol n\,ds = (dy, -dx)$.  
Therefore Green's theorem also gives
$$ \oint_C\left(P\,dy - Q\,dx\right)
= \iint_D\left(\frac{\partial P}{\partial x}
+ \frac{\partial Q}{\partial y}\right)\,dA. $$
The integrand is the divergence of $(P, Q)$,  
so this form equates outward boundary flux with total interior source strength.

### Green's theorem and complex line integrals

Let $h = u + iv$ and let $C = \partial D$ be positively oriented.  
Applying Green's theorem to the two real components of $h\,dz$ gives
$$ \oint_C h(z)\,dz
= -\iint_D\left(\frac{\partial v}{\partial x}
+ \frac{\partial u}{\partial y}\right)\,dA
+ i\iint_D\left(\frac{\partial u}{\partial x}
- \frac{\partial v}{\partial y}\right)\,dA. $$
If $h$ satisfies the Cauchy–Riemann equations, both integrands vanish.  
This is the Green's theorem mechanism behind Cauchy's theorem  
when the region and the partial derivatives satisfy the stated hypotheses.

### Area from Green's theorem

Choose
$$ P(x, y) = -\frac{y}{2},\qquad Q(x, y) = \frac{x}{2}. $$
Then
$$ \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} = 1, $$
so Green's theorem yields the signed-area formula
$$ A_{\mathrm{signed}}
= \frac{1}{2}\oint_C\left(x\,dy - y\,dx\right). $$
It is positive for counterclockwise orientation and negative for clockwise orientation.  
The ordinary area is $|A_{\mathrm{signed}}|$.

### The shoelace formula

Let a simple polygon have vertices
$$ p_k = (x_k, y_k),\qquad k = 1, \ldots, n, $$
listed in boundary order, and set $p_{n + 1} = p_1$.  
Its signed area is
$$ A_{\mathrm{signed}}
= \frac{1}{2}\sum_{k = 1}^n
\left(x_k y_{k + 1} - y_k x_{k + 1}\right). $$
Writing the $x$- and $y$-coordinates in two columns makes the diagonal products  
look like the crossed laces of a shoe, which gives the formula its name.

#### Contribution of one edge

Parametrize the edge from $p_k$ to $p_{k + 1}$ by
$$ x(t) = (1 - t)x_k + t x_{k + 1},\qquad
y(t) = (1 - t)y_k + t y_{k + 1},\qquad 0 \le t \le 1. $$
Along this edge,
$$ \int_0^1\left(x(t)y'(t) - y(t)x'(t)\right)\,dt
= x_k y_{k + 1} - y_k x_{k + 1}. $$
Summing the edge contributions in the area integral obtained from Green's theorem  
produces the shoelace formula.

### Example

Consider the counterclockwise polygon
$$ (0, 0),\quad (4, 0),\quad (4, 2),\quad (1, 3). $$
The edge determinants are
$$ 0,\quad 8,\quad 10,\quad 0. $$
Hence
$$ A = \frac{1}{2}(0 + 8 + 10 + 0) = 9. $$
Reversing the order of the vertices changes every determinant's sign,  
so the signed area becomes $-9$ while the ordinary area remains $9$.

### Complex-number form

Write each vertex as $z_k = x_k + iy_k$.  
Then
$$ \operatorname{Im}\left(\overline{z_k}z_{k + 1}\right)
= x_k y_{k + 1} - y_k x_{k + 1}, $$
and the shoelace formula becomes
$$ A_{\mathrm{signed}}
= \frac{1}{2}\sum_{k = 1}^n
\operatorname{Im}\left(\overline{z_k}z_{k + 1}\right). $$
For a general positively oriented boundary,
$$ A = \frac{1}{2}\operatorname{Im}\oint_C\overline{z}\,dz. $$

### Orientation and self-intersections

For a simple polygon, the sign records the orientation,  
and the absolute value gives the usual geometric area.  
The formula works for both convex and nonconvex simple polygons.

For a self-intersecting polygonal path, the same sum gives **algebraic area**:  
regions traversed with opposite orientations can cancel.  
Computing the total geometric area of all visible lobes then requires  
splitting the path into simple regions or accounting for winding numbers.
