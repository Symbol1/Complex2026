## Convergence of Series

Power series are the main tool of this lecture.  
Before using them, we review when an infinite series of complex numbers converges,  
and the standard tests that decide it.

### Convergence and the Cauchy criterion

A series $\sum_{n = 0}^\infty c_n$ of complex numbers **converges** to $S$  
if its partial sums $S_N = c_0 + c_1 + \cdots + c_N$ tend to $S$.  
Since $|\operatorname{Re}w| \le |w|$, $|\operatorname{Im}w| \le |w|$, and $|w| \le |\operatorname{Re}w| + |\operatorname{Im}w|$,  
a complex series converges exactly when its real and imaginary parts both converge.

**The $n$th-term test.**  
If $\sum c_n$ converges, then $c_N = S_N - S_{N - 1} \to S - S = 0$.  
So a series whose terms do not tend to $0$ diverges.  
The converse is false:  
the terms of the harmonic series $\sum 1/n$ tend to $0$,  
but the series diverges, as we will see.

By the completeness of $\mathbb C$,  
a sequence converges if and only if it is a Cauchy sequence.  
Since $S_M - S_N = c_{N + 1} + \cdots + c_M$, for partial sums this reads as follows.

**Cauchy criterion.**  
The series $\sum c_n$ converges if and only if for every $\varepsilon > 0$ there is $N_0$ such that
$$ |c_{N + 1} + c_{N + 2} + \cdots + c_M| < \varepsilon\qquad\text{for all }M > N \ge N_0. $$
In words: the tails eventually contribute arbitrarily little.

### Absolute and conditional convergence

The series $\sum c_n$ **converges absolutely** if $\sum|c_n|$ converges.  
Absolute convergence implies convergence.  
Indeed, by the triangle inequality,
$$ |c_{N + 1} + \cdots + c_M| \le |c_{N + 1}| + \cdots + |c_M|, $$
and the right-hand side satisfies the Cauchy criterion because $\sum|c_n|$ converges.  
A series that converges but does not converge absolutely is said to **converge conditionally**.

Absolute convergence reduces complex series to series of nonnegative real numbers.  
The partial sums of a series with nonnegative terms increase,  
so such a series converges exactly when its partial sums are bounded.  
Most of the tests below are ways of bounding these partial sums.

### Geometric series and $p$-series

For $q \ne 1$, the finite geometric sum is
$$ 1 + q + \cdots + q^N = \frac{1 - q^{N + 1}}{1 - q}. $$
If $|q| < 1$, then $q^{N + 1} \to 0$, so
$$ \sum_{n = 0}^\infty q^n = \frac{1}{1 - q}\qquad(|q| < 1), $$
and the convergence is absolute, since $\sum|q|^n$ is also geometric.  
If $|q| \ge 1$, then $|q^n| \ge 1$, and the series diverges by the $n$th-term test.

**Integral test.**  
Let $f$ be positive and decreasing on $[1, \infty)$, and let $c_n = f(n)$.  
Since $f(n + 1) \le \int_n^{n + 1}f(x)\,dx \le f(n)$, summing over $n$ gives
$$ \sum_{n = 2}^N c_n \le \int_1^N f(x)\,dx \le \sum_{n = 1}^{N - 1}c_n. $$
So $\sum c_n$ converges if and only if $\int_1^\infty f(x)\,dx$ is finite.

Applying the integral test to $f(x) = x^{-p}$ with $p > 0$,
$$ \sum_{n = 1}^\infty\frac{1}{n^p}\ \text{converges}
\iff \int_1^\infty\frac{dx}{x^p} < \infty
\iff p > 1. $$
In particular, the harmonic series $\sum 1/n$ diverges, with partial sums growing like $\ln N$,  
while $\sum 1/n^2$ converges.  
For $p \le 0$ the terms do not tend to $0$.

### Comparison tests

**Comparison test.**  
If $|c_n| \le b_n$ for all large $n$ and $\sum b_n$ converges, then $\sum c_n$ converges absolutely.  
If $0 \le b_n \le c_n$ for all large $n$ and $\sum b_n$ diverges, then $\sum c_n$ diverges.  
For example, $|e^{in}/n^2| = 1/n^2$, so $\sum e^{in}/n^2$ converges absolutely;  
and $1/\ln n \ge 1/n$ for $n \ge 2$, so $\sum_{n \ge 2}1/\ln n$ diverges.

**Limit comparison test.**  
If $c_n > 0$, $b_n > 0$, and $c_n/b_n \to L$ with $0 < L < \infty$,  
then $\sum c_n$ and $\sum b_n$ converge or diverge together,  
because $\frac L2 b_n \le c_n \le 2Lb_n$ for all large $n$.  
For example,
$$ \frac{(n + 1)/(n^3 - 2n + 5)}{1/n^2} = \frac{n^3 + n^2}{n^3 - 2n + 5} \to 1, $$
so $\sum(n + 1)/(n^3 - 2n + 5)$ converges, like $\sum 1/n^2$.

### Ratio and root tests

**Ratio test.**  
Suppose $|c_{n + 1}/c_n| \to L$.

- If $L < 1$, choose $q$ with $L < q < 1$.  
  Then $|c_{n + 1}| \le q|c_n|$ for $n \ge n_0$, so $|c_n| \le |c_{n_0}|q^{n - n_0}$,  
  and the series converges absolutely by comparison with a geometric series.
- If $L > 1$, then $|c_n|$ eventually increases, so $c_n \not\to 0$ and the series diverges.
- If $L = 1$, the test is inconclusive: $\sum 1/n$ and $\sum 1/n^2$ both have $L = 1$.

**Root test.**  
Suppose $|c_n|^{1/n} \to L$.

- If $L < 1$, choose $q$ with $L < q < 1$; then $|c_n| \le q^n$ for large $n$, and the series converges absolutely.
- If $L > 1$, then $|c_n| > 1$ for large $n$, and the series diverges.
- If $L = 1$, the test is inconclusive.

The ratio test suits factorials, since $(n + 1)!/n! = n + 1$; the root test suits $n$th powers.  
For example, $\sum z^n/n!$ converges absolutely for every $z$, because $|z|/(n + 1) \to 0$;  
and $\sum\left(\dfrac{n}{2n + 1}\right)^n$ converges, because $\dfrac{n}{2n + 1} \to \dfrac12$.

The root test is the stronger one:  
if $|c_{n + 1}/c_n| \to L$, then also $|c_n|^{1/n} \to L$, but not conversely.  
For $c_n = 2^{-n + (-1)^n}$, the ratios alternate between $1/8$ and $2$, so the ratio test does not apply,  
while $|c_n|^{1/n} = 2^{-1 + (-1)^n/n} \to 1/2$, so the series converges.  
When $|c_n|^{1/n}$ has no limit, the root test still works with $L = \limsup|c_n|^{1/n}$,  
the largest limit of a subsequence of $|c_n|^{1/n}$.

### Alternating series

**Alternating series test.**  
If $b_n$ decreases to $0$, then $\sum_{n \ge 1}(-1)^{n + 1}b_n = b_1 - b_2 + b_3 - \cdots$ converges.  
The even partial sums increase, since $S_{2N + 2} - S_{2N} = b_{2N + 1} - b_{2N + 2} \ge 0$;  
the odd partial sums decrease, for the same reason;  
and $S_{2N + 1} - S_{2N} = b_{2N + 1} \to 0$.  
So both converge, to a common limit $S$, and $|S - S_N| \le b_{N + 1}$.

For example, the alternating harmonic series
$$ 1 - \frac12 + \frac13 - \frac14 + \cdots = \ln 2 $$
converges, but only conditionally, since $\sum 1/n$ diverges.  
Complex series can converge conditionally in the same way:  
the real and imaginary parts of
$$ \sum_{n = 1}^\infty\frac{i^n}{n} = \left(-\frac12 + \frac14 - \frac16 + \cdots\right) + i\left(1 - \frac13 + \frac15 - \cdots\right) $$
are alternating series, while $\sum|i^n/n| = \sum 1/n$ diverges.

### Rearrangements

Conditionally convergent series are fragile.  
Rearrange the alternating harmonic series so that each positive term is followed by two negative terms:
$$ \left(1 - \frac12\right) - \frac14 + \left(\frac13 - \frac16\right) - \frac18 + \left(\frac15 - \frac{1}{10}\right) - \frac{1}{12} + \cdots
= \frac12 - \frac14 + \frac16 - \frac18 + \frac{1}{10} - \frac{1}{12} + \cdots. $$
The right-hand side is half of the original series,  
so the rearranged series converges to $\frac12\ln 2$, not to $\ln 2$.

````python
N = 10^5
original = sum(RDF(-1)^(n + 1) / n for n in range(1, 3 * N + 1))
rearranged = sum(1/RDF(2*k - 1) - 1/RDF(4*k - 2) - 1/RDF(4*k) for k in range(1, N + 1))
print(original, RDF(log(2)))
print(rearranged, RDF(log(2) / 2))
````

**Riemann's rearrangement theorem** says more:  
a conditionally convergent real series can be rearranged to converge to any prescribed real number, or to diverge.  
Absolutely convergent series are robust:  
they can be rearranged and regrouped freely,  
and the product of two of them can be expanded and its terms collected in any order, without changing the sums.  
Power series converge absolutely inside their disk of convergence,  
so the manipulations of power series below are legitimate there.

### Summary of tests

| Test | Hypothesis | Conclusion |
|---|---|---|
| $n$th term | $c_n \not\to 0$ | diverges |
| comparison | $\lvert c_n\rvert \le b_n$, $\sum b_n$ converges | converges absolutely |
| limit comparison | $c_n, b_n > 0$, $c_n/b_n \to L \in (0, \infty)$ | same behavior as $\sum b_n$ |
| integral | $c_n = f(n)$, $f$ positive and decreasing | same behavior as $\int_1^\infty f(x)\,dx$ |
| ratio | $\lvert c_{n + 1}/c_n\rvert \to L$ | $L < 1$: converges absolutely; $L > 1$: diverges |
| root | $\lvert c_n\rvert^{1/n} \to L$ | $L < 1$: converges absolutely; $L > 1$: diverges |
| alternating | $c_n = (-1)^{n + 1}b_n$, $b_n$ decreasing to $0$ | converges |

Benchmarks: $\sum q^n$ converges if and only if $|q| < 1$, and $\sum 1/n^p$ converges if and only if $p > 1$.

## Taylor Series

Cauchy's integral formula expresses the values of a holomorphic function inside a circle through its values on the circle.  
Expanding the kernel $1/(\zeta - z)$ into a geometric series turns the formula into a power series.  
So a holomorphic function is the sum of its Taylor series on every disk where it is holomorphic.

### Power series

A **power series** centered at $a$ is a series
$$ \sum_{n = 0}^\infty a_n(z - a)^n = a_0 + a_1(z - a) + a_2(z - a)^2 + \cdots. $$
It converges at $z$ if its partial sums converge.  
For which $z$ does it converge?  
The answer is always a disk, together with some of its boundary points,  
and the geometric series explains why.

Suppose the series converges at some point $z_1 \ne a$.  
Then its terms tend to $0$ by the $n$th-term test, so $|a_n(z_1 - a)^n| \le M$ for all $n$.  
For $|z - a| < |z_1 - a|$, let $q = |z - a|/|z_1 - a| < 1$; then
$$ |a_n(z - a)^n| = |a_n(z_1 - a)^n|\,q^n \le Mq^n. $$
By comparison with the geometric series $\sum Mq^n$,  
the series converges absolutely at $z$.

Let $R \in [0, \infty]$ be the supremum of $|z_1 - a|$ over all points $z_1$ where the series converges.  
This number is the **radius of convergence**.  
If $|z - a| < R$, then $|z - a|$ is not an upper bound,  
so the series converges at some $z_1$ with $|z - a| < |z_1 - a|$,  
and by the previous cell it converges absolutely at $z$.  
If $|z - a| > R$, the series diverges at $z$ by the definition of $R$.  
On the circle $|z - a| = R$ anything can happen.

To compute $R$, apply the ratio or root test to $c_n = a_n(z - a)^n$.  
Since $|c_{n + 1}/c_n| = |z - a|\,|a_{n + 1}/a_n|$ and $|c_n|^{1/n} = |z - a|\,|a_n|^{1/n}$,
$$ R = \lim_{n \to \infty}\left|\frac{a_n}{a_{n + 1}}\right|
\qquad\text{or}\qquad
\frac1R = \lim_{n \to \infty}|a_n|^{1/n}, $$
whenever the limit exists.  
The root formula with $\limsup$ always holds; it is the **Cauchy–Hadamard formula**.

| Series | $R$ | On the boundary circle |
|---|---|---|
| $\sum_{n \ge 0} z^n$ | $1$ | diverges everywhere, by the $n$th-term test |
| $\sum_{n \ge 1} z^n/n$ | $1$ | diverges at $z = 1$, converges conditionally at $z = -1$ and $z = i$ |
| $\sum_{n \ge 1} z^n/n^2$ | $1$ | converges absolutely everywhere |
| $\sum_{n \ge 0} z^n/n!$ | $\infty$ | no boundary |
| $\sum_{n \ge 0} n!\,z^n$ | $0$ | converges only at $z = 0$ |

Inside the disk of convergence, the sum $f(z) = \sum a_n(z - a)^n$ is holomorphic,  
and it may be differentiated term by term:
$$ f'(z) = \sum_{n = 1}^\infty na_n(z - a)^{n - 1}, $$
with the same radius of convergence.  
Repeating this, $f$ has derivatives of all orders, and $f^{(k)}(a) = k!\,a_k$,  
as in the homework on Cauchy's formula for derivatives.

**Power series versus Taylor series.**  
A power series $\sum b_n(z - a)^n$ is given by its coefficients, which can be any numbers.  
A power series exists before the function.


The **Taylor series** of a function $f$ at $a$  
(so it exists after the function)  
is the particular power series whose coefficients are computed be differentiating $f$:
$$ \sum_{n = 0}^\infty\frac{f^{(n)}(a)}{n!}(z - a)^n. $$

Combining these two definitions,  
if $f$ defined by a power series, then $b_n = f^{(n)}(a)/n!$ is automatic,  
and so power $=$ Taylor.  
In particular, $f$ has at most one power series expansion about $a$.

The converse question is, when $f$ is defined elsewhere instead of a series,  
whether the Taylor series of $f$ adds up to $f$.  
We know that for real functions it may not:  
$e^{-1/x^2}$, extended by $0$ at $x = 0$, has all derivatives $0$ at $0$,  
so its Taylor series is the zero series, which converges everywhere, but to $0$ rather than to the function.  
Taylor's theorem below shows that this never happens for holomorphic functions,  
so for them "has a power series expansion" and "equals its Taylor series" mean the same thing.

### Holomorphic functions are analytic

A function is **analytic** at $a$ if it equals a convergent power series on some disk centered at $a$.  
We have just seen that analytic functions are holomorphic.  
The converse is the first big payoff of Cauchy's integral formula.

**Taylor's theorem.**  
Let $f$ be holomorphic on the disk $|z - a| < R$, where $0 < R \le \infty$.  
Then for every $z$ in this disk,
$$ \boxed{f(z) = \sum_{n = 0}^\infty\frac{f^{(n)}(a)}{n!}(z - a)^n}. $$
In particular, the radius of convergence of the Taylor series is at least $R$.

Let's prove this.  
Replacing $f(w)$ by $f(a + w)$ moves the center to $0$ without changing the derivatives there,  
so we may assume $a = 0$.  
Fix $z$ with $|z| < R$, choose $r$ with $|z| < r < R$,  
and let $C_r$ be the circle $|\zeta| = r$, traversed counterclockwise.  



Then
$$ \begin{aligned}
2\pi i\,f(z) & = \oint_{C_r}\frac{f(\zeta)}{\zeta - z}\,d\zeta \\
& = \oint_{C_r}\sum_{n = 0}^\infty\frac{f(\zeta)\,z^n}{\zeta^{n + 1}}\,d\zeta \\
& = \sum_{n = 0}^\infty\left(\oint_{C_r}\frac{f(\zeta)}{\zeta^{n + 1}}\,d\zeta\right)z^n \\
& = \sum_{n = 0}^\infty\frac{2\pi i\,f^{(n)}(0)}{n!}z^n.
\end{aligned} $$
Dividing by $2\pi i$ gives the theorem; it remains to justify the four equalities.

The first equality is Cauchy's integral formula.  
The second is the geometric series:  
for $\zeta$ on $C_r$, the ratio $z/\zeta$ has modulus $q = |z|/r < 1$, so
$$ \frac{1}{\zeta - z} = \frac1\zeta\cdot\frac{1}{1 - z/\zeta} = \sum_{n = 0}^\infty\frac{z^n}{\zeta^{n + 1}}. $$

The last is Cauchy's formula for derivatives.  
Alternatively, even if we don't know it's a derivative,  
the third line already writes $f$ as a power series.  
But if a power series exists,  
the coefficients must be $f^{(n)}(0)/n!$.

The third equality exchanges an integral with an infinite sum, which needs justification.  
After the term $n = N$, the geometric series leaves the remainder
$$ \sum_{n = N + 1}^\infty\frac{z^n}{\zeta^{n + 1}} = \frac{1}{\zeta - z}\left(\frac z\zeta\right)^{N + 1}, $$
whose modulus is at most $q^{N + 1}/(r - |z|)$, because $|\zeta - z| \ge r - |z|$ on $C_r$.

So the second line minus the partial sum of the third line up to $n = N$ is
$$ \oint_{C_r}\frac{f(\zeta)}{\zeta - z}\left(\frac z\zeta\right)^{N + 1}d\zeta. $$
If $M$ is the maximum of $|f|$ on $C_r$, the ML-inequality bounds it by
$$ 2\pi r\cdot\frac{Mq^{N + 1}}{r - |z|} \to 0\qquad(N \to \infty). $$
Hence the partial sums of the third line converge to the second line, as claimed.

The proof uses only the values of $f$ on one circle.  
This is the precise sense in which complex differentiability forces a power series,  
in contrast with the real function $e^{-1/x^2}$ above.

### The radius of convergence

Let $d$ be the distance from $a$ to the nearest point where $f$ is not holomorphic.  
Then $f$ is holomorphic on the disk $|z - a| < d$,  
so by Taylor's theorem its Taylor series about $a$ converges there.  
That is, $R \ge d$, and we learn this without computing a single coefficient.

If that nearest point is a pole $p$, then in fact $R = d$.  
Suppose instead that $R > d$.  
Then $p$ lies inside the disk of convergence,  
where the sum of the series is holomorphic, hence continuous, hence bounded near $p$.  
But the sum equals $f$ on $|z - a| < d$,  
and $|f(z)| \to \infty$ as $z \to p$ within this disk, a contradiction.

Return to the example from an earlier lecture:
$$ \frac{1}{1 + x^2} = \sum_{n = 0}^\infty(-1)^n x^{2n},\qquad |x| < 1. $$
In the complex plane, $1/(1 + z^2)$ has poles at $\pm i$, at distance $1$ from $0$,  
so the radius of convergence is $1$.  
About the center $1$ instead, the nearest poles are at distance $|1 \mp i| = \sqrt 2$,  
so the real Taylor series about $1$ converges exactly for $1 - \sqrt 2 < x < 1 + \sqrt 2$.

````python
var('z')
P = complex_plot(1/(1 + z^2), (-3, 3), (-3, 3), plot_points=300)
P += circle((0, 0), 1, color='black', thickness=2)
P += circle((1, 0), sqrt(2), color='black', thickness=2, linestyle='--')
P += point([(0, 0), (1, 0)], color='black', size=30)
P.show(aspect_ratio=1)
````

Both circles pass through the poles $\pm i$, the bright points.  
The Taylor series about $0$ converges inside the solid circle,  
and the Taylor series about $1$ converges inside the dashed circle.

````python
x = var('x')
h = 1/(1 + x^2)
plots = []
for c, xmin, xmax in [(0, -1.8, 1.8), (1, -0.8, 2.8)]:
    G = plot(h, (x, xmin, xmax), color='black', thickness=3)
    for n, color in [(10, 'red'), (20, 'green'), (40, 'blue')]:
        G += plot(taylor(h, x, c, n), (x, xmin, xmax), color=color, ymin=-0.5, ymax=1.5)
    for x0 in [c - sqrt(1 + c^2), c + sqrt(1 + c^2)]:
        G += line([(x0, -0.5), (x0, 1.5)], color='gray', linestyle='--')
    plots.append(G)
graphics_array(plots).show(figsize=[12, 4])
````

The black curve is $1/(1 + x^2)$;  
the red, green, and blue curves are its Taylor polynomials of degrees $10$, $20$, and $40$,  
about $0$ on the left and about $1$ on the right.  
Nothing special happens to $1/(1 + x^2)$ at $x = \pm 1$ or at $x = 1 \pm \sqrt 2 \approx 2.41, -0.41$,  
yet the Taylor polynomials peel away exactly there, at the dashed lines.  
**The limits of the real approximation are set by complex poles.**

More examples of reading the radius from the singularities:

- $\tan z$ is holomorphic except at the zeros $\pi/2 + k\pi$ of $\cos z$,  
  so its Taylor series about $0$ has radius $\pi/2$.
- $e^z$, $\sin z$, and $\cos z$ are entire, so their Taylor series about any point have radius $\infty$.
- $\operatorname{Log}(1 + z)$ is holomorphic on $|z| < 1$, which misses the branch cut $(-\infty, -1]$;  
  its derivative $1/(1 + z)$ has a pole at $-1$, so the radius is exactly $1$.

### Computing Taylor series

We want the Taylor series of $f$ about $a$.  
But computing every derivative $f^{(n)}(a)$ is usually hopeless.  
Instead, we can use uniqueness: any power series  
we find by other means will be as good as using derivatives.  
Consequently, this gives every derivative at once.

The known series we start from are
$$ \frac{1}{1 - z} = \sum_{n = 0}^\infty z^n\quad(|z| < 1),\qquad
e^z = \sum_{n = 0}^\infty\frac{z^n}{n!}, $$
$$ \cos z = \sum_{k = 0}^\infty\frac{(-1)^k z^{2k}}{(2k)!},\qquad
\sin z = \sum_{k = 0}^\infty\frac{(-1)^k z^{2k + 1}}{(2k + 1)!}\qquad(z \in \mathbb C). $$

**Substitution.**  
Replacing $z$ by $-z^2$ in the geometric series gives the series of $1/(1 + z^2)$ above.  
About another center, rewrite in powers of $z - a$ first:
$$ \frac{1}{3 - z} = \frac{1}{2 - (z - 1)} = \frac12\cdot\frac{1}{1 - (z - 1)/2}
= \sum_{n = 0}^\infty\frac{(z - 1)^n}{2^{n + 1}},\qquad |z - 1| < 2. $$
The radius $2$ is the distance from the center $1$ to the pole $3$.

**Integration.**  
Both $\operatorname{Log}(1 + z)$ and $\sum_{n \ge 1}(-1)^{n + 1}z^n/n$ have derivative
$$ \frac{1}{1 + z} = \sum_{n = 0}^\infty(-1)^n z^n,\qquad |z| < 1, $$
and both vanish at $z = 0$, so
$$ \operatorname{Log}(1 + z) = z - \frac{z^2}{2} + \frac{z^3}{3} - \cdots,\qquad |z| < 1. $$

**Multiplication.**  
Inside their disks of convergence, power series converge absolutely,  
so they can be multiplied like polynomials, collecting equal powers:
$$ e^z\sin z = \left(1 + z + \frac{z^2}{2} + \frac{z^3}{6} + \cdots\right)\left(z - \frac{z^3}{6} + \cdots\right)
= z + z^2 + \frac{z^3}{3} + \cdots. $$

**Division.**  
Since $\tan z$ is odd, write $\tan z = c_1 z + c_3 z^3 + c_5 z^5 + \cdots$ and match coefficients in $\sin z = \cos z\tan z$:
$$ z - \frac{z^3}{6} + \frac{z^5}{120} - \cdots
= \left(1 - \frac{z^2}{2} + \frac{z^4}{24} - \cdots\right)(c_1 z + c_3 z^3 + c_5 z^5 + \cdots). $$
This gives $c_1 = 1$, $c_3 - c_1/2 = -1/6$, and $c_5 - c_3/2 + c_1/24 = 1/120$, so
$$ \tan z = z + \frac{z^3}{3} + \frac{2z^5}{15} + \cdots,\qquad |z| < \frac{\pi}{2}. $$

````python
print(taylor(exp(z) * sin(z), z, 0, 3))
print(taylor(tan(z), z, 0, 7))
print((1/(3 - z)).series(z == 1, 4))
````

### Zeros of holomorphic functions

Let $f$ be holomorphic near $a$ with $f(a) = 0$,  
and suppose that $f$ does not vanish on a whole disk around $a$.  
Then some Taylor coefficient is nonzero (because zero series converges to zero).


Let $a_m$ be the first nonzero coefficient.  
Factoring out $(z - a)^m$ gives
$$ f(z) = (z - a)^m g(z),\qquad g(z) = a_m + a_{m + 1}(z - a) + \cdots,\qquad g(a) = a_m \ne 0. $$
The integer $m \ge 1$ is the **order** of the zero;  
equivalently, $f(a) = f'(a) = \cdots = f^{(m - 1)}(a) = 0$ and $f^{(m)}(a) \ne 0$.

This can be used to explain domain-coloring plots:  
near a zero of order $m$,  
the hue winds $m$ times around the color wheel because that's what $(z - a)^m$ does,  
while $\arg g(z)$ stays close relatively constant.


  
For example, $\sin z = z - z^3/6 + \cdots$ has a simple zero at $0$,  
and $1 - \cos z = z^2/2 - z^4/24 + \cdots$ has a zero of order $2$.

Since $g$ is continuous and $g(a) \ne 0$, also $g(z) \ne 0$ for all $z$ close to $a$.  
Hence $f(z) = (z - a)^m g(z) \ne 0$ for $0 < |z - a| < \delta$, for some $\delta > 0$.  
**Zeros of holomorphic functions are isolated**,  
unless the function vanishes on a whole disk.  
This leads to the following.

**Identity theorem.**  
Let $f$ and $g$ be holomorphic on a region $U$, that is, a connected open set.  
Here $U$ is **connected** if it cannot be split into two disjoint nonempty open sets.  
If $f = g$ on a subset of $U$ that has a limit point $p \in U$, then $f = g$ on all of $U$.

Let $h = f - g$, and let $A$ be the set of points of $U$ at which $h$ and all its derivatives vanish.

- $A$ is not empty: by continuity $h(p) = 0$, and the zeros of $h$ are not isolated at $p$,  
  so $h$ vanishes on a disk around $p$, and $p \in A$.
- $A$ is open: if $z \in A$, the Taylor series of $h$ at $z$ is $0$,  
  so by Taylor's theorem $h = 0$ on a disk around $z$, and this disk lies in $A$.
- $U \setminus A$ is open: if $h^{(n)}(z) \ne 0$, then $h^{(n)} \ne 0$ on a disk around $z$, by continuity.

Since $U$ is connected,  
and connected is defined as "not a union of two open sets",  
so $A = U$.

For example, $\sin(z)^2 + \cos(z)^2 - 1$ is entire and vanishes on the real axis, so it vanishes on $\mathbb C$.  
Every identity between holomorphic functions that holds on a real interval  
holds on every region containing that interval.  
In particular, $e^z$, $\sin z$, and $\cos z$ are the **only** holomorphic functions on $\mathbb C$  
that agree with $e^x$, $\sin x$, and $\cos x$ on the real axis.

## Laurent Series

A Taylor series cannot represent a function beyond a pole  
(a power series is bounded near its center).  
But we have negative powers.



For example, dividing the series of $e^z$ by $z^3$,
$$ \frac{e^z}{z^3} = \frac{1}{z^3} + \frac{1}{z^2} + \frac{1}{2z} + \frac16 + \frac{z}{24} + \cdots,\qquad z \ne 0. $$
In an earlier lecture, only the coefficient of $(z - a)^{-1}$ survived integration around a circle.  
Series with negative powers turn this observation into a method.

### Series in an annulus

A **Laurent series** centered at $a$ is a two-sided series
$$ \sum_{n = -\infty}^\infty a_n(z - a)^n
= \underbrace{\sum_{n = 1}^\infty\frac{a_{-n}}{(z - a)^n}}_{\text{principal part}}
+ \underbrace{\sum_{n = 0}^\infty a_n(z - a)^n}_{\text{regular part}}. $$
It converges at $z$ if both parts converge.

The regular part is a power series in $z - a$; it converges for $|z - a| < R$.  
The principal part is a power series in $w = 1/(z - a)$;  
it converges for $|w| < 1/r$, that is, for $|z - a| > r$.  



So a Laurent series converges on an **annulus**
$$ \{z : r < |z - a| < R\},\qquad 0 \le r < R \le \infty, $$
and its sum is holomorphic there.  
A punctured disk $0 < |z - a| < R$ is the case $r = 0$.

But can it happen that $r > R$?

### Laurent's theorem

**Laurent's theorem.**  
Let $f$ be holomorphic on the annulus $r < |z - a| < R$.  
Then $f$ is the sum of a Laurent series on the annulus:
$$ \boxed{f(z) = \sum_{n = -\infty}^\infty a_n(z - a)^n,\qquad
a_n = \frac{1}{2\pi i}\oint_{|\zeta - a| = \rho}\frac{f(\zeta)}{(\zeta - a)^{n + 1}}\,d\zeta}, $$
where $\rho$ is any radius with $r < \rho < R$, and the circle is traversed counterclockwise.  
The coefficients do not depend on $\rho$, by deformation of contours.

The Laurent series of $f$ on a given annulus is **unique**.  
Indeed, suppose $f(z) = \sum b_n(z - a)^n$ on the annulus.  
Divide by $(z - a)^{k + 1}$ and integrate term by term over $|z - a| = \rho$,  
as in the homework on Cauchy's formula for derivatives.  
By the model integrals, only the term with $n - k - 1 = -1$ survives:
$$ \oint_{|z - a| = \rho}\frac{f(z)}{(z - a)^{k + 1}}\,dz = 2\pi i\,b_k, $$
so $b_k = a_k$.  
Any method that produces a Laurent series on the annulus produces **the** Laurent series.

### Proof of Laurent's theorem

As in the proof of Taylor's theorem, replacing $f(w)$ by $f(a + w)$ moves the center to $0$,  
so we may assume $a = 0$.  
Fix $z$ in the annulus and choose radii $r < \rho_1 < |z| < \rho_2 < R$.  
In the figure below,  
the blue circle $|\zeta| = \rho_2$ runs counterclockwise,  
the purple circle $|\zeta| = \rho_1$ runs clockwise,  
and the small red circle around $z$ runs counterclockwise.  
The gray dashed circles bound the annulus $r < |\zeta| < R$.

````python
G = circle((0, 0), 3, color='blue', thickness=2) + circle((0, 0), 1, color='purple', thickness=2)
G += circle((0, 0), 0.6, color='gray', linestyle='--') + circle((0, 0), 3.5, color='gray', linestyle='--')
G += circle((1.2, 1.6), 0.45, color='red', thickness=2)
G += arrow((0.3, 3), (-0.3, 3), color='blue', width=2, arrowsize=4)
G += arrow((-0.25, 1), (0.25, 1), color='purple', width=2, arrowsize=4)
G += arrow((1.35, 2.05), (1.05, 2.05), color='red', width=2, arrowsize=4)
G += point([(0, 0), (1.2, 1.6)], color='black', size=30)
G += text('$0$', (0.2, -0.2), color='black', fontsize=16) + text('$z$', (1.4, 1.45), color='black', fontsize=16)
show(G, aspect_ratio=1, axes=False)
````

Then
$$ \begin{aligned}
2\pi i\,f(z) & = \oint_{\textcolor{red}{\text{red}}}\frac{f(\zeta)}{\zeta - z}\,d\zeta \\
& = \oint_{\textcolor{blue}{\text{blue}}}\frac{f(\zeta)}{\zeta - z}\,d\zeta
&& \qquad + \oint_{\textcolor{purple}{\text{purple}}}\frac{f(\zeta)}{\zeta - z}\,d\zeta \\
& = \oint_{\textcolor{blue}{\text{blue}}}\sum_{n = 0}^\infty\frac{f(\zeta)\,z^n}{\zeta^{n + 1}}\,d\zeta
&& \qquad - \oint_{\textcolor{purple}{\text{purple}}}\sum_{n = -\infty}^{-1}\frac{f(\zeta)\,z^n}{\zeta^{n + 1}}\,d\zeta \\
& = \sum_{n = 0}^\infty\left(\oint_{\textcolor{blue}{\text{blue}}}\frac{f(\zeta)}{\zeta^{n + 1}}\,d\zeta\right)z^n
&& \qquad + \sum_{n = -\infty}^{-1}\left(-\oint_{\textcolor{purple}{\text{purple}}}\frac{f(\zeta)}{\zeta^{n + 1}}\,d\zeta\right)z^n \\
& = 2\pi i\sum_{n = 0}^\infty a_n z^n
&& \qquad + 2\pi i\sum_{n = -\infty}^{-1}a_n z^n.
\end{aligned} $$
Dividing by $2\pi i$ and combining the two sums gives the theorem;  
it remains to justify the five equalities.

The first equality is Cauchy's integral formula on the red circle.  
For the second, remove the small red disk from the region between the blue and purple circles.  
On what remains, $f(\zeta)/(\zeta - z)$ is holomorphic,  
and the positively oriented boundary consists of the blue circle, the purple circle, and the red circle reversed.  
Cauchy's theorem says that these three integrals add up to $0$, which is the second equality.

The third equality expands $1/(\zeta - z)$ by the geometric series on each circle.  
On the blue circle $|z| < |\zeta|$, so, as in the proof of Taylor's theorem,
$$ \frac{1}{\zeta - z} = \frac1\zeta\cdot\frac{1}{1 - z/\zeta} = \sum_{n = 0}^\infty\frac{z^n}{\zeta^{n + 1}}. $$
On the purple circle $|\zeta| < |z|$, so we expand in powers of $\zeta/z$ instead:
$$ \frac{1}{\zeta - z} = -\frac1z\cdot\frac{1}{1 - \zeta/z} = -\sum_{m = 0}^\infty\frac{\zeta^m}{z^{m + 1}}
= -\sum_{n = -\infty}^{-1}\frac{z^n}{\zeta^{n + 1}}, $$
where $n = -m - 1$.  
The fourth equality integrates term by term;  
on both circles, the ML estimate from the proof of Taylor's theorem justifies this.

### One function, three annuli

Let
$$ f(z) = \frac{1}{(z - 1)(z - 2)} = \frac{1}{z - 2} - \frac{1}{z - 1}. $$
The poles $1$ and $2$ split the plane into three annuli centered at $0$:
$$ |z| < 1,\qquad 1 < |z| < 2,\qquad |z| > 2. $$
On each annulus, expand each partial fraction by the geometric series,  
choosing the ratio that has modulus less than $1$ there.

For $c \ne 0$, the two choices are
$$ \frac{1}{z - c} = -\frac1c\cdot\frac{1}{1 - z/c} = -\sum_{n = 0}^\infty\frac{z^n}{c^{n + 1}}\qquad(|z| < |c|), $$
$$ \frac{1}{z - c} = \frac1z\cdot\frac{1}{1 - c/z} = \sum_{n = 1}^\infty\frac{c^{n - 1}}{z^n}\qquad(|z| > |c|). $$

On $|z| < 1$, both fractions use the first expansion:
$$ f(z) = \sum_{n = 0}^\infty\left(1 - \frac{1}{2^{n + 1}}\right)z^n. $$
On $1 < |z| < 2$, use the second expansion for $c = 1$ and the first for $c = 2$:
$$ f(z) = -\sum_{n = 1}^\infty\frac{1}{z^n} - \sum_{n = 0}^\infty\frac{z^n}{2^{n + 1}}. $$
On $|z| > 2$, both fractions use the second expansion:
$$ f(z) = \sum_{n = 2}^\infty\frac{2^{n - 1} - 1}{z^n}. $$

````python
P = complex_plot(1/((z - 1) * (z - 2)), (-3, 3), (-3, 3), plot_points=300)
P += circle((0, 0), 1, color='black', thickness=2) + circle((0, 0), 2, color='black', thickness=2)
P.show(aspect_ratio=1)
````

One function has three different Laurent series about $0$, one on each annulus;  
uniqueness holds only once the annulus is fixed.  
The coefficient of $1/z$ is $0$, $-1$, and $0$ on the three annuli.  
So $\oint_{|z| = \rho}f(z)\,dz = 2\pi i\,a_{-1}$ equals $0$, $-2\pi i$, and $0$  
for $\rho < 1$, for $1 < \rho < 2$, and for $\rho > 2$.

### Expansions about a singular point

Most often we expand on a punctured disk $0 < |z - a| < R$ around one singular point $a$.  
Known Taylor series do the work:
$$ \frac{\sin z}{z^4} = \frac{1}{z^3} - \frac{1}{6z} + \frac{z}{120} - \cdots,\qquad
e^{1/z} = 1 + \frac1z + \frac{1}{2!\,z^2} + \frac{1}{3!\,z^3} + \cdots, $$
both valid for all $z \ne 0$.

About a center $a \ne 0$, expand in powers of $w = z - a$.  
For $f(z) = \dfrac{1}{(z - 1)(z - 2)}$ near $a = 1$,
$$ f(z) = \frac{1}{w(w - 1)} = -\frac1w\cdot\frac{1}{1 - w} = -\frac1w - 1 - w - w^2 - \cdots,\qquad 0 < |w| < 1. $$
So $f(z) = -\dfrac{1}{z - 1} - \displaystyle\sum_{n = 0}^\infty(z - 1)^n$ for $0 < |z - 1| < 1$.  
The outer radius $1$ is the distance from $1$ to the other pole $2$.

In Sage, `series` computes Laurent expansions about a point:

## Isolated Singularities

A point $a$ is an **isolated singularity** of $f$ if $f$ is holomorphic on a punctured disk $0 < |z - a| < R$,  
but not at $a$, where $f$ may not even be defined.  
By Laurent's theorem $f$ has a Laurent series on the punctured disk,  
and its principal part sorts isolated singularities into three kinds.

### Three kinds of singularities

| Kind | Principal part | Example at $0$ |
|---|---|---|
| **removable** | $0$ | $\dfrac{\sin z}{z} = 1 - \dfrac{z^2}{6} + \cdots$ |
| **pole of order $m$** | ends at $a_{-m}(z - a)^{-m}$, $a_{-m} \ne 0$ | $\dfrac{\sin z}{z^4} = \dfrac{1}{z^3} - \dfrac{1}{6z} + \cdots$ |
| **essential** | infinitely many terms | $e^{1/z} = 1 + \dfrac1z + \dfrac{1}{2z^2} + \cdots$ |

For a pole of order $m$, the function $g(z) = (z - a)^m f(z) = a_{-m} + a_{-m + 1}(z - a) + \cdots$  
is holomorphic at $a$ with $g(a) = a_{-m} \ne 0$.  
So $f(z) = g(z)/(z - a)^m$, as in the definition of a pole of order $m$ from an earlier lecture.

````python
plots = [complex_plot(f, (-1, 1), (-1, 1), plot_points=300) for f in [sin(z)/z, sin(z)/z^4, exp(1/z)]]
graphics_array(plots).show(figsize=[12, 4])
````

From left to right:  
$\sin(z)/z$ shows nothing special at $0$;  
$\sin(z)/z^4$ has a white pole around which the hue winds $3$ times in the reverse direction;  
$e^{1/z}$ shows every hue and every brightness in every neighborhood of $0$.  
Each picture is typical of its kind, as the next three sections explain.

### Removable singularities

If $f$ has a removable singularity at $a$, then setting $f(a) = a_0$ makes $f$ holomorphic on the whole disk.  
For example, $\sin(z)/z$ becomes the entire function $\sum_{k \ge 0}(-1)^k z^{2k}/(2k + 1)!$, with value $1$ at $0$.  
How can we recognize a removable singularity without computing the Laurent series?

**Riemann's removable singularity theorem.**  
If $f$ is holomorphic and bounded on $0 < |z - a| < R$, then $a$ is a removable singularity.  
To see this, let $|f| \le M$.  
For $k \ge 1$ and $0 < \rho < R$, the ML-inequality on the circle $|z - a| = \rho$ gives
$$ |a_{-k}| = \left|\frac{1}{2\pi i}\oint_{|z - a| = \rho}f(z)(z - a)^{k - 1}\,dz\right|
\le \frac{1}{2\pi}\cdot 2\pi\rho\cdot M\rho^{k - 1} = M\rho^k. $$
Letting $\rho \to 0$ gives $a_{-k} = 0$, so the principal part vanishes.

This is the argument for bounded integrands from an earlier lecture,  
applied to every negative coefficient at once.  
For example, $(e^z - 1)/z \to (e^z)'|_{z = 0} = 1$ as $z \to 0$,  
so $z/(e^z - 1)$ is bounded near $0$ and has a removable singularity there.

### Poles

If $f$ has a pole of order $m$ at $a$, then $f(z) = g(z)/(z - a)^m$ with $g(a) \ne 0$, so
$$ |f(z)| \approx \frac{|g(a)|}{|z - a|^m} \to \infty\qquad(z \to a). $$
Conversely, if $|f(z)| \to \infty$ as $z \to a$, then $1/f$ is bounded near $a$.  
By Riemann's theorem $1/f$ has a removable singularity, where its value is $0$;  
if this zero has order $m$, then $f$ has a pole of order $m$.  
So **poles are exactly the isolated singularities where $|f| \to \infty$**, the white points in domain coloring.

Poles usually come from zeros of denominators.  
Let $g$ and $h$ be holomorphic at $a$, with zeros of orders $k$ and $m$ at $a$, where order $0$ means nonzero.  
Then $g = (z - a)^k g_1$ and $h = (z - a)^m h_1$ with $g_1(a) \ne 0$ and $h_1(a) \ne 0$, so
$$ \frac{g(z)}{h(z)} = (z - a)^{k - m}\frac{g_1(z)}{h_1(z)}. $$
Hence $g/h$ has a pole of order $m - k$ at $a$ if $m > k$, and a removable singularity if $m \le k$.

For example:

- $\tan z = \sin z/\cos z$ has simple poles at $\pi/2 + k\pi$, the simple zeros of $\cos z$;
- $(1 - \cos z)/z^4$ has a pole of order $4 - 2 = 2$ at $0$;
- $z/\sin z$ has a removable singularity at $0$ and simple poles at $k\pi$, $k \ne 0$;
- $1/(z^2\sin z)$ has a pole of order $3$ at $0$.

A function whose only singularities in $U$ are poles is called **meromorphic** on $U$.  
Rational functions and $\tan z$ are meromorphic on $\mathbb C$.

### Essential singularities

Near an essential singularity, $f$ neither stays bounded nor tends to $\infty$.  
For $e^{1/z}$ near $0$:

- along the positive real axis, $e^{1/x} \to \infty$;
- along the negative real axis, $e^{1/x} \to 0$; and
- along the imaginary axis, $|e^{1/(iy)}| = |e^{-i/y}| = 1$.

In fact $e^{1/z}$ takes **every** nonzero value $w$ in every punctured disk $0 < |z| < \delta$.  
Indeed, $e^{1/z} = w$ means $1/z = \operatorname{Log}w + 2\pi ik$ for some $k \in \mathbb Z$, that is,
$$ z = z_k = \frac{1}{\operatorname{Log}w + 2\pi ik},\qquad |z_k| \to 0\quad(|k| \to \infty). $$
Infinitely many of the solutions $z_k$ lie in the punctured disk.

**Casorati–Weierstrass theorem.**  
If $a$ is an essential singularity of $f$,  
then in every punctured disk around $a$, $f$ comes arbitrarily close to every complex number.  
The homework due Friday proves it.  
This is why a domain-coloring plot near an essential singularity contains every hue and every brightness.  
Picard's great theorem says even more: $f$ takes every complex value, with at most one exception,  
infinitely often near $a$.

## Residues

### Definition

Let $a$ be an isolated singularity of $f$, with Laurent series $\sum a_n(z - a)^n$ on $0 < |z - a| < R$.  
The **residue** of $f$ at $a$ is the coefficient
$$ \operatorname{Res}(f, a) = a_{-1}. $$
By Laurent's coefficient formula with $n = -1$, for every $0 < \rho < R$,
$$ \boxed{\oint_{|z - a| = \rho}f(z)\,dz = 2\pi i\operatorname{Res}(f, a)}. $$
Integrating the Laurent series term by term gives the same result:  
every power other than $(z - a)^{-1}$ integrates to $0$.

This agrees with the residue $g(a)$ at a simple pole from an earlier lecture,  
and extends it to poles of every order and to essential singularities.  
At a removable singularity the residue is $0$.  
From the expansions above,
$$ \operatorname{Res}\left(\frac{e^z}{z^3}, 0\right) = \frac12,\qquad
\operatorname{Res}\left(\frac{\sin z}{z^4}, 0\right) = -\frac16,\qquad
\operatorname{Res}\left(e^{1/z}, 0\right) = 1. $$

### Computing residues

#### Simple poles

At a simple pole, $f(z) = a_{-1}/(z - a) + a_0 + a_1(z - a) + \cdots$, so
$$ \operatorname{Res}(f, a) = \lim_{z \to a}(z - a)f(z). $$
For example,
$$ \operatorname{Res}\left(\frac{e^z}{z^2 + 1}, i\right) = \lim_{z \to i}\frac{e^z}{z + i} = \frac{e^i}{2i}. $$

Often $f = g/h$, where $g$ is holomorphic at $a$ and $h$ has a simple zero at $a$, that is, $h(a) = 0 \ne h'(a)$.  
Then
$$ \operatorname{Res}\left(\frac gh, a\right) = \lim_{z \to a}\frac{g(z)}{\bigl(h(z) - h(a)\bigr)/(z - a)} = \frac{g(a)}{h'(a)}. $$
For example,
$$ \operatorname{Res}\left(\tan z, \frac{\pi}{2}\right) = \frac{\sin(\pi/2)}{-\sin(\pi/2)} = -1,\qquad
\operatorname{Res}\left(\frac{1}{z^4 + 1}, a\right) = \frac{1}{4a^3} = -\frac a4\quad\text{if }a^4 = -1. $$

#### Poles of higher order

At a pole of order $m$, multiply by $(z - a)^m$:
$$ (z - a)^m f(z) = a_{-m} + \cdots + a_{-1}(z - a)^{m - 1} + a_0(z - a)^m + \cdots. $$
The residue is the Taylor coefficient of $(z - a)^{m - 1}$ of this holomorphic function, so
$$ \boxed{\operatorname{Res}(f, a) = \frac{1}{(m - 1)!}\lim_{z \to a}\frac{d^{m - 1}}{dz^{m - 1}}\Bigl((z - a)^m f(z)\Bigr)}. $$
This is Cauchy's formula for derivatives in disguise:  
if $f(z) = g(z)/(z - a)^m$, the residue is $g^{(m - 1)}(a)/(m - 1)!$.

For example, $e^z/(z - 1)^3$ has a pole of order $3$ at $1$, and
$$ \operatorname{Res}\left(\frac{e^z}{(z - 1)^3}, 1\right) = \frac{1}{2!}\left.\frac{d^2}{dz^2}e^z\right|_{z = 1} = \frac e2. $$
The function $1/\bigl(z^2(z - 1)\bigr)$ has a pole of order $2$ at $0$, and
$$ \operatorname{Res}\left(\frac{1}{z^2(z - 1)}, 0\right) = \left.\frac{d}{dz}\frac{1}{z - 1}\right|_{z = 0} = -\frac{1}{(0 - 1)^2} = -1. $$

#### Using the Laurent series

When the order is high, or the singularity is essential, expand instead.  
For $1/(z^2\sin z)$, which has a pole of order $3$ at $0$,
$$ \frac{1}{z^2\sin z} = \frac{1}{z^3}\cdot\frac{1}{1 - z^2/6 + z^4/120 - \cdots}
= \frac{1}{z^3}\left(1 + \frac{z^2}{6} + \cdots\right), $$
so $\operatorname{Res}\bigl(1/(z^2\sin z), 0\bigr) = 1/6$.  
For $z^3e^{1/z}$, which has an essential singularity at $0$, the only term with $z^{-1}$ is  
$z^3\cdot\dfrac{1}{4!\,z^4}$, so $\operatorname{Res}(z^3e^{1/z}, 0) = 1/24$.

In Sage, read the residue off the series, or use `residue` at a pole.  
The command `residue` expands in powers of $z - a$, which fails at an essential singularity:  
for $e^{1/z}$ at $0$ it returns $0$ instead of $1$.

````python
print((1/(z^2 * sin(z))).series(z, 2))
print((exp(z)/(z - 1)^3).residue(z == 1))
print(tan(z).residue(z == pi/2))
````

### The residue theorem

**Residue theorem.**  
Let $C$ be a positively oriented simple closed curve,  
and let $f$ be holomorphic on and inside $C$ except at finitely many points $a_1, \ldots, a_k$ inside $C$.  
Then
$$ \boxed{\oint_C f(z)\,dz = 2\pi i\sum_{j = 1}^k\operatorname{Res}(f, a_j)}. $$

Deform $C$ into small disjoint circles $C_j$ around the points $a_j$, as in an earlier lecture.  
Cauchy's theorem on the region inside $C$ and outside the small circles gives
$$ \oint_C f(z)\,dz = \sum_{j = 1}^k\oint_{C_j}f(z)\,dz, $$
and each integral on the right is $2\pi i\operatorname{Res}(f, a_j)$ by the definition of the residue.

````python
var('t')
outer = parametric_plot((2 * cos(t) + cos(3 * t) / 2, 3 * sin(t) / 2), (t, 0, 2 * pi),
                        color='blue', thickness=2)
inside = [(-1.2, 0.4), (0.3, -0.6), (1.4, 0.5)]
small = sum(circle(p, 0.3, color='red', thickness=2) for p in inside)
dots = point(inside, size=30, color='black') + point((2.6, 1.4), size=30, color='gray')
show(outer + small + dots, aspect_ratio=1, axes=False)
````

Only the black singularities inside the blue curve $C$ contribute; the gray one outside does not.

Cauchy's theorem, Cauchy's integral formula, and Cauchy's formula for derivatives are special cases:  
no singularities, $g(z)/(z - a)$, and $g(z)/(z - a)^{k + 1}$.  
The residue theorem also handles essential singularities, where no integral formula applies.  
Computing a contour integral comes down to two tasks:  
locate the singularities inside $C$, and find a Laurent coefficient at each of them.

### Examples

**Two simple poles.**  
Both poles $\pm i$ of $e^z/(z^2 + 1)$ lie inside $|z| = 2$, so
$$ \oint_{|z| = 2}\frac{e^z}{z^2 + 1}\,dz = 2\pi i\left(\frac{e^i}{2i} + \frac{e^{-i}}{-2i}\right) = 2\pi i\sin 1. $$
This is the value found with partial fractions in an earlier lecture.

**The poles of $\tan z$.**  
Inside $|z| = 2$, $\tan z$ has the poles $\pm\pi/2 \approx \pm 1.57$, but not $\pm 3\pi/2 \approx \pm 4.71$.  
By the $g/h$ rule, each pole $a$ has residue $\sin(a)/(-\sin a) = -1$, so
$$ \oint_{|z| = 2}\tan z\,dz = 2\pi i(-1 - 1) = -4\pi i. $$

**A pole of order $3$.**  
Of the singularities $0, \pm\pi, \pm 2\pi, \ldots$ of $1/(z^2\sin z)$, only $0$ lies inside $|z| = 1$, so
$$ \oint_{|z| = 1}\frac{dz}{z^2\sin z} = 2\pi i\cdot\frac16 = \frac{\pi i}{3}. $$

**Essential singularities.**  
From the residues computed above,
$$ \oint_{|z| = 1}e^{1/z}\,dz = 2\pi i,\qquad \oint_{|z| = 1}z^3e^{1/z}\,dz = \frac{2\pi i}{24} = \frac{\pi i}{12}. $$
Cauchy's integral formula cannot compute these:  
$e^{1/z}$ is not of the form $g(z)/(z - a)^m$ with $g$ holomorphic at $a$.

### Numerical check

As in an earlier lecture, approximate $\oint_{|z - a| = r}f(z)\,dz$, counterclockwise, by the trapezoid rule.  
Each line below prints the numerical integral followed by the value from the residue theorem.

````python
def tidy(w, tol=1e-9):
    x0, y0 = w.real(), w.imag()
    return CDF(0 if abs(x0) < tol else x0, 0 if abs(y0) < tol else y0)
````

````python
def circle_integral(f, a=0, r=1, n=2000):
    """Integral of f over |z - a| = r, counterclockwise (trapezoid rule)."""
    h = 2 * RDF(pi) / n
    total = CDF(0)
    for k in range(n):
        w = CDF(cos(k * h), sin(k * h))
        total += f(a + r * w) * (CDF(0, 1) * r * w) * h
    return tidy(total)
````

### Summary

| Singularity at $a$ | Principal part | Behavior near $a$ | $\operatorname{Res}(f, a)$ |
|---|---|---|---|
| removable | $0$ | bounded | $0$ |
| simple pole | $a_{-1}/(z - a)$ | $\lvert f\rvert \to \infty$ | $\lim_{z \to a}(z - a)f(z)$, or $g(a)/h'(a)$ for $f = g/h$ |
| pole of order $m$ | ends at $a_{-m}(z - a)^{-m}$ | $\lvert f\rvert \to \infty$ | $\dfrac{1}{(m - 1)!}\lim_{z \to a}\dfrac{d^{m - 1}}{dz^{m - 1}}\bigl((z - a)^m f(z)\bigr)$ |
| essential | infinitely many terms | comes close to every value | $a_{-1}$ from the Laurent series |

For a positively oriented simple closed curve $C$,  
$\oint_C f(z)\,dz$ is $2\pi i$ times the sum of the residues at the singularities inside $C$.

````python
def winding_number(gamma, a, n=2000):
    """Winding number of the closed curve gamma(t), 0 <= t <= 2 pi, around a."""
    g = fast_callable(gamma, vars=[t], domain=CDF)
    dg = fast_callable(diff(gamma, t), vars=[t], domain=CDF)
    h = 2 * RDF(pi) / n
    total = sum(dg(k * h) / (g(k * h) - a) for k in range(n)) * h
    return tidy(total / CDF(0, 2 * RDF(pi)))
````
