## Exponential, Logarithm, and Trigonometric Functions

### The exponential function

Euler's formula was introduced for real angles: $e^{i\theta} = \cos\theta + i\sin\theta$.  
We now define $e^z$ for every $z = x + iy$ so that

- $e^z$ agrees with the real exponential when $y = 0$;
- the law of exponents $e^{z + w} = e^z e^w$ holds; and
- Euler's formula holds on the imaginary axis.

These requirements leave only one choice:
$$ e^z = e^x e^{iy} = e^x(\cos y + i\sin y). $$

Reading off the polar form,
$$ |e^z| = e^x > 0,\qquad \arg e^z = \{y + 2\pi k : k \in \mathbb Z\}. $$
In particular, $e^z$ is never zero.  
The law of exponents says that lengths multiply and angles add:  
for $z = x_1 + iy_1$ and $w = x_2 + iy_2$,
$$ e^z e^w = e^{x_1 + x_2}e^{i(y_1 + y_2)} = e^{z + w}. $$

Write $e^z = u + iv$ with $u = e^x\cos y$ and $v = e^x\sin y$.  
Then
$$ u_x = e^x\cos y = v_y,\qquad u_y = -e^x\sin y = -v_x. $$
The Cauchy–Riemann equations hold at every point,  
and the partial derivatives are continuous.  

So $e^z$ is holomorphic on all of $\mathbb C$, that is, **entire**, and
$$ (e^z)' = u_x + iv_x = e^x\cos y + ie^x\sin y = e^z. $$

Since $e^{iy}$ repeats every $2\pi$,
$$ e^{z + 2\pi i} = e^z. $$
More precisely, $e^z = e^w$ if and only if $z - w \in 2\pi i\mathbb Z$.  
So $e^z$ is periodic with period $2\pi i$, and it is not injective on $\mathbb C$.  
The strip $\{z : -\pi < \operatorname{Im}z \le \pi\}$ is mapped one-to-one onto $\mathbb C \setminus \{0\}$.

The domain-coloring plot of $e^z$ from the previous lecture shows exactly this.  
The hue depends only on $y$ and repeats every $2\pi$ vertically,  
while the brightness depends only on $x$.  
Horizontal lines are sent to rays from the origin,  
and vertical segments of length $2\pi$ are wrapped once around circles.

The Taylor series of the real exponential converges for every complex $z$ and sums to $e^z$:
$$ e^z = \sum_{n = 0}^\infty \frac{z^n}{n!}. $$
Setting $z = iy$ and separating even and odd powers,
$$ e^{iy} = \sum_{k = 0}^\infty \frac{(-1)^k y^{2k}}{(2k)!}
+ i\sum_{k = 0}^\infty \frac{(-1)^k y^{2k + 1}}{(2k + 1)!}
= \cos y + i\sin y. $$
This justifies Euler's formula.

### Trigonometric and hyperbolic functions

Euler's formula gives $e^{i\theta} = \cos\theta + i\sin\theta$ and $e^{-i\theta} = \cos\theta - i\sin\theta$.  
Adding and subtracting,
$$ \cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2},\qquad
\sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i}. $$
The right-hand sides make sense for every complex number, so we **define**
$$ \cos z = \frac{e^{iz} + e^{-iz}}{2},\qquad
\sin z = \frac{e^{iz} - e^{-iz}}{2i},\qquad z \in \mathbb C. $$

Both functions are built from $e^{iz}$ and $e^{-iz}$, so both are entire.  
The chain rule gives the familiar derivatives
$$ (\sin z)' = \cos z,\qquad (\cos z)' = -\sin z. $$
Both are $2\pi$-periodic, $\sin$ is odd, and $\cos$ is even.  
Expanding the squares of the exponential formulas gives
$$ \sin(z)^2 + \cos(z)^2 = 1. $$

The addition formulas follow from the same algebra:
$$ \sin(z + w) = \sin(z)\cos(w) + \cos(z)\sin(w), $$
$$ \cos(z + w) = \cos(z)\cos(w) - \sin(z)\sin(w). $$
Every trigonometric identity that can be derived from the exponential formulas  
holds for all complex arguments.

For example, apply the addition formula for the sine to $\pi/2$ and $-z$:
$$ \sin(\pi/2 - z) = \sin(\pi/2)\cos(-z) + \cos(\pi/2)\sin(-z) = \cos z, $$
because $\sin(\pi/2) = 1$, $\cos(\pi/2) = 0$, and $\cos$ is even.  
So $\cos z = \sin(\pi/2 - z)$ for every complex $z$, and in the same way $\sin z = \cos(\pi/2 - z)$.

The triple-angle formulas come out of Euler's formula.  
For real $\theta$, the law of exponents gives
$$ \cos(3\theta) + i\sin(3\theta) = e^{3i\theta} = \bigl(e^{i\theta}\bigr)^3 = (\cos\theta + i\sin\theta)^3, $$
which is de Moivre's formula with $n = 3$.

Expand the cube with the binomial theorem, using $i^2 = -1$ and $i^3 = -i$:
$$ (\cos\theta + i\sin\theta)^3
= \cos(\theta)^3 + 3i\cos(\theta)^2 \sin\theta - 3\cos(\theta)\sin(\theta)^2 - i\sin(\theta)^3. $$

Since $\cos\theta$ and $\sin\theta$ are real, we may compare real and imaginary parts:
$$ \cos(3\theta) = \cos(\theta)^3 - 3\cos(\theta)\sin(\theta)^2,\qquad
\sin(3\theta) = 3\cos(\theta)^2 \sin\theta - \sin(\theta)^3. $$

Substituting $\sin(\theta)^2 = 1 - \cos(\theta)^2$ in the first formula  
and $\cos(\theta)^2 = 1 - \sin(\theta)^2$ in the second gives the **triple-angle formulas**
$$ \boxed{\cos(3\theta) = 4\cos(\theta)^3 - 3\cos\theta,\qquad \sin(3\theta) = 3\sin\theta - 4\sin(\theta)^3}. $$

The addition formulas give the same result with more work, by expanding $\cos(2\theta + \theta)$ and $\sin(2\theta + \theta)$.  
Both formulas also hold for complex arguments.  
For instance, with $u = e^{iz}$,
$$ 4\cos(z)^3 - 3\cos z = \frac{(u + u^{-1})^3}{2} - \frac{3(u + u^{-1})}{2} = \frac{u^3 + u^{-3}}{2} = \cos(3z). $$

The hyperbolic functions are
$$ \cosh z = \frac{e^z + e^{-z}}{2},\qquad \sinh z = \frac{e^z - e^{-z}}{2}. $$
Comparing with the definitions above,
$$ \cos(iz) = \cosh z,\qquad \sin(iz) = i\sinh z. $$
So the trigonometric and hyperbolic functions form one family, viewed along rotated axes.

Use the addition formulas together with $\cos(iy) = \cosh y$ and $\sin(iy) = i\sinh y$:
$$ \sin(x + iy) = \sin(x)\cosh(y) + i\cos(x)\sinh(y), $$
$$ \cos(x + iy) = \cos(x)\cosh(y) - i\sin(x)\sinh(y). $$
Taking squared moduli and using $\cosh(y)^2 = 1 + \sinh(y)^2$,
$$ |\sin(x + iy)|^2 = \sin(x)^2 + \sinh(y)^2,\qquad
|\cos(x + iy)|^2 = \cos(x)^2 + \sinh(y)^2. $$

Two consequences follow at once.

- **Zeros.** If $\sin z = 0$, then $\sin(x)^2 + \sinh(y)^2 = 0$, so $y = 0$ and $\sin x = 0$.  
  Hence the zeros of $\sin z$ are exactly the real numbers $k\pi$, $k \in \mathbb Z$.  
  Similarly, the zeros of $\cos z$ are exactly $\pi/2 + k\pi$.
- **Unboundedness.** $|\sin z| \ge |\sinh y|$, which tends to $\infty$ as $|y| \to \infty$.  
  So $\sin z$ and $\cos z$ are **not bounded** on $\mathbb C$,  
  even though $|\sin x| \le 1$ on the real axis.

The tangent is $\tan z = \sin z/\cos z$.  
It is holomorphic except at the zeros $z = \pi/2 + k\pi$ of the cosine, where it has poles.  
These are the bright points on the real axis  
in the domain-coloring plot of $\tan z$ from the previous lecture,  
and the dark points of that plot are the zeros $z = k\pi$ of the sine.

### The logarithm

Given $z \ne 0$, we look for all $w$ with $e^w = z$.  
Write $z = re^{i\theta}$ and $w = s + it$.  
Then $e^w = e^s e^{it}$, so the equation $e^w = z$ means
$$ e^s = r,\qquad t = \theta + 2\pi k,\quad k \in \mathbb Z. $$
Because $e^z$ is periodic, every nonzero $z$ has infinitely many solutions $w$,  
which differ by integer multiples of $2\pi i$.

These solutions are the **logarithms** of $z$:
$$ \log z = \ln|z| + i\arg z = \{\ln|z| + i(\operatorname{Arg}z + 2\pi k) : k \in \mathbb Z\}. $$
The **principal logarithm** uses the principal argument $\operatorname{Arg}z \in (-\pi, \pi]$:
$$ \operatorname{Log}z = \ln|z| + i\operatorname{Arg}z. $$
Here $\ln$ is the real natural logarithm of a positive number.  
For example, $\operatorname{Log}1 = 0$, $\operatorname{Log}(-1) = i\pi$, and $\operatorname{Log}(1 + i) = \frac12\ln2 + \frac{i\pi}{4}$.

Since $e^{\operatorname{Log}z} = z$, the principal logarithm undoes the exponential in one direction.  
In the other direction,
$$ \operatorname{Log}(e^z) = z \quad\text{only if}\quad -\pi < \operatorname{Im}z \le \pi. $$
For instance, $\operatorname{Log}(e^{2\pi i}) = \operatorname{Log}1 = 0 \ne 2\pi i$.

The principal argument jumps by $2\pi$ across the negative real axis,  
and so does $\operatorname{Im}\operatorname{Log}z$.  
Hence $\operatorname{Log}$ is continuous on $\mathbb C \setminus (-\infty, 0]$  
but not at points of the negative real axis,  
which we call the **branch cut**.  
Cutting along a different ray from the origin gives a different **branch** of the logarithm.

On $\mathbb C \setminus (-\infty, 0]$, write $\operatorname{Log}z = u + iv$ with $u = \frac12\ln(x^2 + y^2)$ and $v = \operatorname{Arg}z$.  
For $x > 0$ we have $v = \arctan(y/x)$, and
$$ u_x = \frac{x}{x^2 + y^2} = v_y,\qquad u_y = \frac{y}{x^2 + y^2} = -v_x. $$
The same formulas hold at every point of the slit plane, so $\operatorname{Log}$ is holomorphic there, with
$$ (\operatorname{Log}z)' = u_x + iv_x = \frac{x - iy}{x^2 + y^2} = \frac1z. $$

The cut cannot be avoided.  
Let $z$ travel once counterclockwise around the origin.  
The modulus returns to its starting value, but the argument increases by $2\pi$,  
so a logarithm followed continuously along the path gains $2\pi i$.  
No continuous branch of $\log z$ exists on a region that encircles the origin.

```python
var('z')
complex_plot(log(z), (-5, 5), (-5, 5), plot_points=500)
```

The plot of $\operatorname{Log}z$ has one zero, at $z = 1$, where $\operatorname{Log}1 = 0$.  
Along the negative real axis the colors jump:  
the two sides of the cut carry the values $\ln|x| + i\pi$ and $\ln|x| - i\pi$.  
Near the origin $|\operatorname{Log}z| \to \infty$, but only logarithmically slowly.

## Integration

### Step 1: $f : \mathbb R \to \mathbb R$ on an interval

For an integrable function $f : \mathbb R \to \mathbb R$,  
the definite integral over $[a, b]$ is the limit of signed rectangular sums:
$$ \int_a^b f(x)\,dx
= \lim_{\|\mathcal P\| \to 0}
\sum_j f(x_j^*)\,\Delta x_j. $$

The differential $dx$ records the oriented displacement along the real line.  
Consequently,
$$ \int_b^a f(x)\,dx = -\int_a^b f(x)\,dx. $$

### Step 2: $g : \mathbb R^2 \to \mathbb R^2$ along a curve $\gamma$

Let
$$ \gamma(t) = (x(t), y(t)),\qquad a \le t \le b, $$
be a piecewise continuously differentiable curve,  
and let $g = (P, Q)$ be a vector field. 

The line integral of $g$ along $\gamma$ is
$$ \int_\gamma g \mathbin{\boldsymbol{\cdot}} d\boldsymbol r
= \int_a^b g(\gamma(t)) \mathbin{\boldsymbol{\cdot}} \gamma'(t)\,dt. $$
It measures the tangential accumulation of the field along the curve.

#### Riemann-sum form

For a partition
$$ a = t_0 < t_1 < \cdots < t_n = b $$
and sample points $t_j^* \in [t_{j - 1}, t_j]$, the line integral is
$$ \int_\gamma g \mathbin{\boldsymbol{\cdot}} d\boldsymbol r
= \lim_{\|\mathcal P\| \to 0}
\sum_{j = 1}^n g(\gamma(t_j^*)) \mathbin{\boldsymbol{\cdot}}
\left(\gamma(t_j) - \gamma(t_{j - 1})\right). $$
Each term pairs the field sampled on one small arc  
with that arc's displacement vector.  
Since $\Delta\boldsymbol r_j \approx \gamma'(t_j^*)\Delta t_j$,  
the limit gives the parametrized integral in the next cell.

In coordinates,
$$ \int_\gamma g \mathbin{\boldsymbol{\cdot}} d\boldsymbol r
= \int_\gamma\left(P\,dx + Q\,dy\right)
= \int_a^b
\left(P(x(t), y(t))x'(t) + Q(x(t), y(t))y'(t)\right)dt. $$
An orientation-preserving reparametrization does not change the integral.  
Reversing the direction of $\gamma$ changes its sign.

### Step 3: $h : \mathbb C \to \mathbb C$ along a curve $\gamma$

Let $\gamma : [a, b] \to \mathbb C$ be piecewise continuously differentiable.  
For a continuous function $h : \mathbb C \to \mathbb C$, define
$$ \int_\gamma h(z)\,dz
= \int_a^b h(\gamma(t))\gamma'(t)\,dt. $$
This is formally the same parametrized integral as before,  
but the product is now complex multiplication.  
It is unchanged by orientation-preserving reparametrization  
and changes sign when the curve is reversed.

#### Riemann-sum form

For a partition
$$ a = t_0 < t_1 < \cdots < t_n = b $$
and sample points $t_j^* \in [t_{j - 1}, t_j]$, write $z_j = \gamma(t_j)$ and $z_j^* = \gamma(t_j^*)$.  
Then
$$ \int_\gamma h(z)\,dz
= \lim_{\|\mathcal P\| \to 0}
\sum_{j = 1}^n h(z_j^*)\,(z_j - z_{j - 1}). $$

Each term multiplies the value of $h$ sampled on one small arc  
by that arc's complex displacement $\Delta z_j = z_j - z_{j - 1}$.  
Because the product is complex multiplication,  
$h(z_j^*)$ rotates and scales $\Delta z_j$.  
Since $\Delta z_j \approx \gamma'(t_j^*)\Delta t_j$,  
the limit gives the parametrized integral above.

Write
$$ h(z) = u(x, y) + iv(x, y),\qquad dz = dx + i\,dy. $$
Then
$$ h(z)\,dz
= (u\,dx - v\,dy) + i(v\,dx + u\,dy), $$
so
$$ \int_\gamma h(z)\,dz
= \int_\gamma\left(u\,dx - v\,dy\right)
+ i\int_\gamma\left(v\,dx + u\,dy\right). $$
A complex line integral is therefore two real vector-field line integrals  
carried by the same oriented curve.

### The three constructions

| Integrand | Domain of integration | Definition after parametrization |
|---|---|---|
| $f : \mathbb R \to \mathbb R$ | Interval $[a, b]$ | $\displaystyle\int_a^b f(x)\,dx$ |
| $g : \mathbb R^2 \to \mathbb R^2$ | Curve $\gamma$ in $\mathbb R^2$ | $\displaystyle\int_a^b g(\gamma(t)) \mathbin{\boldsymbol{\cdot}} \gamma'(t)\,dt$ |
| $h : \mathbb C \to \mathbb C$ | Curve $\gamma$ in $\mathbb C$ | $\displaystyle\int_a^b h(\gamma(t))\gamma'(t)\,dt$ |

Green's theorem relates the second type, taken around a closed curve,  
to an ordinary double integral over the enclosed planar region.

## Green's Theorem and the Shoelace Formula

### Green's theorem in the plane

Let $D \subset \mathbb R^2$ be a bounded region whose boundary $C = \partial D$  
consists of finitely many positively oriented, piecewise smooth, simple closed curves.  
Let $P$ and $Q$ have continuous first partial derivatives  
on an open set containing $D$ and its boundary.

Positive orientation means that $D$ stays on the left as we traverse $C$.  
Thus an outer boundary runs counterclockwise; boundaries around holes run clockwise.

### Circulation form

Green's theorem converts a line integral around the boundary  
into a double integral over the region:
$$ \oint_C\left(P\,dx + Q\,dy\right)
= \iint_D\left(\frac{\partial Q}{\partial x}
- \frac{\partial P}{\partial y}\right)\,dA. $$
The integrand on the right is the scalar curl of the planar vector field $(P, Q)$.  
The theorem says that total microscopic rotation inside $D$  
equals macroscopic circulation around $\partial D$.

### Flux form

For a counterclockwise parametrization, the outward normal satisfies  
$\boldsymbol n\,ds = (dy, -dx)$.  
Therefore Green's theorem also gives
$$ \oint_C\left(P\,dy - Q\,dx\right)
= \iint_D\left(\frac{\partial P}{\partial x}
+ \frac{\partial Q}{\partial y}\right)\,dA. $$
The integrand is the divergence of $(P, Q)$,  
so this form equates outward boundary flux with total interior source strength.

### Green's theorem and complex line integrals

Let $h = u + iv$ and let $C = \partial D$ be positively oriented.  
Applying Green's theorem to the two real components of $h\,dz$ gives
$$ \oint_C h(z)\,dz
= -\iint_D\left(\frac{\partial v}{\partial x}
+ \frac{\partial u}{\partial y}\right)\,dA
+ i\iint_D\left(\frac{\partial u}{\partial x}
- \frac{\partial v}{\partial y}\right)\,dA. $$
If $h$ satisfies the Cauchy–Riemann equations, both integrands vanish.  
This is the Green's theorem mechanism behind Cauchy's theorem  
when the region and the partial derivatives satisfy the stated hypotheses.

### Area from Green's theorem

Choose
$$ P(x, y) = -\frac{y}{2},\qquad Q(x, y) = \frac{x}{2}. $$
Then
$$ \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} = 1, $$
so Green's theorem yields the signed-area formula
$$ A_{\mathrm{signed}}
= \frac{1}{2}\oint_C\left(x\,dy - y\,dx\right). $$
It is positive for counterclockwise orientation and negative for clockwise orientation.  
The ordinary area is $|A_{\mathrm{signed}}|$.

### The shoelace formula

Let a simple polygon have vertices
$$ p_k = (x_k, y_k),\qquad k = 1, \ldots, n, $$
listed in boundary order, and set $p_{n + 1} = p_1$.  
Its signed area is
$$ A_{\mathrm{signed}}
= \frac{1}{2}\sum_{k = 1}^n
\left(x_k y_{k + 1} - y_k x_{k + 1}\right). $$
Writing the $x$- and $y$-coordinates in two columns makes the diagonal products  
look like the crossed laces of a shoe, which gives the formula its name.

#### Contribution of one edge

Parametrize the edge from $p_k$ to $p_{k + 1}$ by
$$ x(t) = (1 - t)x_k + t x_{k + 1},\qquad
y(t) = (1 - t)y_k + t y_{k + 1},\qquad 0 \le t \le 1. $$
Along this edge,
$$ \int_0^1\left(x(t)y'(t) - y(t)x'(t)\right)\,dt
= x_k y_{k + 1} - y_k x_{k + 1}. $$
Summing the edge contributions in the area integral obtained from Green's theorem  
produces the shoelace formula.

### Example

Consider the counterclockwise polygon
$$ (0, 0),\quad (4, 0),\quad (4, 2),\quad (1, 3). $$
The edge determinants are
$$ 0,\quad 8,\quad 10,\quad 0. $$
Hence
$$ A = \frac{1}{2}(0 + 8 + 10 + 0) = 9. $$
Reversing the order of the vertices changes every determinant's sign,  
so the signed area becomes $-9$ while the ordinary area remains $9$.

### Complex-number form

Write each vertex as $z_k = x_k + iy_k$.  
Then
$$ \operatorname{Im}\left(\overline{z_k}z_{k + 1}\right)
= x_k y_{k + 1} - y_k x_{k + 1}, $$
and the shoelace formula becomes
$$ A_{\mathrm{signed}}
= \frac{1}{2}\sum_{k = 1}^n
\operatorname{Im}\left(\overline{z_k}z_{k + 1}\right). $$
For a general positively oriented boundary,
$$ A = \frac{1}{2}\operatorname{Im}\oint_C\overline{z}\,dz. $$

### Orientation and self-intersections

For a simple polygon, the sign records the orientation,  
and the absolute value gives the usual geometric area.  
The formula works for both convex and nonconvex simple polygons.

For a self-intersecting polygonal path, the same sum gives **algebraic area**:  
regions traversed with opposite orientations can cancel.  
Computing the total geometric area of all visible lobes then requires  
splitting the path into simple regions or accounting for winding numbers.

## Cauchy's Integral Theorem

### Statement

Let $D$ be a bounded region whose boundary $C = \partial D$ is positively oriented  
and consists of finitely many piecewise smooth, simple closed curves, as in Green's theorem.  
Suppose $f$ is holomorphic on an open set containing $D$ and $C$.  
Then
$$ \boxed{\oint_C f(z)\,dz = 0}. $$

In the most common case, $D$ is the inside of a single simple closed curve $C$.  
The hypothesis then says that $f$ is holomorphic **on and inside** $C$.  
Holomorphy on the curve alone is not enough.

A region is **simply connected** if it has no holes:  
every closed path in it can be shrunk to a point without leaving the region.  
Disks, half-planes, and $\mathbb C$ itself are simply connected;  
the punctured plane $\mathbb C \setminus \{0\}$ and annuli are not.  
If $f$ is holomorphic on a simply connected region $U$,  
then $\oint_\gamma f(z)\,dz = 0$ for every closed path $\gamma$ in $U$.

### Proof from Green's theorem

Write $f = u + iv$.  
Green's theorem, applied to the real and imaginary parts of $f(z)\,dz$, gives
$$ \oint_C f(z)\,dz
= -\iint_D\left(v_x + u_y\right)dA
+ i\iint_D\left(u_x - v_y\right)dA. $$

The Cauchy–Riemann equations say that $u_x = v_y$ and $u_y = -v_x$.  
So $v_x + u_y = 0$ and $u_x - v_y = 0$ at every point of $D$,  
and both double integrals vanish.  
Hence $\oint_C f(z)\,dz = 0$.

Green's theorem needs $u$ and $v$ to have continuous first partial derivatives, that is, $f'$ continuous.  
Goursat showed that the continuity of $f'$ is not needed;  
we do not prove this refinement.

### Examples

Polynomials, $e^z$, $\sin z$, and $\cos z$ are entire,  
so their integrals over every closed curve vanish.  
For example, on the unit circle $z = e^{i\theta}$, $dz = ie^{i\theta}\,d\theta$,
$$ \oint_{|z| = 1} z^2\,dz = i\int_0^{2\pi}e^{3i\theta}\,d\theta
= \frac{e^{6\pi i} - 1}{3} = 0. $$

Now take $f(z) = \bar z$.  
It fails the Cauchy–Riemann equations, since $u_x = 1$ but $v_y = -1$.  
Expanding, $\bar z\,dz = (x\,dx + y\,dy) + i(x\,dy - y\,dx)$.  
The first part is the differential of $(x^2 + y^2)/2$,  
so it integrates to $0$ over a closed curve.  
The second part integrates to twice the area,  
by the area formula of the previous section.  
Hence
$$ \oint_C \bar z\,dz = 2i\cdot\operatorname{area}(D) \ne 0. $$

```python
def tidy(w, tol=1e-9):
    x0, y0 = w.real(), w.imag()
    return CDF(0 if abs(x0) < tol else x0, 0 if abs(y0) < tol else y0)
```

```python
def circle_integral(f, a=0, r=1, n=2000):
    """Integral of f over |z - a| = r, counterclockwise (trapezoid rule)."""
    h = 2 * RDF(pi) / n
    total = CDF(0)
    for k in range(n):
        w = CDF(cos(k * h), sin(k * h))
        total += f(a + r * w) * (CDF(0, 1) * r * w) * h
    return tidy(total)
```

```python
print(circle_integral(lambda z: z^2))
print(circle_integral(lambda z: exp(z)))
print(circle_integral(lambda z: conjugate(z)))
```

The first two integrals vanish, as Cauchy's theorem predicts.  
The third is $2\pi i = 2i\cdot\operatorname{area}(D)$ for the unit disk,  
so a function that is not holomorphic can have a nonzero integral.  
(Values below $10^{-9}$ are printed as $0$; the rest agree with the exact values up to rounding error.)

### Consequences

#### Antiderivatives

Suppose $F$ is holomorphic with $F' = f$ continuous on an open set containing a path $\gamma : [a, b] \to \mathbb C$.  
The chain rule gives $\dfrac{d}{dt}F(\gamma(t)) = f(\gamma(t))\gamma'(t)$,  
and integrating over $[a, b]$ yields the fundamental theorem of calculus for contour integrals:
$$ \int_\gamma f(z)\,dz = F(\gamma(b)) - F(\gamma(a)). $$
For a closed path the right-hand side is $0$.

#### Path independence

Let $f$ be holomorphic on a simply connected region $U$,  
and let $\gamma_1$ and $\gamma_2$ be paths in $U$ from $p$ to $q$.  
The path $\gamma_1$ followed by the reverse of $\gamma_2$ is a closed path in $U$,  
so Cauchy's theorem gives
$$ \int_{\gamma_1} f(z)\,dz = \int_{\gamma_2} f(z)\,dz. $$
The integral depends only on the endpoints, not on the route.

![alt text](1002/image-2.png)

#### Deformation of contours

Let $C_1$ be a simple closed curve inside a simple closed curve $C_2$, both counterclockwise,  
and let $f$ be holomorphic on the region $D$ between them, including the curves themselves.  
The boundary of $D$ is $C_2$ counterclockwise together with $C_1$ clockwise,  
so Cauchy's theorem reads $\oint_{C_2} f\,dz - \oint_{C_1} f\,dz = 0$, that is,
$$ \oint_{C_2} f(z)\,dz = \oint_{C_1} f(z)\,dz. $$

The integral does not change when the contour is deformed through a region where $f$ is holomorphic.  
Nothing is required of $f$ **inside** $C_1$:  
it may fail to be holomorphic, or even to be defined, at points we have cut out.  
This is the key to what follows.

## Integrals near a Pole

Cauchy's theorem needs $f$ to be holomorphic at every point inside the curve.  
What happens to $\oint_C f\,dz$ when $f$ has a pole inside?  
We say that $f$ has a **pole of order $m$** at $a$ if $f(z) = g(z)/(z - a)^m$ near $a$,  
where $g$ is holomorphic at $a$ and $g(a) \ne 0$;  
a pole of order $1$ is called a **simple pole**.  
The simplest examples are the powers $(z - a)^{-m}$.

### The model integrals

Let $C_r$ be the circle $|z - a| = r$, traversed counterclockwise.  
With $z = a + re^{i\theta}$ and $dz = ire^{i\theta}\,d\theta$, we have for every integer $n$
$$ \oint_{C_r}(z - a)^n\,dz
= ir^{n + 1}\int_0^{2\pi}e^{i(n + 1)\theta}\,d\theta. $$

Over a full period, $\int_0^{2\pi}e^{im\theta}\,d\theta$ equals $0$ for $m \ne 0$ and $2\pi$ for $m = 0$.  
Therefore, for every radius $r > 0$,
$$ \boxed{\oint_{C_r}(z - a)^n\,dz
= \begin{cases}
2\pi i, & n = -1, \\
0, & n \ne -1.
\end{cases}} $$

Three observations.

- For $n \ge 0$ the integrand is entire, in agreement with Cauchy's theorem.
- For $n \le -2$ there is a pole of order $|n|$ inside the circle, yet the integral is $0$.
- Only the power $(z - a)^{-1}$ contributes, and it contributes $2\pi i$ **independently of $r$**.

### Why the power $-1$ is special

For $n \ne -1$, the function $(z - a)^n$ has the single-valued antiderivative $(z - a)^{n + 1}/(n + 1)$ on $\mathbb C \setminus \{a\}$,  
so its integral over every closed path avoiding $a$ vanishes.  
For $n = -1$ the antiderivative would be $\log(z - a)$, which is multivalued.  
Going once around $a$ increases $\arg(z - a)$ by $2\pi$, so $\log(z - a)$ gains $2\pi i$.  
Hence $1/(z - a)$ has no single-valued antiderivative on $\mathbb C \setminus \{a\}$.

The same fact in polar form.  
Write $z - a = \rho e^{i\varphi}$, so that $dz = e^{i\varphi}(d\rho + i\rho\,d\varphi)$ and
$$ \frac{dz}{z - a} = \frac{d\rho}{\rho} + i\,d\varphi. $$
Over a closed path, the real part $d\rho/\rho = d(\ln\rho)$ integrates to $0$,  
while the imaginary part integrates to the total change of the angle $\varphi$.  
For a counterclockwise simple closed curve $C$ around $a$ that change is $2\pi$, so $\oint_C \dfrac{dz}{z - a} = 2\pi i$.

If $a$ lies outside $C$, then $1/(z - a)$ is holomorphic on and inside $C$, and the integral is $0$ by Cauchy's theorem.  
If $C$ winds $n$ times counterclockwise around $a$, the angle increases by $2\pi n$, and the integral is $2\pi i\,n$.  
Thus
$$ \frac{1}{2\pi i}\oint_C\frac{dz}{z - a} = \operatorname{wind}(C, a), $$
the **winding number** of $C$ around $a$.

Domain coloring has been showing us this number all along.  
For $f(z) = (z - a)^m$ with $m \in \mathbb Z$, we have $f'(z)/f(z) = m/(z - a)$, so
$$ \frac{1}{2\pi i}\oint_{C_r}\frac{f'(z)}{f(z)}\,dz = m. $$
This is the signed number of times the hue of $f$ winds around the color wheel along $C_r$:  
positive for a zero of order $m$, and negative for a pole of order $-m$.

```python
for r in [1, 0.1, 0.01]:
    print(r, circle_integral(lambda z: 1/z, r=r), circle_integral(lambda z: 1/z^2, r=r))
```

The integral of $1/z$ equals $2\pi i$ for every radius,  
even though the length of the circle tends to $0$ while $|1/z|$ tends to $\infty$.  
The integral of $1/z^2$ is $0$ for every radius,  
although $|1/z^2| = r^{-2}$ is huge on small circles.

### Deformation to a small circle

Let $f$ be holomorphic on a region except at one point $a$, and let $C$ be a counterclockwise simple closed curve around $a$.  
Choose $r$ so small that $C_r$ lies inside $C$.  
Cauchy's theorem on the region between $C$ and $C_r$ gives
$$ \oint_C f(z)\,dz = \oint_{C_r} f(z)\,dz\qquad\text{for every small }r > 0. $$
So the integral around $C$ depends only on how $f$ behaves arbitrarily close to $a$.

### The ML-inequality

Let $\gamma$ be a piecewise smooth path of length $L$, and suppose $|h(z)| \le M$ on $\gamma$.  
Then
$$ \left|\int_\gamma h(z)\,dz\right| \le ML. $$
Indeed,
$$ \left|\int_a^b h(\gamma(t))\gamma'(t)\,dt\right|
\le \int_a^b|h(\gamma(t))|\,|\gamma'(t)|\,dt
\le M\int_a^b|\gamma'(t)|\,dt = ML. $$

### A simple pole

Suppose $f(z) = g(z)/(z - a)$ with $g$ holomorphic near $a$.  
Write $g(z) = g(a) + \bigl(g(z) - g(a)\bigr)$ and integrate over $C_r$:
$$ \oint_{C_r} f(z)\,dz
= g(a)\oint_{C_r}\frac{dz}{z - a}
+ \oint_{C_r}\frac{g(z) - g(a)}{z - a}\,dz
= 2\pi i\,g(a) + E(r). $$

On $C_r$ we have $|z - a| = r$, and the length is $2\pi r$, so the ML-inequality gives
$$ |E(r)| \le 2\pi r\cdot\frac{\max_{C_r}|g(z) - g(a)|}{r}
= 2\pi\max_{|z - a| = r}|g(z) - g(a)|. $$
Since $g$ is continuous at $a$, the right-hand side tends to $0$ as $r \to 0$.

On the other hand, $\oint_{C_r} f\,dz$ does not depend on $r$ by deformation,  
so $E(r)$ is a constant that tends to $0$.  
Hence $E \equiv 0$, and for every curve $C$ around $a$,
$$ \boxed{\oint_C f(z)\,dz = 2\pi i\,g(a) = 2\pi i\lim_{z \to a}(z - a)f(z)}. $$

The number $g(a) = \lim_{z \to a}(z - a)f(z)$ is called the **residue** of $f$ at $a$.  
Compare the effects as $r \to 0$:  
the length $2\pi r$ tends to $0$, but $|f| \approx |g(a)|/r$ grows at exactly the compensating rate,  
so the integral stays at $2\pi i\,g(a)$.

At a pole of order $m \ge 2$, $|f| \approx r^{-m}$ outgrows the length,  
and the ML bound $2\pi r^{1 - m}$ blows up, yet the integral can still be finite because of cancellation.  
If instead $|f| \le M$ near $a$, then $\left|\oint_{C_r} f\,dz\right| \le 2\pi rM \to 0$,  
and since the integral does not depend on $r$, it equals $0$.

## Cauchy's Integral Formula

### Statement

Let $C$ be a positively oriented simple closed curve, and let $f$ be holomorphic on and inside $C$.  
For every point $a$ inside $C$,
$$ \boxed{f(a) = \frac{1}{2\pi i}\oint_C\frac{f(z)}{z - a}\,dz}. $$

The values of $f$ on the curve $C$ determine the value of $f$ at **every** point inside.  
A holomorphic function cannot be altered inside $C$ without being altered on $C$.  
Renaming $a \to z$ and $z \to \zeta$ gives the equivalent form
$$ f(z) = \frac{1}{2\pi i}\oint_C\frac{f(\zeta)}{\zeta - z}\,d\zeta. $$

![alt text](1002/image-4.png)

### Proof

The function $z \mapsto f(z)/(z - a)$ has the form $g(z)/(z - a)$ with $g = f$ holomorphic.  
It has at most a simple pole at $a$,  
so the computation of the previous section applies verbatim:
$$ \oint_C\frac{f(z)}{z - a}\,dz = 2\pi i\,f(a). $$
Dividing by $2\pi i$ gives the formula.

The proof has three steps.

1. Deform $C$ to a small circle $C_r$ around $a$ (Cauchy's theorem).
2. Replace $f(z)$ by the constant $f(a)$; the ML-inequality controls the error.
3. Integrate $1/(z - a)$ exactly (the model integral).

### Mean value property

Take $C$ to be the circle $|z - a| = r$, with $z = a + re^{i\theta}$ and $dz = ire^{i\theta}\,d\theta$:
$$ f(a) = \frac{1}{2\pi i}\int_0^{2\pi}\frac{f(a + re^{i\theta})}{re^{i\theta}}\,ire^{i\theta}\,d\theta
= \frac{1}{2\pi}\int_0^{2\pi}f(a + re^{i\theta})\,d\theta. $$
The value at the center is the average of the values on the circle.  
In particular, $|f(a)| \le \max_{|z - a| = r}|f(z)|$.

### Examples

#### A pole inside the circle

Compute $\displaystyle\oint_{|z| = 2}\frac{e^z}{z - 1}\,dz$.  
The point $a = 1$ lies inside the circle, and $f(z) = e^z$ is entire, so
$$ \oint_{|z| = 2}\frac{e^z}{z - 1}\,dz = 2\pi i\,f(1) = 2\pi i\,e. $$

#### A pole outside the circle

Compute $\displaystyle\oint_{|z| = 1}\frac{e^z}{z - 2}\,dz$.  
The only singularity is at $z = 2$, outside the circle,  
so the integrand is holomorphic on and inside the circle, and the integral is $0$ by Cauchy's theorem.

#### Two poles

Compute $\displaystyle\oint_{|z| = 2}\frac{e^z}{z^2 + 1}\,dz$.  
Both poles $\pm i$ lie inside the circle.  
Partial fractions give
$$ \frac{1}{z^2 + 1} = \frac{1}{2i}\left(\frac{1}{z - i} - \frac{1}{z + i}\right), $$
and the formula applies to each term:
$$ \oint_{|z| = 2}\frac{e^z}{z^2 + 1}\,dz
= \frac{1}{2i}\left(2\pi i\,e^i - 2\pi i\,e^{-i}\right)
= \pi\left(e^i - e^{-i}\right) = 2\pi i\sin 1. $$

When several poles lie inside $C$, deform $C$ into small disjoint circles, one around each pole.  
Cauchy's theorem on the region inside $C$ and outside the small circles shows that the integral over $C$ is the sum of the integrals over the small circles.  
Each of these is computed as above.

```python
print(circle_integral(lambda z: exp(z)/(z - 1), r=2))
print(CDF(2 * pi * i * e))
print(circle_integral(lambda z: exp(z)/(z^2 + 1), r=2))
print(CDF(2 * pi * i * sin(1)))
```

Each printed pair agrees: the numerical integral and the value given by the formula.

### Derivatives

The kernel $1/(z - a)$ depends smoothly on $a$,  
so the formula can be differentiated with respect to $a$ under the integral sign:
$$ f^{(k)}(a) = \frac{k!}{2\pi i}\oint_C\frac{f(z)}{(z - a)^{k + 1}}\,dz,\qquad k = 0, 1, 2, \ldots. $$
In particular, a holomorphic function is infinitely differentiable.  
The second homework derives this formula for functions given by power series.

### Summary

| Situation | $\displaystyle\oint_C f(z)\,dz$ |
|---|---|
| $f$ holomorphic on and inside $C$ | $0$ |
| $f(z) = \dfrac{g(z)}{z - a}$, with $g$ holomorphic on and inside $C$ and $a$ inside $C$ | $2\pi i\,g(a)$ |
| $f(z) = \dfrac{g(z)}{(z - a)^{k + 1}}$, with the same assumptions | $\dfrac{2\pi i}{k!}\,g^{(k)}(a)$ |

Only the coefficient of $(z - a)^{-1}$ survives the integration; residues will measure exactly this.

#### (c) From countably many points to the whole set

Let $(g_k)$ be the diagonal subsequence from (b), with $S$ as in (a).  
Show that $(g_k)$ is **uniformly Cauchy** on $K$.  
Given $\varepsilon > 0$, choose $\delta$ by equicontinuity, then $m$ with $1/m < \delta$, and let $F = F_{1/m} \subset S$.  
Choose $N$ such that $|g_k(s) - g_l(s)| < \varepsilon$ for all $s \in F$ and all $k, l \ge N$.  
For an arbitrary $z \in K$, pick $s \in F$ with $|z - s| < \delta$ and estimate
$$ |g_k(z) - g_l(z)| \le |g_k(z) - g_k(s)| + |g_k(s) - g_l(s)| + |g_l(s) - g_l(z)|. $$
Conclude, using the completeness of $\mathbb C$, that $(g_k)$ converges uniformly on $K$.

#### (d) Holomorphic functions give equicontinuous families

Surprise: it's all about Cauchy Integral formulachallenge

Let each $f_n$ be holomorphic on an open set containing the closed disk $|z - a| \le 2r$, with $|f_n(z)| \le M$ there.  
Take for granted that Cauchy's formula for derivatives and Cauchy's estimate hold for such functions.  
If $|z - a| \le r$, then the circle $|\zeta - z| = r$ lies in the disk $|\zeta - a| \le 2r$,  
so Cauchy's estimate gives $|f_n'(z)| \le M/r$.  
Integrate $f_n'$ along the line segment from $w$ to $z$ to show
$$ |f_n(z) - f_n(w)| \le \frac{M}{r}|z - w| $$
whenever $|z - a| \le r$ and $|w - a| \le r$.  
Conclude that $(f_n)$ has a subsequence converging uniformly on $|z - a| \le r$.  
A family with this property on every compact subset is called **normal**.
