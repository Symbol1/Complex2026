### Two definitions of a limit point

Let $S \subset \mathbb C$, let $a \in \mathbb C$, and write
$$ D(a, r) = \{z \in \mathbb C : |z - a| < r\}. $$
The point $a$ is allowed to belong to $S$, but neither formulation below  
may use $a$ itself as evidence that nearby points of $S$ exist.

#### Neighborhood formulation

Call $a$ a **neighborhood limit point** of $S$ if every punctured disk  
centered at $a$ intersects $S$:
$$ \bigl(D(a, r) \setminus \{a\}\bigr) \cap S \ne \varnothing
\qquad \text{for every }r > 0. $$

#### Sequential formulation

Call $a$ a **sequential limit point** of $S$ if there exists a sequence  
$(z_n)$ in $S \setminus \{a\}$ such that
$$ z_n \to a. $$
You may strengthen the formulation by requiring the terms $z_n$  
to be pairwise distinct.

#### Prove the equivalence

Prove that $a$ is a neighborhood limit point of $S$  
if and only if $a$ is a sequential limit point of $S$.

Your proof must include both directions:

1. construct a convergent sequence using punctured disks around $a$; and
2. use convergence of a sequence to prove that every punctured disk meets $S$.

Also explain why removing $\{a\}$ from the disks and from the sequence  
is necessary when $a \in S$.

### Closed sets and their limit points

For $F \subset \mathbb C$, let
$$ F' = \{a \in \mathbb C : a \text{ is a limit point of }F\}. $$
Prove the equivalence
$$ \mathbb C \setminus F \text{ is open}
\quad\Longleftrightarrow\quad
F' \subset F. $$
Thus a set is closed exactly when it contains every one of its limit points.

Organize the proof in both directions:

1. Assume $\mathbb C \setminus F$ is open.  
   Show that a point outside $F$ cannot be a limit point of $F$.
2. Assume $F' \subset F$.  
   For each $a \notin F$, find a disk centered at $a$ contained in $\mathbb C \setminus F$.

Finally, use the first problem to derive the sequential version:  
$F$ is closed if and only if every convergent sequence $(z_n)$ in $F$  
has its limit in $F$.
