# ENGINEERING MATH - COMPLEX VARIABLE

Recall that in 1106 we found the steady temperature $\frac1\pi\operatorname{Arg}z$  
in a half-plane  
whose edge is held at $0$ on one side and at $1$ on the other.  
Today we solve such problems on other regions:  
map the region onto the half-plane by a holomorphic function,  
solve the problem there, and carry the solution back.

Two facts make this work.  
A holomorphic map preserves angles where its derivative is nonzero.  
Composed with a holomorphic map,  
a harmonic function stays harmonic.

We first study such angle-preserving maps,  
called **conformal mappings**,  
then collect a dictionary of standard regions,  
and end with the Poisson integral formulas  
and the Schwarz–Christoffel formula.

## Conformal Mappings

### Tangent vectors

Let $\gamma(t)$ be a smooth curve with $\gamma(t_0) = z_0$ and $\gamma'(t_0) \ne 0$,  
and let $f$ be holomorphic at $z_0$.  
The image curve $f\bigl(\gamma(t)\bigr)$ has the tangent vector
$$ \frac{d}{dt}f\bigl(\gamma(t)\bigr)\Big|_{t=t_0} = f'(z_0)\gamma'(t_0). $$

This is the chain rule, proved as in calculus.  
For small $h \ne 0$, we have $\gamma(t_0 + h) \ne z_0$, and
$$ \begin{aligned}
&\frac{f\bigl(\gamma(t_0 + h)\bigr) - f(z_0)}{h} \\
&\quad = \frac{f\bigl(\gamma(t_0 + h)\bigr) - f(z_0)}{\gamma(t_0 + h) - z_0} \cdot \frac{\gamma(t_0 + h) - z_0}{h}.
\end{aligned} $$
As $h \to 0$, the two factors tend to $f'(z_0)$ and $\gamma'(t_0)$.

Write $f'(z_0) = \rho e^{i\varphi}$ with $\rho > 0$.  
Multiplication by $f'(z_0)$ rotates every tangent vector at $z_0$  
by the same angle $\varphi$  
and stretches it by the same factor $\rho$.

### Angles are preserved

**Theorem.**  
If $f$ is holomorphic at $z_0$ and $f'(z_0) \ne 0$,  
then $f$ preserves the angle between any two curves through $z_0$,  
both in size and in orientation.

The angle from a curve $\gamma_1$ to a curve $\gamma_2$ at $z_0$  
is $\arg(\gamma_2'/\gamma_1')$, with the derivatives taken at $z_0$.  
After the map, it becomes
$$ \arg\frac{f'(z_0)\gamma_2'}{f'(z_0)\gamma_1'} = \arg\frac{\gamma_2'}{\gamma_1'}. $$

A map that preserves angles in this way at every point of a region  
is called **conformal** there.  
So a holomorphic function is conformal wherever $f' \ne 0$.  
The reflection $z \mapsto \bar z$ preserves the sizes of angles  
but reverses their orientation, and it is not holomorphic.

Moreover, $|f(z) - f(z_0)| \approx |f'(z_0)||z - z_0|$ near $z_0$:  
the map stretches all directions by the same factor $|f'(z_0)|$,  
so small circles go to approximate circles,  
and small squares to approximate squares.

Recall that in 0911 we represented $a + bi$  
by the real matrix $\begin{pmatrix} a & -b \\ b & a \end{pmatrix}$.  
By the Cauchy–Riemann equations,  
the Jacobian matrix of $f = u + iv$ is
$$ \begin{pmatrix} u_x & u_y \\ v_x & v_y \end{pmatrix} = \begin{pmatrix} u_x & -v_x \\ v_x & u_x \end{pmatrix}, $$
the matrix of $f'(z_0) = u_x + iv_x$.

So, to first order, $f$ acts near $z_0$ as a rotation followed by a scaling,  
and it multiplies areas by the determinant $u_x^2 + v_x^2 = |f'(z_0)|^2$.

The picture shows the images of the lines  
$x = c$ (red) and $y = c$ (blue) under $w = z^2$.  
They become two families of parabolas,  
which still cross at right angles.

```python
var('t')
red = sum(parametric_plot((c^2 - t^2, 2*c*t), (t, 0, 2), color='red') for c in srange(0.25, 2.01, 0.25))
blue = sum(parametric_plot((t^2 - c^2, 2*c*t), (t, 0, 2), color='blue') for c in srange(0.25, 2.01, 0.25))
(red + blue).show(aspect_ratio=1)
```

### Critical points

Where $f'(z_0) = 0$, angles are not preserved.  
If $f(z) - f(z_0)$ has a zero of order $m \ge 2$ at $z_0$,  
then $f(z) - f(z_0) = (z - z_0)^m g(z)$ with $g(z_0) \ne 0$, so
$$ \arg\bigl(f(z) - f(z_0)\bigr) \approx m\arg(z - z_0) + \arg g(z_0): $$
angles at $z_0$ are multiplied by $m$.

Recall that in 1106 the map $z^2$  
took the first quadrant onto the upper half-plane:  
the right angle at $0$ became a straight angle.  
This is how we will straighten the corners of regions.

## Möbius Transformations

### Conformal on the whole sphere

Recall from the 1106 homework the Möbius transformations
$$ T(z) = \frac{az + b}{cz + d},\qquad ad - bc \ne 0. $$
They are compositions of translations, complex scalings,  
and the inversion $1/z$,  
and they map circles and lines onto circles and lines.

Since
$$ T'(z) = \frac{ad - bc}{(cz + d)^2} \ne 0, $$
the map $T$ is conformal at every point of $\mathbb C$ except $-d/c$.  
With $T(-d/c) = \infty$ and $T(\infty) = a/c$,  
it maps $\hat{\mathbb C}$ one-to-one onto itself.

Measured in the coordinate $1/z$ near $\infty$,  
it is conformal at $-d/c$ and at $\infty$ as well.  
In fact, the Möbius transformations  
are the only one-to-one meromorphic maps of $\hat{\mathbb C}$ onto itself;  
the exercises of 1126 give a guided proof.

### Three points determine a Möbius transformation

Given three distinct points $z_1$, $z_2$, $z_3$ in $\mathbb C$,
$$ S(z) = \frac{(z - z_1)(z_2 - z_3)}{(z - z_3)(z_2 - z_1)} $$
is a Möbius transformation  
with $S(z_1) = 0$, $S(z_2) = 1$, and $S(z_3) = \infty$.  
In the notation of the 1106 homework,  
$S(z)$ is the cross ratio $(z, z_2; z_1, z_3)$.

A Möbius transformation that fixes $0$, $1$, and $\infty$ is the identity:  
$T(\infty) = \infty$ forces $c = 0$,  
then $T(0) = 0$ forces $b = 0$,  
and $T(1) = 1$ forces $a = d$.

Hence, for distinct $w_1$, $w_2$, $w_3$,  
there is exactly one Möbius transformation $T$ with $T(z_k) = w_k$.  
If $S_w$ is the map above for $w_1$, $w_2$, $w_3$,  
then $S_w \circ T$ and $S$ agree at $z_1$, $z_2$, $z_3$,  
so $S_w \circ T = S$ and $T = S_w^{-1} \circ S$.

In practice, solve
$$ \frac{(w - w_1)(w_2 - w_3)}{(w - w_3)(w_2 - w_1)} = \frac{(z - z_1)(z_2 - z_3)}{(z - z_3)(z_2 - z_1)} $$
for $w$.  
If one of the points is $\infty$,  
delete the two factors that contain it.

For example, to send $-1$, $0$, $1$ to $0$, $i$, $\infty$,  
delete the factors with $w_3 = \infty$:
$$ \frac{w}{i} = \frac{(z + 1)(0 - 1)}{(z - 1)(0 + 1)},\qquad\text{so}\qquad w = i\,\frac{1 + z}{1 - z}. $$

### The disk and the half-plane

The **Cayley transformation**
$$ C(z) = \frac{z - i}{z + i} $$
maps the upper half-plane onto the unit disk.  
Indeed, $|C(z)| < 1$ exactly when $z$ is closer to $i$ than to $-i$,  
that is, when $\operatorname{Im}z > 0$.

The real axis goes onto the unit circle minus the point $1 = C(\infty)$.  
The inverse map is the example above,
$$ C^{-1}(w) = i\,\frac{1 + w}{1 - w}, $$
which sends $-1$, $0$, $1$ to $0$, $i$, $\infty$.

For $|a| < 1$, the map
$$ \varphi_a(z) = \frac{z - a}{1 - \bar az} $$
maps the unit disk onto itself and sends $a$ to $0$.

Indeed, on $|z| = 1$,  
$|z - a| = |\bar z(z - a)| = |1 - a\bar z| = |1 - \bar az|$,  
so $\varphi_a$ maps the unit circle to itself,  
and the point $a$ inside goes to $0$ inside.

Its inverse is $\varphi_{-a}(z) = (z + a)/(1 + \bar az)$, with $\varphi_{-a}(0) = a$.  
So in a disk, every point looks like the center:  
we will use this for the Poisson formula in the disk.

## A Dictionary of Mappings

### Standard maps

| $w = f(z)$ | region in $z$ | image in $w$ |
|---|---|---|
| $z^{\pi/\alpha}$, principal branch | sector $0 < \arg z < \alpha$, $0 < \alpha \le 2\pi$ | upper half-plane |
| $e^z$ | strip $0 < \operatorname{Im}z < \pi$ | upper half-plane |
| $\operatorname{Log}z$ | upper half-plane | strip $0 < \operatorname{Im}w < \pi$ |
| $\sin z$ | half-strip $\lvert\operatorname{Re}z\rvert < \pi/2$, $\operatorname{Im}z > 0$ | upper half-plane |
| $(z - i)/(z + i)$ | upper half-plane | unit disk |
| $(1 + z)/(1 - z)$ | unit disk | right half-plane |
| $(z - a)/(1 - \bar az)$, $\lvert a\rvert < 1$ | unit disk | unit disk |

The power multiplies angles at $0$ by $\pi/\alpha$,  
as in 1106.  
For the exponential, $e^{x+iy} = e^xe^{iy}$:  
the horizontal line $\operatorname{Im}z = c$ goes onto the ray $\arg w = c$,  
and the vertical segment $\operatorname{Re}z = c$, $0 < \operatorname{Im}z < \pi$,  
goes onto the upper half of the circle $|w| = e^c$.

For the sine,
$$ \sin(x + iy) = \sin(x)\cosh y + i\cos(x)\sinh y. $$
On the half-strip, $\operatorname{Im}\sin z = \cos(x)\sinh y > 0$.  
The bottom edge $-\pi/2 < x < \pi/2$ goes onto $(-1, 1)$,  
and the sides $x = \pm\pi/2$ go onto $\pm\cosh y$,  
that is, onto $[1, \infty)$ and $(-\infty, -1]$.

The picture shows the images under $\sin z$  
of the vertical lines in the half-strip (red),  
which become hyperbolas,  
and of the horizontal segments (blue),  
which become upper halves of ellipses;  
all of them have foci $\pm1$.

```python
red = sum(parametric_plot((sin(c)*cosh(t), cos(c)*sinh(t)), (t, 0, 1.6), color='red') for c in srange(-1.4, 1.41, 0.2))
blue = sum(parametric_plot((sin(t)*cosh(c), cos(t)*sinh(c)), (t, -pi/2, pi/2), color='blue') for c in srange(0.2, 1.61, 0.2))
(red + blue + point([(-1, 0), (1, 0)], size=30, color='black')).show(aspect_ratio=1)
```

### Composing maps

More regions are reached by composing the maps in the table.  
For example, $z^2$ maps the first quadrant onto the upper half-plane,  
and the Cayley transformation maps that onto the unit disk, so
$$ w = \frac{z^2 - i}{z^2 + i} $$
maps the first quadrant onto the unit disk.

For the upper half-disk $|z| < 1$, $\operatorname{Im}z > 0$,  
the map $(1 + z)/(1 - z)$ sends $-1$, $0$, $1$, $i$ to $0$, $1$, $\infty$, $i$.  
It maps the segment $(-1, 1)$ onto the positive real axis  
and the upper semicircle onto the positive imaginary axis,  
so it maps the half-disk onto the first quadrant.

Squaring then gives the upper half-plane:
$$ w = \Bigl(\frac{1 + z}{1 - z}\Bigr)^2 $$
maps the upper half-disk onto the upper half-plane.  
To check which side is the image, test one point,  
for example $z = i/2 \mapsto \bigl((3 + 4i)/5\bigr)^2$.

## Dirichlet Problems

### Harmonic functions are carried along

**Theorem.**  
Let $f$ be holomorphic on a region $D$ with values in a region $D'$,  
and let $U$ be harmonic on $D'$.  
Then $u = U \circ f$ is harmonic on $D$.

Recall that in 1106 we constructed harmonic conjugates on disks.  
Near $f(z_0)$, write $U = \operatorname{Re}F$ with $F$ holomorphic on a disk.  
Near $z_0$, then $u = \operatorname{Re}(F \circ f)$  
is the real part of a holomorphic function,  
so it is harmonic, as shown in 1106.

Boundary values are carried along as well:  
if $f$ maps an arc of the boundary of $D$  
onto an arc of the boundary of $D'$ on which $U = c$,  
then $u = c$ on that arc.

The **Dirichlet problem** asks for a function  
that is harmonic in a region $D$  
and has prescribed values on the boundary of $D$:  
for example, the steady temperature in a plate  
whose edges are held at given temperatures.

The strategy of this lecture:  
map $D$ conformally onto the upper half-plane,  
solve the problem there,  
and compose the solution with the map.  
For a bounded region, the maximum principle from 1106 shows that  
two solutions with the same continuous boundary values coincide.

### Piecewise constant data on the half-plane

Shifting the solution from 1106 by $a$,  
$\operatorname{Arg}(z - a)/\pi$ is harmonic in the upper half-plane,  
tends to $1$ on $(-\infty, a)$ and to $0$ on $(a, \infty)$.

So for $a < b$,
$$ u(z) = \frac1\pi\bigl(\operatorname{Arg}(z - b) - \operatorname{Arg}(z - a)\bigr) $$
is $1$ on the segment $(a, b)$ and $0$ on the rest of the real axis.

The difference $\operatorname{Arg}(z - b) - \operatorname{Arg}(z - a)$  
is the angle at $z$ under which the segment $[a, b]$ is seen.  
By the inscribed angle theorem from the 1106 homework,  
the isotherms $u = c$ are circular arcs through $a$ and $b$.

```python
var('x y')
u = (atan2(y, x - 1) - atan2(y, x + 1))/pi
T = contour_plot(u, (x, -3, 3), (y, 0, 3), contours=srange(0.1, 0.91, 0.1), cmap='coolwarm', colorbar=True, plot_points=300)
T.show(aspect_ratio=1)
```

More generally, let the boundary value be $c_0$ on $(-\infty, a_1)$,  
$c_k$ on $(a_k, a_{k+1})$, and $c_n$ on $(a_n, \infty)$,  
where $a_1 < \cdots < a_n$.  
Adding such steps gives
$$ u(z) = c_n + \frac1\pi\sum_{k=1}^n(c_{k-1} - c_k)\operatorname{Arg}(z - a_k). $$

On $(a_j, a_{j+1})$, the terms with $k \le j$ vanish  
and those with $k > j$ have $\operatorname{Arg} = \pi$,  
so the sum telescopes to $c_j - c_n$, and $u = c_j$ there.

### Example: a disk

Find the steady temperature $u$ in the unit disk  
whose upper semicircle is held at $1$ and lower semicircle at $0$.

The inverse Cayley map $w = i(1 + z)/(1 - z)$  
maps the disk onto the upper half-plane.  
On the circle $z = e^{i\theta}$,
$$ i\,\frac{1 + e^{i\theta}}{1 - e^{i\theta}} = i\,\frac{e^{-i\theta/2} + e^{i\theta/2}}{e^{-i\theta/2} - e^{i\theta/2}} = -\cot\frac\theta2, $$
which is negative for $0 < \theta < \pi$ and positive for $\pi < \theta < 2\pi$.

So the upper semicircle goes onto the negative real axis,  
where we want $1$,  
and the lower semicircle onto the positive real axis,  
where we want $0$.  
The solution in the $w$-plane is $\operatorname{Arg}(w)/\pi$, so
$$ u(z) = \frac1\pi\operatorname{Arg}\Bigl(i\,\frac{1 + z}{1 - z}\Bigr) = \frac12 + \frac1\pi\operatorname{Arg}\frac{1 + z}{1 - z}. $$

Since
$$ \frac{1 + z}{1 - z} = \frac{(1 + z)(1 - \bar z)}{|1 - z|^2} = \frac{1 - x^2 - y^2 + 2iy}{|1 - z|^2} $$
has positive real part in the disk, we get
$$ u(x, y) = \frac12 + \frac1\pi\arctan\frac{2y}{1 - x^2 - y^2}. $$

At the center, $u = 1/2$, the average of the boundary values,  
as the mean value property from 1106 requires.  
The isotherms are the images of the rays from $0$ in the $w$-plane,  
which are circular arcs through $\pm1$.

```python
u = 1/2 + atan2(2*y, 1 - x^2 - y^2)/pi
T = contour_plot(u, (x, -1, 1), (y, -1, 1), contours=srange(0.1, 0.91, 0.1), cmap='coolwarm', colorbar=True, plot_points=300, region=1 - x^2 - y^2)
T.show(aspect_ratio=1)
```

### Example: a half-strip

Let the bottom edge of the half-strip $|x| < \pi/2$, $y > 0$  
be held at $1$ and its two sides at $0$.  
The map $w = \sin z$ sends the bottom edge onto $(-1, 1)$  
and the sides onto the rest of the real axis, so
$$ u(z) = \frac1\pi\bigl(\operatorname{Arg}(\sin z - 1) - \operatorname{Arg}(\sin z + 1)\bigr). $$

## The Poisson Integral Formulas

### The upper half-plane

For $z = x + iy$ with $y > 0$,  
as $t$ runs along the real axis from $-\infty$ to $\infty$,  
$\operatorname{Arg}(z - t)$ increases from $0$ to $\pi$, with
$$ \frac{d}{dt}\operatorname{Arg}(z - t) = \frac{y}{(x - t)^2 + y^2}. $$

So the solution for a single step is
$$ \frac1\pi\bigl(\operatorname{Arg}(z - b) - \operatorname{Arg}(z - a)\bigr) = \frac1\pi\int_a^b\frac{y\,dt}{(x - t)^2 + y^2}. $$
Adding steps, and passing to a limit, suggests the following formula.

**Poisson integral formula for the upper half-plane.**  
Let $g$ be bounded and piecewise continuous on $\mathbb R$. Then
$$ \boxed{u(x, y) = \frac y\pi\int_{-\infty}^\infty\frac{g(t)}{(x - t)^2 + y^2}\,dt} $$
is bounded and harmonic in the upper half-plane,  
and $u(x, y) \to g(x)$ as $y \to 0$ wherever $g$ is continuous.

The **Poisson kernel** is harmonic in $z$,  
because it is the imaginary part of a holomorphic function of $z$:
$$ \frac1\pi\frac{y}{(x - t)^2 + y^2} = \frac1\pi\operatorname{Im}\frac{1}{t - z}. $$
Differentiating under the integral sign shows that $u$ is harmonic.

The kernel has total integral $1$ for every $y > 0$.  
As $y \to 0$, it concentrates near $t = x$,  
as the picture shows for $x = 0$,  
so $u(x, y)$ becomes an average of $g$  
over a shrinking neighborhood of $x$.

```python
plot([y/(pi*(t^2 + y^2)) for y in [1, 0.5, 0.2]], (t, -4, 4), legend_label=['$y = 1$', '$y = 0.5$', '$y = 0.2$'],
     color=['blue', 'green', 'red']).show()
```

For example, let $g(t) = 1/(1 + t^2)$.  
Then
$$ \frac{1}{1 + t^2} = \operatorname{Re}\frac{i}{t + i}, $$
and $i/(z + i)$ is holomorphic and bounded on $\operatorname{Im}z > 0$.

So the bounded solution is
$$ u(x, y) = \operatorname{Re}\frac{i}{z + i} = \frac{1 + y}{x^2 + (1 + y)^2}. $$
The Poisson integral gives the same function,  
as the residue theorem from 1120 confirms (Optional Exercise 11).

### The unit disk

**Poisson integral formula for the disk.**  
Let $u$ be harmonic on an open set containing the closed unit disk,  
with boundary values $g(\varphi) = u(e^{i\varphi})$.  
Then, for $0 \le r < 1$,
$$ u(re^{i\theta}) = \frac{1}{2\pi}\int_0^{2\pi}\frac{1 - r^2}{1 - 2r\cos(\theta - \varphi) + r^2}\,g(\varphi)\,d\varphi. $$

At the center $r = 0$, this is the mean value property from 1106.  
We move an arbitrary point $a = re^{i\theta}$ to the center with $\varphi_a$.

The map $\psi = \varphi_{-a}$ sends $0$ to $a$,  
and $u \circ \psi$ is harmonic by the theorem above.  
The mean value property for $u \circ \psi$ at $0$ gives
$$ u(a) = \frac{1}{2\pi}\int_0^{2\pi}u\bigl(\psi(e^{is})\bigr)\,ds. $$

Substitute $e^{is} = \varphi_a(e^{i\varphi})$, so that $\psi(e^{is}) = e^{i\varphi}$.  
As $e^{i\varphi}$ runs counterclockwise around the circle, so does $\varphi_a(e^{i\varphi})$,  
with speed $|\varphi_a'(e^{i\varphi})|$.  
Hence $ds = |\varphi_a'(e^{i\varphi})|\,d\varphi$.

The derivative is
$$ \varphi_a'(z) = \frac{1 - |a|^2}{(1 - \bar az)^2}, $$
and $|1 - \bar az| = |z - a|$ on $|z| = 1$.

Hence
$$ ds = \frac{1 - |a|^2}{|e^{i\varphi} - a|^2}\,d\varphi = \frac{1 - r^2}{1 - 2r\cos(\theta - \varphi) + r^2}\,d\varphi, $$
which proves the formula.

For $g = 1$, the formula must give $1$.  
Write the denominator as $2r\bigl(A - \cos(\theta - \varphi)\bigr)$  
with $A = (1 + r^2)/(2r)$;  
this follows from $\int_0^{2\pi}d\varphi/(A - \cos\varphi) = 2\pi/\sqrt{A^2 - 1}$,  
the family of integrals evaluated in 1120,  
since $\sqrt{A^2 - 1} = (1 - r^2)/(2r)$.

## The Schwarz–Christoffel Formula

### Polygons

The dictionary reaches only a few regions.  
For a polygon there is a general formula,  
which maps the upper half-plane onto the polygon.

**Schwarz–Christoffel formula.**  
Let $x_1 < x_2 < \cdots < x_n$ be real,  
let $0 < \alpha_k < 2\pi$, and let $A \ne 0$.  
Let $f$ be an antiderivative of
$$ f'(z) = A(z - x_1)^{\alpha_1/\pi-1}\cdots(z - x_n)^{\alpha_n/\pi-1} $$
in the upper half-plane, with the principal branch of each power.

Then $f$ maps the real axis onto a polygonal path  
whose interior angle at each vertex $f(x_k)$ is $\alpha_k$.

When the angles satisfy $\sum_k(\pi - \alpha_k) = 2\pi$,  
the path closes up into a polygon,  
and $f$ maps the upper half-plane conformally onto its inside.  
Otherwise, the polygon has a vertex at $\infty$,  
which is the image of $\infty$.

### Why the edges are straight

On the real axis between $x_k$ and $x_{k+1}$,  
each factor $(x - x_j)^{\alpha_j/\pi-1}$ has a constant argument:  
$0$ when $x > x_j$, and $(\alpha_j/\pi - 1)\pi$ when $x < x_j$.  
So $\arg f'(x)$ is constant there,  
and $f$ maps the interval onto a straight segment.

As $x$ passes $x_k$ from left to right,  
$\arg(x - x_k)$ drops from $\pi$ to $0$,  
so $\arg f'$ changes by $-(\alpha_k/\pi - 1)\pi = \pi - \alpha_k$.  
The path turns by the exterior angle $\pi - \alpha_k$,  
so its interior angle at $f(x_k)$ is $\alpha_k$.  
The total turning of a closed polygon is $2\pi$.

### Examples

**One vertex.**  
With $n = 1$, $x_1 = 0$, and $\alpha_1 = \alpha$,
$$ f(z) = A\int^z\zeta^{\alpha/\pi-1}\,d\zeta = \frac{A\pi}{\alpha}z^{\alpha/\pi}, $$
which maps the upper half-plane onto a sector of angle $\alpha$:  
the inverse of the power in the dictionary.

**A half-strip.**  
With vertices at $x = \pm1$ and angles $\pi/2$ at both,
$$ f'(z) = A(z + 1)^{-1/2}(z - 1)^{-1/2}. $$
Choosing $A$ suitably gives $f'(z) = (1 - z^2)^{-1/2}$, so $f = \arcsin$,  
which maps the upper half-plane onto the half-strip  
$|\operatorname{Re}w| < \pi/2$, $\operatorname{Im}w > 0$.

This is the inverse of the sine in the dictionary,  
and the third vertex of the half-strip is at $\infty$.

For most polygons, the integral is not elementary;  
for a rectangle, it is an elliptic integral.  
Still, the formula shows that such a map exists.  
More generally, the **Riemann mapping theorem** says that  
every simply connected region other than $\mathbb C$ itself  
can be mapped conformally onto the unit disk,  
so in principle this strategy works for every such region.
