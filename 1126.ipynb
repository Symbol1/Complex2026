These notes look at two kinds of points  
where the residue theorem from 1016 does not apply as stated:  
the point $\infty$ and branch points.

At $\infty$ we define limits, singularities, and residues  
through the substitution $w = 1/z$.  
The residue at $\infty$ often replaces a sum of many residues  
by a single one.

A branch point is not an isolated singularity,  
so it has no residue.  
Still, a branch of $\log z$ or of $z^a$ evaluates new real integrals  
once we indent the contour around the branch point  
or run along both sides of the branch cut.

## Limits Involving $\infty$

### The definitions

Recall that in 0918 we defined the neighborhoods
$$ \{z \in \mathbb C : |z| > R\} \cup \{\infty\} $$
of $\infty$ on the Riemann sphere.  
Replacing the disks in the definition of a limit from 1106  
by these neighborhoods gives three new kinds of limits.

We write $\lim_{z\to z_0}f(z) = \infty$  
if for every $M > 0$ there is a $\delta > 0$ such that
$$ 0 < |z - z_0| < \delta \quad\Longrightarrow\quad |f(z)| > M. $$

We write $\lim_{z\to\infty}f(z) = L$  
if for every $\varepsilon > 0$ there is an $R > 0$ such that
$$ |z| > R \quad\Longrightarrow\quad |f(z) - L| < \varepsilon, $$
and $\lim_{z\to\infty}f(z) = \infty$  
if for every $M > 0$ there is an $R > 0$  
such that $|z| > R$ implies $|f(z)| > M$.

### Back to limits at finite points

Each new limit is an ordinary limit in disguise:
$$ \begin{aligned}
\lim_{z\to z_0}f(z) = \infty &\quad\Longleftrightarrow\quad \lim_{z\to z_0}\frac{1}{f(z)} = 0, \\
\lim_{z\to\infty}f(z) = L &\quad\Longleftrightarrow\quad \lim_{w\to0}f\Bigl(\frac1w\Bigr) = L, \\
\lim_{z\to\infty}f(z) = \infty &\quad\Longleftrightarrow\quad \lim_{w\to0}\frac{1}{f(1/w)} = 0.
\end{aligned} $$

This is because $|z| > R$ exactly when $0 < |1/z| < 1/R$,  
and $|f| > M$ exactly when $|1/f| < 1/M$.

For example, let $f(z) = (3z - i)/(iz + 2)$.  
Dividing the numerator and the denominator by $z$ gives
$$ \lim_{z\to\infty}f(z) = \lim_{w\to0}\frac{3 - iw}{i + 2w} = \frac3i = -3i. $$

The denominator vanishes at $z = 2i$,  
where the numerator is $5i \ne 0$.  
Hence $1/f(z) \to 0/(5i) = 0$ as $z \to 2i$,  
and $\lim_{z\to2i}f(z) = \infty$.

For polynomials $P$ and $Q$ of degrees $n$ and $m$,  
with leading coefficients $p_n$ and $q_m$,
$$ \frac{P(z)}{Q(z)} = z^{n-m} \frac{p_n + p_{n-1}z^{-1} + \cdots + p_0z^{-n}}{q_m + q_{m-1}z^{-1} + \cdots + q_0z^{-m}}, $$
and the last fraction tends to $p_n/q_m \ne 0$ as $z \to \infty$.

Therefore
$$ \lim_{z\to\infty}\frac{P(z)}{Q(z)} = \begin{cases} 0, & n < m, \\ p_n/q_m, & n = m, \\ \infty, & n > m. \end{cases} $$
In particular, every nonconstant polynomial tends to $\infty$,  
as in the 1002 homework.

With these limits,  
a Möbius transformation $T(z) = (az + b)/(cz + d)$ with $c \ne 0$  
is continuous on $\hat{\mathbb C}$  
once we set $T(-d/c) = \infty$ and $T(\infty) = a/c$,  
as in the 1106 homework.

Some functions have no limit at $\infty$, not even $\infty$.  
For $e^z$, the positive real axis gives $e^x \to \infty$,  
the negative real axis gives $e^x \to 0$,  
and on the imaginary axis $|e^{iy}| = 1$.  
As in 1106, different paths show that the limit does not exist.

## Singularities at $\infty$

### The definition

Let $f$ be holomorphic on $|z| > R$ for some $R \ge 0$.  
Then
$$ g(w) = f\Bigl(\frac1w\Bigr) $$
is holomorphic on the punctured disk $0 < |w| < 1/R$,  
and we classify $f$ at $\infty$ by the behavior of $g$ at $0$.

- $f$ is **holomorphic at $\infty$** if $g$ has a removable singularity at $0$;  
  then we set $f(\infty) = g(0) = \lim_{z\to\infty}f(z)$.
- $f$ has a **zero of order $m$** at $\infty$ if $g$ has a zero of order $m$ at $0$.
- $f$ has a **pole of order $m$** at $\infty$ if $g$ has a pole of order $m$ at $0$.
- $f$ has an **essential singularity** at $\infty$ if $g$ has one at $0$.

### Laurent series at $\infty$

By Laurent's theorem from 1016, on $|z| > R$,
$$ f(z) = \sum_{n=-\infty}^\infty c_n z^n,\qquad\text{so}\qquad g(w) = \sum_{n=-\infty}^\infty c_n w^{-n}. $$
The principal part of $g$ at $0$ comes from the **positive** powers of $z$:  
at $\infty$, positive and negative powers exchange their roles.

| Kind at $\infty$ | Laurent series on $\lvert z\rvert > R$ | Example |
|---|---|---|
| holomorphic | no positive powers | $1 + \dfrac1z$ |
| zero of order $m$ | starts at $c_{-m}z^{-m}$, $c_{-m} \ne 0$ | $\dfrac{1}{z^2 + 1} = \dfrac{1}{z^2} - \dfrac{1}{z^4} + \cdots$ |
| pole of order $m$ | positive powers end at $c_m z^m$, $c_m \ne 0$ | $z^2 + \dfrac1z$ |
| essential | infinitely many positive powers | $e^z = \sum_{n\ge0}z^n/n!$ |

The factorization of $P/Q$ above shows that  
a rational function has a pole of order $n - m$ at $\infty$ if $n > m$,  
and a zero of order $m - n$ there if $n < m$.  
For example, $z^3/(z^2 + 1)$ has a simple pole at $\infty$,  
and $(z + 2)/(z^3 - 1)$ has a zero of order $2$ there.

Since $e^{1/z} = 1 + 1/z + 1/(2!\,z^2) + \cdots$,  
the function $e^{1/z}$ is holomorphic at $\infty$ with value $1$,  
although it has an essential singularity at $0$;  
and $ze^{1/z} = z + 1 + 1/(2z) + \cdots$ has a simple pole at $\infty$.

### Entire functions at $\infty$

**Theorem.**  
Let $f$ be a nonconstant entire function.  
Then $f$ has a pole at $\infty$ exactly when $f$ is a polynomial;  
otherwise, $f$ has an essential singularity at $\infty$.

Recall that in 1016 we saw that the Taylor series  
$f(z) = \sum_{n\ge0}a_n z^n$ of an entire function converges on all of $\mathbb C$.  
On $|z| > 0$ it is also a Laurent series, with no negative powers,  
so by uniqueness it is the Laurent series of $f$ at $\infty$.

If only finitely many $a_n$ are nonzero,  
$f$ is a polynomial of some degree $d \ge 1$,  
with a pole of order $d$ at $\infty$.  
Otherwise, infinitely many positive powers appear,  
and $\infty$ is an essential singularity.

The singularity cannot be removable,  
since then $f$ would be the constant $a_0$.  
This also follows from Liouville's theorem from the 1002 homework:  
an entire function that is holomorphic at $\infty$ is bounded.

So $e^z$, $\sin z$, and $\cos z$ have essential singularities at $\infty$.  
By the Casorati–Weierstrass theorem from 1016,  
on every region $|z| > R$  
they come arbitrarily close to every complex number.  
For instance, $\sin z = 2$ has infinitely many solutions,  
as we found in 1106.

## Non-Isolated Singularities

### Accumulating poles

A point $a$ is a **singular point** of $f$  
if $f$ is not holomorphic at $a$  
but every disk around $a$ contains points where $f$ is holomorphic.  
Recall that in 1016 we called $a$ isolated  
if $f$ is holomorphic on some punctured disk $0 < |z - a| < R$.

A singular point that is not isolated is a **non-isolated singularity**.  
There $f$ has no Laurent series on a punctured disk,  
so it has no residue,  
and the three kinds of isolated singularities do not apply to it.

For example, $\sin(\pi/z)$ vanishes at $z = 1/n$  
for every nonzero integer $n$, and these zeros are simple,  
since the derivative $-\pi z^{-2}\cos(\pi/z)$ is nonzero there.  
So $1/\sin(\pi/z)$ has simple poles at the points $1/n$,  
which accumulate at $0$.

Every punctured disk $0 < |z| < \delta$ contains a pole $1/n$,  
so $0$ is a non-isolated singularity of $1/\sin(\pi/z)$.  
In the plot below, the white poles  
crowd toward $0$ along the real axis,  
until the picture becomes too fine to resolve.

```python
var('z')
P = complex_plot(1/sin(pi/z), (-0.6, 0.6), (-0.3, 0.3), plot_points=500)
P.show(aspect_ratio=1, axes=False, frame=True)
```

The same happens at $\infty$:  
the poles $\pi/2 + k\pi$ of $\tan z$ leave every disk,  
so $\tan z$ is holomorphic on no region $|z| > R$,  
and $\infty$ is a non-isolated singularity of $\tan z$.  
Since $\tan z$ is not entire, the theorem above does not apply to it.

### Branch points

Recall that in 1002 we found $\operatorname{Log}z$ discontinuous  
at every point of the negative real axis.  
Every disk around $0$ meets this branch cut,  
so $0$ is a non-isolated singularity of $\operatorname{Log}z$.  
Moving the cut does not help:  
no branch of $\log z$ is holomorphic on a punctured disk around $0$.

Indeed, suppose $L$ is holomorphic on $0 < |z| < \delta$ and $e^{L(z)} = z$.  
Differentiating gives $L'(z)e^{L(z)} = 1$, so $L'(z) = 1/z$.  
Then $1/z$ would have an antiderivative on the punctured disk,  
and $\oint_{|z|=\delta/2}dz/z$ would be $0$, not $2\pi i$.

Going once around $0$ changes every value of $\log z$ by $2\pi i$,  
and every value of $z^a = e^{a\log z}$ by the factor $e^{2\pi ia}$.  
A point with this property is a **branch point**:  
$0$ is a branch point of $\log z$,  
and of $z^a$ for every non-integer $a$.

On the Riemann sphere,  
a large circle $|z| = R$ also goes once around $\infty$,  
so $\infty$ is a branch point as well.  
Hence every branch cut of $\log z$ must join $0$ to $\infty$.

## The Residue at $\infty$

### One residue instead of many

**Theorem.**  
Let $C$ be a positively oriented simple closed curve,  
and let $f$ be holomorphic on $\mathbb C$  
except at finitely many points inside $C$. Then
$$ \oint_C f(z)\,dz = 2\pi i\operatorname{Res}\Bigl(\frac{1}{z^2}f\Bigl(\frac1z\Bigr), 0\Bigr). $$

Choose $R$ so large that $C$ and all the singular points lie in $|z| < R$.  
By deformation of contours from 1002,  
the integral over $C$ equals the integral over the circle $|z| = R$.

On this circle, substitute $z = 1/w$.  
As $z = Re^{i\theta}$ runs counterclockwise, $w = R^{-1}e^{-i\theta}$ runs clockwise,  
and $dz = -dw/w^2$, so
$$ \oint_{|z|=R}f(z)\,dz = \oint_{|w|=1/R}f\Bigl(\frac1w\Bigr)\frac{dw}{w^2}, $$
where both circles run counterclockwise.

On and inside $|w| = 1/R$,  
the new integrand is holomorphic except at $w = 0$,  
because $f$ is holomorphic on $|z| \ge R$.  
The residue theorem with the single point $0$ proves the theorem.

### The residue at $\infty$

In terms of the Laurent series $f(z) = \sum c_n z^n$ on $|z| > R$,
$$ \frac{1}{z^2}f\Bigl(\frac1z\Bigr) = \sum_{n=-\infty}^\infty c_n z^{-n-2}, $$
and the coefficient of $z^{-1}$ comes from $n = -1$.  
So the theorem says that $\oint_C f(z)\,dz = 2\pi i c_{-1}$,  
which also follows by integrating term by term over $|z| = R$.

The **residue of $f$ at $\infty$** is defined by
$$ \operatorname{Res}(f, \infty) = -c_{-1} = -\operatorname{Res}\Bigl(\frac{1}{z^2}f\Bigl(\frac1z\Bigr), 0\Bigr). $$
The minus sign appears because a counterclockwise circle around $0$  
runs clockwise around $\infty$, as the substitution $w = 1/z$ showed.

With this sign, the theorem and the residue theorem together say  
that the residues of $f$ at its singular points $a_1, \ldots, a_k$ in $\mathbb C$  
and at $\infty$ add up to $0$:
$$ \sum_{j=1}^k\operatorname{Res}(f, a_j) + \operatorname{Res}(f, \infty) = 0. $$

**Caution.**  
The residue at $\infty$ is not determined by the kind of singularity there.  
For example, $1/z$ is holomorphic at $\infty$, with a zero of order $1$,  
but $\operatorname{Res}(1/z, \infty) = -1$.  
The factor $1/z^2$ comes from $dz = -dw/w^2$:  
the residue belongs to $f(z)\,dz$, not to $f$ alone.

### Examples

**Rational functions.**  
Let $f = P/Q$ with $\deg P = n$ and $\deg Q = m$,  
and let $C$ enclose all the zeros of $Q$.  
By the factorization of $P/Q$ above,
$$ \frac{1}{z^2}f\Bigl(\frac1z\Bigr) = z^{m-n-2} \frac{p_n + p_{n-1}z + \cdots + p_0z^n}{q_m + q_{m-1}z + \cdots + q_0z^m}. $$

If $m \ge n + 2$, this is holomorphic at $0$, so $\oint_C f(z)\,dz = 0$.  
If $m = n + 1$, it has a simple pole at $0$ with residue $p_n/q_m$, so
$$ \oint_C\frac{P(z)}{Q(z)}\,dz = 2\pi i\,\frac{p_n}{q_m}. $$

Recall that in 1120 the integral of $P/Q$ over a large semicircle  
tends to $0$ when $m \ge n + 2$.  
Over a whole circle around all the poles, it is exactly $0$.

For example, all the poles of
$$ f(z) = \frac{z^7}{(z^4 - 1)(z^4 + 2)} $$
lie on the circles $|z| = 1$ and $|z| = 2^{1/4}$, inside $|z| = 2$.

Here $m = n + 1$ and $p_n/q_m = 1$,  
so $\oint_{|z|=2}f(z)\,dz = 2\pi i$  
without computing any of the eight residues.

**An essential singularity inside.**  
Let $f(z) = e^{1/z}/(z - 1)$.  
Inside $|z| = 2$, the residue at $1$ is $e$.

At the essential singularity $0$,
$$ \frac{e^{1/z}}{z - 1} = -\Bigl(1 + \frac1z + \frac{1}{2!\,z^2} + \cdots\Bigr)(1 + z + z^2 + \cdots), $$
and the coefficient of $z^{-1}$ is the series  
$-(1/1! + 1/2! + \cdots) = -(e - 1)$.

The residue at $\infty$ avoids this series:
$$ \frac{1}{z^2}f\Bigl(\frac1z\Bigr) = \frac{1}{z^2} \cdot \frac{e^z}{1/z - 1} = \frac{e^z}{z(1 - z)}, $$
whose residue at $0$ is $1$.  
Both ways, $\oint_{|z|=2}f(z)\,dz = 2\pi i\bigl(e - (e - 1)\bigr) = 2\pi i$.

## Branch Cuts in Real Integrals

### Logarithms and powers, revisited

Recall from 1002 and 1106 that  
$\log z = \{\ln|z| + i(\operatorname{Arg}z + 2\pi k) : k \in \mathbb Z\}$  
and $z^a = \{e^{a\ell} : \ell \in \log z\}$.  
For instance, $z^{1/n} = \exp\bigl((\log z)/n\bigr)$  
gives exactly the $n$ roots from 1106,  
since $\exp\bigl((\ell + 2\pi ik)/n\bigr)$ depends only on $k$ modulo $n$.

As sets of values,
$$ \log(z_1z_2) = \log z_1 + \log z_2, $$
where the right side means all sums $\ell_1 + \ell_2$.  
Indeed, $e^{\ell_1+\ell_2} = z_1z_2$;  
conversely, if $e^\ell = z_1z_2$ and $\ell_1 \in \log z_1$, then $\ell - \ell_1 \in \log z_2$.

For principal values, the identity can fail by $2\pi i$:
$$ \begin{aligned}
\operatorname{Log}\bigl((-1)(-1)\bigr) &= \operatorname{Log}1 = 0, \\
\operatorname{Log}(-1) + \operatorname{Log}(-1) &= 2\pi i.
\end{aligned} $$

In general, $\operatorname{Log}(z_1z_2) = \operatorname{Log}z_1 + \operatorname{Log}z_2$  
exactly when $\operatorname{Arg}z_1 + \operatorname{Arg}z_2 \in (-\pi, \pi]$.  
The same lost $2\pi$ made  
$\bigl((-1)(-1)\bigr)^{1/2} \ne (-1)^{1/2}(-1)^{1/2}$ in 1106.

Likewise, $2\log z$ and $\log(z^2)$ differ in general:  
$2\log i = \{(4k + 1)\pi i : k \in \mathbb Z\}$  
is only half of $\log(-1) = \{(2k + 1)\pi i : k \in \mathbb Z\}$.

### Choosing the branch cut

For a real integral over $(0, \infty)$ involving $\ln x$ or $x^a$,  
we first choose a branch $L(z) = \ln|z| + i\Theta(z)$ of $\log z$.  
Recall from 1106 that $L$ is holomorphic off its cut  
with $L'(z) = 1/z$, and that $e^{aL(z)}$ is a holomorphic branch of $z^a$.

The cut must avoid the contour, or run along part of it on purpose.  
We use two branches:

- $-\pi/2 < \Theta(z) < 3\pi/2$,  
  with the cut along the negative imaginary axis,  
  so that $L(x) = \ln x$ and $L(-x) = \ln x + i\pi$ for $x > 0$;
- $0 < \Theta(z) < 2\pi$,  
  with the cut along the positive real axis.

### Indenting around a branch point

Let $a > 0$.  
We evaluate
$$ I = \int_0^\infty\frac{\ln x}{x^2 + a^2}\,dx. $$
It converges at $0$ because $\int_0^1|\ln x|\,dx = 1$,  
and at $\infty$ because $\ln x \le \sqrt x$, so $\ln(x)/x^2 \le x^{-3/2}$.

Let $f(z) = L(z)/(z^2 + a^2)$ with the first branch,  
and integrate around the indented contour from 1120:  
the segments $[-R, -r]$ and $[r, R]$, the big semicircle $C_R$,  
and the small semicircle $C_r$, run clockwise.

```python
R, r, a = 3, 0.6, 2
G = line([(-R, 0), (-r, 0)], color='blue', thickness=2) + line([(r, 0), (R, 0)], color='blue', thickness=2)
G += arc((0, 0), R, sector=(0, pi), color='blue', thickness=2)
G += arc((0, 0), r, sector=(0, pi), color='red', thickness=2)
G += line([(0, 0), (0, -R)], color='gray', linestyle='--', thickness=2)  # the branch cut
G += point((0, a), size=40, color='black', zorder=10)
G += text('$C_R$', (2.5, 2.3), fontsize=14, color='blue') + text('$C_r$', (0.75, 0.65), fontsize=14, color='red')
G.show(aspect_ratio=1, axes=False)
```

The dashed cut lies outside the contour,  
and the only singular point inside is the pole $ia$, shown for $a = 2$.  
The small semicircle keeps the branch point $0$ outside as well.

Since $L(ia) = \ln a + i\pi/2$,
$$ \operatorname{Res}(f, ia) = \frac{L(ia)}{2ia} = \frac{\ln a + i\pi/2}{2ia}. $$

On the negative real axis, we substitute $x = -t$  
and use $L(-t) = \ln t + i\pi$:
$$ \int_{-R}^{-r}f(x)\,dx = \int_r^R\frac{\ln t + i\pi}{t^2 + a^2}\,dt. $$

Orient $C_r$ counterclockwise as in 1120.  
Then the residue theorem gives
$$ \begin{aligned}
&2\int_r^R\frac{\ln x}{x^2 + a^2}\,dx + i\pi\int_r^R\frac{dx}{x^2 + a^2} \\
&\quad + \int_{C_R}f(z)\,dz - \int_{C_r}f(z)\,dz \\
&\quad = 2\pi i \cdot \frac{\ln a + i\pi/2}{2ia} = \frac{\pi\ln a}{a} + \frac{i\pi^2}{2a}.
\end{aligned} $$

On $C_R$, $|L(z)| \le \ln R + \pi$ and $|z^2 + a^2| \ge R^2 - a^2$, so
$$ \Bigl|\int_{C_R}f(z)\,dz\Bigr| \le \frac{\pi R(\ln R + \pi)}{R^2 - a^2} \to 0\qquad(R \to \infty). $$

On $C_r$, $|L(z)| \le |\ln r| + \pi$ and $|z^2 + a^2| \ge a^2 - r^2$, so
$$ \Bigl|\int_{C_r}f(z)\,dz\Bigr| \le \frac{\pi r(|\ln r| + \pi)}{a^2 - r^2} \to 0\qquad(r \to 0), $$
because $r\ln r \to 0$.

Letting $r \to 0$ and $R \to \infty$ and comparing real parts,
$$ \int_0^\infty\frac{\ln x}{x^2 + a^2}\,dx = \frac{\pi\ln a}{2a}. $$
The imaginary parts give $\pi\int_0^\infty dx/(x^2 + a^2) = \pi^2/(2a)$,  
which we already knew from $\arctan$.

For $a = 1$ the integral is $0$,  
as the substitution $x = 1/t$ confirms:  
it turns the integral into its own negative.

Unlike a pole, the branch point contributes nothing in the limit.  
On $C_r$, $|f|$ grows only like $|\ln r|$, slower than $1/r$,  
so the length $\pi r$ of $C_r$ wins.

The semicircle works here because $x^2 + a^2$ is even:  
the negative real axis returns the same integral plus a known one.  
For other rational functions, the next contour is better suited.

### Integrating along a branch cut

Let $0 < a < 1$.  
We show that
$$ \int_0^\infty\frac{x^{-a}}{x + 1}\,dx = \frac{\pi}{\sin(a\pi)}. $$
It converges at $0$ because $a < 1$, and at $\infty$ because $a > 0$.

Use the second branch, $0 < \Theta(z) < 2\pi$, and let
$$ f(z) = \frac{e^{-aL(z)}}{z + 1}. $$
This branch of $z^{-a}/(z + 1)$ is holomorphic  
off the positive real axis, except at the pole $-1$.

The **keyhole contour** consists of  
the upper edge of the cut from $r$ to $R$,  
the circle $C_R$ counterclockwise,  
the lower edge of the cut from $R$ back to $r$,  
and the circle $C_r$ clockwise.

```python
R, r, d = 3, 0.6, 0.15
s, t = arcsin(d/R), arcsin(d/r)
G = line([(r*cos(t), d), (R*cos(s), d)], color='blue', thickness=2)
G += line([(r*cos(t), -d), (R*cos(s), -d)], color='blue', thickness=2)
G += arc((0, 0), R, sector=(s, 2*pi - s), color='blue', thickness=2)
G += arc((0, 0), r, sector=(t, 2*pi - t), color='red', thickness=2)
G += line([(0, 0), (R + 0.5, 0)], color='gray', linestyle='--', thickness=2)  # the branch cut
G += arrow((1.4, d), (2, d), color='blue', width=2) + arrow((2, -d), (1.4, -d), color='blue', width=2)
G += point((-1, 0), size=40, color='black', zorder=10)
G += text('$C_R$', (-2.3, 2.3), fontsize=14, color='blue') + text('$C_r$', (-0.5, 0.65), fontsize=14, color='red')
G.show(aspect_ratio=1, axes=False)
```

In the picture the two edges are drawn slightly apart from the cut.  
Strictly, the straight parts run along the rays  
$\arg z = \delta$ and $\arg z = 2\pi - \delta$,  
and we let $\delta \to 0$ at the end;  
the integrands converge uniformly for $r \le |z| \le R$.

On the upper edge, $\Theta \to 0$ and $f(x) \to x^{-a}/(x + 1)$.  
On the lower edge, $\Theta \to 2\pi$ and
$$ f(x) \to \frac{x^{-a}e^{-2\pi ia}}{x + 1}. $$
The two edges do not cancel,  
because $z^{-a}$ jumps by the factor $e^{-2\pi ia}$ across the cut.

The only pole inside is $-1 = e^{i\pi}$, with
$$ \operatorname{Res}(f, -1) = e^{-aL(-1)} = e^{-i\pi a}. $$

Write $I_{r,R} = \int_r^R x^{-a}/(x + 1)\,dx$,  
and orient $C_r$ counterclockwise.  
Then the residue theorem gives
$$ (1 - e^{-2\pi ia})I_{r,R} + \int_{C_R}f(z)\,dz - \int_{C_r}f(z)\,dz = 2\pi ie^{-i\pi a}. $$

Since $|e^{-aL(z)}| = |z|^{-a}$,
$$ \begin{aligned}
\Bigl|\int_{C_R}f(z)\,dz\Bigr| &\le 2\pi R \cdot \frac{R^{-a}}{R - 1} = \frac{2\pi R^{1-a}}{R - 1} \to 0, \\
\Bigl|\int_{C_r}f(z)\,dz\Bigr| &\le 2\pi r \cdot \frac{r^{-a}}{1 - r} = \frac{2\pi r^{1-a}}{1 - r} \to 0,
\end{aligned} $$
the first because $a > 0$ and the second because $a < 1$.

Hence $(1 - e^{-2\pi ia})I = 2\pi ie^{-i\pi a}$.  
Multiplying by $e^{i\pi a}$,
$$ I = \frac{2\pi i}{e^{i\pi a} - e^{-i\pi a}} = \frac{2\pi i}{2i\sin(a\pi)} = \frac{\pi}{\sin(a\pi)}. $$

For $a = 1/2$, the substitution $x = t^2$ confirms it:  
$\int_0^\infty dx/\bigl(\sqrt x(x + 1)\bigr) = 2\int_0^\infty dt/(t^2 + 1) = \pi$.

More generally, substituting $x = t^n$ with $a = 1 - 1/n$ gives
$$ \int_0^\infty\frac{dt}{t^n + 1} = \frac1n\int_0^\infty\frac{x^{-a}}{x + 1}\,dx = \frac{\pi}{n\sin(\pi/n)}, $$
the value found with a sector in 1120.  
The keyhole gives such integrals for every real exponent at once.

### Summary

| Tool | When | Key fact |
|---|---|---|
| residue at $\infty$ | all singular points inside $C$ | $\oint_C f(z)\,dz = 2\pi i\operatorname{Res}\bigl(z^{-2}f(1/z), 0\bigr)$ |
| indented semicircle | $\ln x$ or $x^a$ times an even function | cut on the negative imaginary axis; $C_r$ contributes $0$ |
| keyhole | $x^{-a}$ times a rational function | cut on the positive real axis; the lower edge gains $e^{-2\pi ia}$ |

On $|z| = 2$, write $z = 2e^{it}$, so $dz = iz\,dt$ and
$$ \frac{1}{2\pi i}\oint_{|z|=2}f(z)\,dz = \frac{1}{2\pi}\int_0^{2\pi}zf(z)\,dt. $$
For both examples, the cell below prints  
the real and imaginary parts of the right side;  
we expect $1$ and $0$.

```python
def contour_average(f):
    zf = lambda t: CDF(2*cos(t), 2*sin(t)) * f(CDF(2*cos(t), 2*sin(t))) / (2*RDF(pi))
    return numerical_integral(lambda t: zf(t).real(), 0, 2*pi)[0], numerical_integral(lambda t: zf(t).imag(), 0, 2*pi)[0]

print(contour_average(lambda z: z^7/((z^4 - 1)*(z^4 + 2))))
print(contour_average(lambda z: exp(1/z)/(z - 1)))
```
