# ENGINEERING MATH - COMPLEX VARIABLE

Today we collect topics that come before the residue theorem  
but were skipped in earlier lectures.  
They fall into two groups.

The first group is about the functions themselves:  
limits, on which continuity and the complex derivative rest;  
roots and general powers,  
together with how $z^n$, $z^{1/n}$, and $1/z$ map regions;  
and the inverse trigonometric and hyperbolic functions,  
which, like general powers, are built from the logarithm.

The second group completes the theory of contour integrals:  
a proof of the Cauchy–Goursat theorem  
without assuming that $f'$ is continuous;  
its converse, Morera's theorem;  
and harmonic functions and the maximum modulus principle,  
both resting on the Cauchy integral formula.

## Limits of Complex Functions

### The definition

Let $f$ be defined near $z_0$, except possibly at $z_0$ itself.  
We write $\lim_{z\to z_0}f(z) = L$ if for every $\varepsilon > 0$ there is a $\delta > 0$ such that
$$ 0 < |z - z_0| < \delta \quad\Longrightarrow\quad |f(z) - L| < \varepsilon. $$
The input may approach $z_0$ from every direction in the plane.  
The value $f(z_0)$ is irrelevant to this limit.

The sequential criterion is equivalent to the $\varepsilon$–$\delta$ definition:
$$ \lim_{z\to z_0}f(z) = L $$
if and only if $f(z_n) \to L$ for every sequence $z_n \to z_0$ with $z_n \ne z_0$.

If $f = u + iv$ and $L = A + iB$, then
$$ f(z) \to L \quad\Longleftrightarrow\quad u(x, y) \to A \text{ and } v(x, y) \to B. $$
Thus complex limits are exactly  
two-variable vector limits written in complex notation.

### Limit laws

If $f(z) \to A$ and $g(z) \to B$ as $z \to z_0$, then
$$ f + g \to A + B,\qquad cf \to cA,\qquad fg \to AB. $$
If $B \ne 0$, then $f/g \to A/B$.  
Also $|f(z)| \to |A|$, $\overline{f(z)} \to \overline A$,  
and every polynomial is continuous.

Consequently, a rational function $p(z)/q(z)$  
is continuous wherever $q(z) \ne 0$.  
These statements follow from  
the triangle inequality and $|zw| = |z||w|$.

For example, since $2 + i \ne 0$, the limit laws give
$$ \lim_{z\to i}\frac{z^3 + 2z}{z + 2} = \frac{i^3 + 2i}{i + 2} = \frac{i}{2 + i} = \frac{1 + 2i}{5}. $$
Direct substitution is justified here by continuity;  
it is not a separate rule that can be used at a zero of the denominator.

### Detecting nonexistence with paths

If a two-dimensional limit exists,  
every path approaching the point gives the same limit.  
Therefore two paths with different limiting values  
prove that the limit does not exist.

For $f(z) = \bar z/z$ at $0$, approaching along the real axis gives
$$ \frac{\bar x}{x} = 1, $$
while approaching along the imaginary axis gives
$$ \frac{\overline{iy}}{iy} = -1. $$
Hence $\lim_{z\to0}\bar z/z$ does not exist.

Checking only straight lines is not enough.  
For
$$ F(x + iy) = \frac{x^2y}{x^4 + y^2}, $$
every line $y = mx$ gives $F(x + imx) \to 0$.

Along the parabola $y = x^2$, however, $F = 1/2$.  
Thus the limit at $0$ does not exist.  
Paths disprove a limit;  
agreement along a collection of paths does not prove one.

## Powers, Roots, and Elementary Mappings

### The $n$th roots of a complex number

Let $n \ge 1$ and $z = re^{i\theta} \ne 0$.  
Solving $w^n = z$ gives exactly $n$ values:
$$ w_k = r^{1/n}e^{i(\theta+2\pi k)/n},\qquad k = 0, 1, \ldots, n - 1. $$

Changing $\theta$ by $2\pi$ merely cycles through the same set.  
All roots have modulus $r^{1/n}$,  
and consecutive roots differ in argument by $2\pi/n$;  
they are the vertices of a regular $n$-gon centered at the origin.  
The only $n$th root of $0$ is $0$.

For example, $32i = 32e^{i\pi/2}$.  
Its fifth roots are
$$ w_k = 2e^{i(\pi/10+2\pi k/5)},\qquad k = 0, 1, 2, 3, 4. $$
The factor $2$ is the fifth root of $32$,  
and multiplying one root by $e^{2\pi i/5}$ produces the next one.

### The map $w = z^n$

In polar coordinates,
$$ z = re^{i\theta} \quad\longmapsto\quad w = r^n e^{in\theta}. $$
The map raises every radius to the $n$th power  
and multiplies every angle by $n$.

It maps each circle $|z| = r$ onto $|w| = r^n$  
and each ray $\arg z = \theta$ onto $\arg w = n\theta$ modulo $2\pi$.  
Every nonzero $w$ has $n$ preimages,  
related by multiplication by the $n$th roots of unity.

Radial inequalities transform without ambiguity:
$$ \{z : a < |z| < b\} \longmapsto \{w : a^n < |w| < b^n\}. $$
Conversely,
$$ (z^n)^{-1}\bigl(\{w : A < |w| < B\}\bigr) = \{z : A^{1/n} < |z| < B^{1/n}\}. $$

The same rule handles disks, punctured disks, exteriors, circles, and annuli.  
Boundary inclusion is preserved because  
$r \mapsto r^n$ is strictly increasing for $r \ge 0$.

For a sector $a < r < b$, $\alpha < \theta < \beta$, the image under $z^n$ has
$$ a^n < |w| < b^n,\qquad n\alpha < \arg w < n\beta, $$
with angles interpreted modulo $2\pi$.

If $\beta - \alpha \le 2\pi/n$,  
the angle interval does not wrap more than once,  
and $z^n$ is one-to-one on the sector  
(after one boundary ray is omitted when necessary).

For example, $z^2$ maps the first quadrant  
one-to-one onto the upper half-plane.

Suppose $0 < \delta - \gamma < 2\pi$.  
The preimage of the sector $\gamma < \arg w < \delta$  
under $w = z^n$ is the union of $n$ sectors
$$ \frac{\gamma + 2\pi k}{n} < \arg z < \frac{\delta + 2\pi k}{n},\qquad k = 0, 1, \ldots, n - 1. $$

Add the corresponding condition $A^{1/n} < |z| < B^{1/n}$  
when the target also has radial bounds.  
This description separates the $n$ inverse sheets  
instead of hiding them in a multivalued symbol.

### Branches of the root map

Let $D$ be a region on which  
a continuous argument $\Theta(z)$ has been chosen.  
Then the $n$ functions
$$ R_k(z) = |z|^{1/n}e^{i(\Theta(z)+2\pi k)/n},\qquad k = 0, 1, \ldots, n - 1, $$
are the branches of $z^{1/n}$ on $D$, and $R_k(z)^n = z$.

On a logarithm branch $L(z) = \ln|z| + i\Theta(z)$,  
the same formula is
$$ R_k(z) = e^{(L(z)+2\pi i k)/n}. $$
Each branch is holomorphic and $R_k'(z) = 1/(nR_k(z)^{n-1})$.

Using $\operatorname{Arg}z \in (-\pi, \pi)$ on $\mathbb C \setminus (-\infty, 0]$ gives the principal branch
$$ z^{1/n} = \exp\Bigl(\frac{\operatorname{Log}z}{n}\Bigr),\qquad -\frac{\pi}{n} < \arg z^{1/n} < \frac{\pi}{n}. $$
For $n = 2$, this branch takes values in the right half-plane.

A branch cut is not part of the algebraic equation $w^n = z$;  
it is the boundary removed  
so that one root can be selected continuously and holomorphically.

On a branch with $\gamma < \Theta(z) < \delta$, the image of
$$ A < |z| < B,\qquad \gamma < \Theta(z) < \delta, $$
under $R_k$ is
$$ A^{1/n} < |w| < B^{1/n},\qquad \frac{\gamma + 2\pi k}{n} < \arg w < \frac{\delta + 2\pi k}{n}. $$

A preimage under $R_k$ is found  
by intersecting the target set with the range of $R_k$,  
then applying $w \mapsto w^n$.  
For example, the principal square root maps the slit plane  
one-to-one onto the right half-plane,  
and the preimage of the first quadrant is the upper half-plane.

### The map $w = 1/z$

In polar coordinates,
$$ z = re^{i\theta} \quad\longmapsto\quad w = \frac1r e^{-i\theta}. $$
Thus inversion reciprocates radii and reverses angles.

It interchanges the inside and outside of the unit circle,  
maps $|z| = r$ to $|w| = 1/r$,  
and maps the sector $\alpha < \arg z < \beta$ to $-\beta < \arg w < -\alpha$.  
Because $1/(1/z) = z$,  
image and preimage questions use the same rule.

A circle or line can be written in the unified form
$$ A|z|^2 + Bz + \overline B\bar z + C = 0, $$
where $A, C \in \mathbb R$.

Substituting $z = 1/w$ and multiplying by $|w|^2$ gives
$$ C|w|^2 + B\bar w + \overline B w + A = 0. $$

Therefore inversion maps circles and lines to circles and lines.  
A circle through $0$ becomes a line,  
and a line not through $0$ becomes a circle through $0$.

For example, let $a > 0$ and write $w = u + iv$.  
Since $z = 1/w = (u - iv)/|w|^2$,  
the half-plane $\operatorname{Re}z > a$ becomes
$$ \frac{u}{u^2 + v^2} > a \quad\Longleftrightarrow\quad \Bigl(u - \frac{1}{2a}\Bigr)^2 + v^2 < \frac{1}{4a^2}. $$

Its image is the open disk with diameter from $0$ to $1/a$.  
Testing one point decides  
which side of a transformed boundary is the image.

For images and preimages under $z^n$, $z^{1/n}$, and $1/z$,  
use the same workflow:

1. write the region using radial and angular inequalities,  
   or an equation for its boundary;
2. transform the boundary and the inequalities separately;
3. count the inverse sheets and test an interior point.

The root symbol needs a stated branch when it denotes a function.  
Without a branch, $z^{1/n}$ denotes all $n$ roots  
and its image or preimage is a relation, not a single-valued map.

## General Complex Powers

### A fixed exponent

For $z \ne 0$ and $\alpha \in \mathbb C$, the multivalued power is
$$ z^\alpha = \{e^{\alpha\ell} : \ell \in \log z\} = \{e^{\alpha(\ln|z|+i(\theta+2\pi k))} : k \in \mathbb Z\}. $$

Successive values differ by the factor $e^{2\pi i\alpha}$.  
If $\alpha$ is an integer, all values coincide.  
If $\alpha = p/q$ in lowest terms, there are exactly $q$ values.  
For every other exponent, there are infinitely many values.

Choose a logarithm branch $L$ on a region $D$.  
The corresponding single-valued branch of the power is
$$ z_L^\alpha = e^{\alpha L(z)}, $$
which is holomorphic on $D$ and satisfies
$$ \frac{d}{dz} z_L^\alpha = \alpha e^{(\alpha-1)L(z)} = \frac{\alpha z_L^\alpha}{z}. $$

For integer $\alpha$, this agrees with the usual power  
and does not depend on $L$.  
For $\alpha = 1/n$, it gives one of the root branches above.

On one fixed logarithm branch,
$$ z_L^{\alpha+\beta} = z_L^\alpha z_L^\beta. $$
Laws involving different bases require compatible logarithms  
and can fail for principal values.

For example, using $\operatorname{Arg}z \in (-\pi, \pi]$,
$$ \bigl((-1) \cdot (-1)\bigr)^{1/2} = 1,\qquad (-1)^{1/2} (-1)^{1/2} = i^2 = -1. $$
The lost sign records the $2\pi$ jump in the arguments.

### A complex base and a complex exponent

For $z \ne 0$ and $w \in \mathbb C$, define the multivalued expression
$$ z^w = \{e^{w\ell} : \ell \in \log z\}. $$

After choosing a logarithm branch $L$ in the $z$-variable,
$$ P(z, w) = e^{wL(z)} $$
is single-valued and holomorphic in both variables, with
$$ \frac{\partial P}{\partial z} = \frac{wP}{z},\qquad \frac{\partial P}{\partial w} = L(z)P. $$

The branch choice can change even the size:  
$1^i = \{e^{-2\pi k} : k \in \mathbb Z\}$.

## Inverse Trigonometric and Hyperbolic Functions

The exponential is periodic,  
so the trigonometric and hyperbolic functions built from it  
are not one-to-one on $\mathbb C$.  
Their inverses are therefore multivalued relations.

A single-valued inverse function  
requires compatible branches of the logarithm and square root.  
Different choices give different inverse branches,  
just as they do for $z^{1/n}$.

### Inverse trigonometric functions

To solve $\sin z = \zeta$, set $y = e^{iz}$.  
The equation $(y - y^{-1})/(2i) = \zeta$ becomes
$$ y^2 - 2i\zeta y - 1 = 0, $$
so $y = i\zeta \pm \sqrt{1 - \zeta^2}$.

Taking logarithms and using $z = -i\log y$ gives
$$ \arcsin\zeta = -i\log\bigl(i\zeta + \sqrt{1 - \zeta^2}\bigr), $$
where both $\log$ and $\sqrt{\phantom z}$ are multivalued until branches are chosen.

The other inverse trigonometric functions can be written as
$$ \begin{aligned}
\arccos\zeta &= \frac\pi2 - \arcsin\zeta, \\
\arctan\zeta &= \frac{1}{2i}\bigl(\log(1 + i\zeta) - \log(1 - i\zeta)\bigr).
\end{aligned} $$

With compatible branches, these satisfy
$$ \begin{aligned}
\sin(\arcsin\zeta) &= \zeta, \\
\cos(\arccos\zeta) &= \zeta, \\
\tan(\arctan\zeta) &= \zeta.
\end{aligned} $$

The reverse compositions need not return the original input  
because of periods and symmetries.

If $a$ is one value of $\arcsin\zeta$, all solutions of $\sin z = \zeta$ are
$$ z = a + 2\pi k\quad\text{or}\quad z = \pi - a + 2\pi k,\qquad k \in \mathbb Z. $$

For instance, $\sin z = 2$ has the solutions
$$ z = \frac\pi2 + 2\pi k \pm i\ln(2 + \sqrt3),\qquad k \in \mathbb Z, $$
because $\sin(\pi/2 + iy) = \cosh y$.

Values outside $[-1, 1]$ cause no problem  
once complex inputs are allowed.

### Inverse hyperbolic functions

Solving the exponential definitions in the same way gives
$$ \begin{aligned}
\operatorname{arsinh}\zeta &= \log\bigl(\zeta + \sqrt{\zeta^2 + 1}\bigr), \\
\operatorname{arcosh}\zeta &= \log\bigl(\zeta + \sqrt{\zeta - 1} \sqrt{\zeta + 1}\bigr), \\
\operatorname{artanh}\zeta &= \frac12\bigl(\log(1 + \zeta) - \log(1 - \zeta)\bigr).
\end{aligned} $$

These formulas represent multivalued inverses  
unless all logarithm and square-root branches are specified.

Common principal branches are holomorphic on slit planes.  
For $\arcsin$ and $\arccos$, remove $(-\infty, -1] \cup [1, \infty)$.  
For $\arctan$ and $\operatorname{arsinh}$,  
remove the two imaginary rays from $i$ and $-i$ to infinity.  
For $\operatorname{arcosh}$, remove $(-\infty, 1]$;  
for $\operatorname{artanh}$, remove $(-\infty, -1] \cup [1, \infty)$.

Other cuts are possible.  
The branch points occur  
where the original function has a critical value  
or where a logarithm argument vanishes.

On any compatible branches, implicit differentiation gives
$$ \begin{aligned}
(\arcsin z)' &= \frac{1}{\sqrt{1 - z^2}}, \\
(\arccos z)' &= -\frac{1}{\sqrt{1 - z^2}}, \\
(\arctan z)' &= \frac{1}{1 + z^2}.
\end{aligned} $$

Likewise, for the inverse hyperbolic functions,
$$ \begin{aligned}
(\operatorname{arsinh}z)' &= \frac{1}{\sqrt{1 + z^2}}, \\
(\operatorname{arcosh}z)' &= \frac{1}{\sqrt{z - 1} \sqrt{z + 1}}, \\
(\operatorname{artanh}z)' &= \frac{1}{1 - z^2}.
\end{aligned} $$

The square roots in the derivative formulas  
must use the same branches as the inverse functions.

## The Cauchy–Goursat Theorem

Recall that in 1002 we proved Cauchy's theorem  
from Green's theorem,  
assuming that the real and imaginary parts of $f$  
have continuous first partial derivatives.  
That is stronger than holomorphy,  
which only says that $f'(z)$ exists at every point.

Goursat's argument removes the extra continuity assumption.  
Its key step proves the theorem on a triangle  
by repeatedly subdividing the triangle.

### The triangle theorem

**Triangle theorem.**  
If $f$ is holomorphic on an open set  
containing a triangle $T$ and its interior, then
$$ \oint_{\partial T}f(z)\,dz = 0. $$
No continuity of $f'$ is assumed.

Write $I(T) = \oint_{\partial T}f(z)\,dz$.  
Join the midpoints of the three sides  
to divide $T$ into four congruent triangles.

The integrals along the new interior edges  
cancel in opposite directions, so
$$ I(T) = I(T_1) + I(T_2) + I(T_3) + I(T_4). $$

At least one subtriangle, call it $T_1$, satisfies $|I(T_1)| \ge |I(T)|/4$.  
Repeat the construction inside the selected triangle.

This produces nested closed triangles
$$ T \supset T_1 \supset T_2 \supset \cdots $$
such that
$$ |I(T_n)| \ge \frac{|I(T)|}{4^n}. $$

If $d$ and $P$ are the diameter and perimeter of $T$,  
then $T_n$ has diameter $d/2^n$ and perimeter $P/2^n$.  
The diameters tend to $0$,  
so the nested triangles have exactly one common point, say $z_*$.

Complex differentiability at $z_*$ means that
$$ f(z) = f(z_*) + f'(z_*)(z - z_*) + (z - z_*)\eta(z), $$
where $\eta(z) \to 0$ as $z \to z_*$.

The integrals of the constant and linear terms around $\partial T_n$ vanish.  
Given $\varepsilon > 0$, for all sufficiently large $n$ we have $|\eta(z)| < \varepsilon$ on $T_n$.  
The ML-inequality then gives
$$ |I(T_n)| \le \varepsilon\frac{d}{2^n}\frac{P}{2^n} = \frac{\varepsilon dP}{4^n}. $$

Combining the lower and upper bounds gives
$$ \frac{|I(T)|}{4^n} \le |I(T_n)| \le \frac{\varepsilon dP}{4^n}, $$
hence $|I(T)| \le \varepsilon dP$.  
Since $\varepsilon$ is arbitrary, $I(T) = 0$.

The proof used $f'(z_*)$ at one point but never used continuity of $f'$.

### From triangles to closed contours

A polygonal region can be triangulated.  
Adding the triangle integrals cancels every interior edge,  
leaving only the boundary integral.  
Local polygonal approximation  
gives the corresponding result for piecewise smooth boundaries.

Therefore, if $f$ is holomorphic on and inside a closed contour $C$, then
$$ \oint_C f(z)\,dz = 0. $$

Equivalently, if $f$ is holomorphic on a simply connected region $U$,  
its integral around every closed contour in $U$ is zero.  
This is the Cauchy–Goursat theorem  
in the form used throughout the course.

## Morera's Theorem

**Morera's theorem.**  
Let $f$ be continuous on a region $U$.  
If
$$ \oint_{\partial T}f(z)\,dz = 0 $$
for every triangle $T$ whose interior and boundary lie in $U$,  
then $f$ is holomorphic on $U$.

Thus continuity plus vanishing local contour integrals  
forces complex differentiability.

Fix a disk $D \subset U$ and a point $a \in D$.  
For $z \in D$, define
$$ F(z) = \int_{[a,z]} f(\zeta)\,d\zeta, $$
where $[a,z]$ is the straight segment from $a$ to $z$.

For small $h$, the triangle with vertices $a$, $z$, and $z + h$ lies in $D$.  
Its integral is zero, so
$$ F(z + h) - F(z) = \int_{[z,z+h]}f(\zeta)\,d\zeta. $$

Parameterizing the last segment by $\zeta = z + th$ gives
$$ \frac{F(z + h) - F(z)}{h} = \int_0^1 f(z + th)\,dt. $$
Continuity of $f$ makes the right-hand side tend to $f(z)$ as $h \to 0$.  
Hence $F'(z) = f(z)$, so $F$ is holomorphic.

A holomorphic function is analytic,  
and its derivative is holomorphic;  
therefore $f = F'$ is holomorphic on $D$.  
Since every point of $U$ lies in such a disk, $f$ is holomorphic on $U$.

A useful consequence concerns limits of holomorphic functions.  
Suppose $f_n$ is holomorphic on $U$  
and $f_n \to f$ uniformly on every compact subset of $U$.  
Then $f$ is continuous.

For every triangle $T$ compactly contained in $U$,  
uniform convergence on $\partial T$ gives
$$ \oint_{\partial T}f(z)\,dz = \lim_{n\to\infty}\oint_{\partial T}f_n(z)\,dz = 0. $$

Morera's theorem shows that $f$ is holomorphic.  
Ordinary pointwise convergence would not be enough.

## Harmonic Functions

The real and imaginary parts of a holomorphic function  
are far from arbitrary functions of two real variables.

They satisfy Laplace's equation,  
the equation of steady-state temperature,  
of electrostatic potential in a charge-free region,  
and of the velocity potential of an ideal two-dimensional flow.  
This is the bridge from complex analysis to these physical problems.

### Laplace's equation

A real function $u(x, y)$ with continuous second partial derivatives  
on an open set $U \subset \mathbb R^2$  
is **harmonic** if it satisfies **Laplace's equation**
$$ \Delta u = u_{xx} + u_{yy} = 0\qquad\text{on }U. $$
The operator $\Delta = \partial^2/\partial x^2 + \partial^2/\partial y^2$ is called the **Laplacian**.

For example, the following functions are harmonic:

- $x^2 - y^2$ and $xy$, since $2 - 2 = 0$ and $0 + 0 = 0$;
- $e^x\cos y$, since $e^x\cos y - e^x\cos y = 0$; and
- $\ln(x^2 + y^2)$ on $\mathbb R^2 \setminus \{0\}$, since  
  $u_{xx} = \dfrac{2(y^2 - x^2)}{(x^2 + y^2)^2}$ and $u_{yy} = \dfrac{2(x^2 - y^2)}{(x^2 + y^2)^2}$.

By contrast, $x^2 + y^2$ is not harmonic: its Laplacian is $4$.

### Real and imaginary parts of holomorphic functions

**Theorem.**  
If $f = u + iv$ is holomorphic on $U$,  
then $u$ and $v$ are harmonic on $U$.

Recall that in 1016 we proved Taylor's theorem,  
so $f$ has derivatives of all orders,  
and $u$ and $v$ have continuous partial derivatives of all orders.

Differentiating the Cauchy–Riemann equations  
$u_x = v_y$ and $u_y = -v_x$ gives
$$ u_{xx} = v_{yx},\qquad u_{yy} = -v_{xy}, $$
and mixed partial derivatives are equal,  
so $u_{xx} + u_{yy} = v_{yx} - v_{xy} = 0$.  
In the same way, $v_{xx} + v_{yy} = -u_{yx} + u_{xy} = 0$.

Every holomorphic function we know produces harmonic functions:
$$ \begin{aligned}
z^2 &= (x^2 - y^2) + i\,2xy, \\
e^z &= e^x\cos y + i\,e^x\sin y, \\
\operatorname{Log}z &= \tfrac12\ln(x^2 + y^2) + i\operatorname{Arg}z.
\end{aligned} $$
This explains the examples above.

The function $\ln(x^2 + y^2) = 2\ln|z|$  
is harmonic on all of $\mathbb C \setminus \{0\}$,  
even across the branch cut of $\operatorname{Log}$:  
harmonicity is a local property, and near the cut,  
$\ln|z|$ is the real part of another branch of the logarithm.

### Harmonic conjugates

Conversely, given a harmonic function $u$,  
can we find $v$ such that $u + iv$ is holomorphic?  
Such a $v$ is called a **harmonic conjugate** of $u$.  
By the Cauchy–Riemann equations, it must satisfy
$$ v_x = -u_y,\qquad v_y = u_x. $$

If $v_1$ and $v_2$ are harmonic conjugates of $u$ on a region,  
then $v_1 - v_2$ has zero partial derivatives,  
so it is constant:  
a harmonic conjugate is unique up to an additive real constant.

In practice, integrate one equation  
and use the other to fix the remaining function.

For $u = x^3 - 3xy^2$,  
integrating $v_y = u_x = 3x^2 - 3y^2$ with respect to $y$ gives
$$ v = 3x^2y - y^3 + h(x). $$

Then $v_x = 6xy + h'(x)$ must equal $-u_y = 6xy$,  
so $h'(x) = 0$, and $h$ is a constant $C$.  
Hence
$$ u + iv = (x^3 - 3xy^2) + i(3x^2y - y^3) + iC = z^3 + iC. $$

On a disk, a harmonic conjugate always exists.  
Fix a point $p$ of the disk, and for each point $q$ of the disk let
$$ v(q) = \int_\gamma\left(-u_y\,dx + u_x\,dy\right), $$
where $\gamma$ is a path from $p$ to $q$ in the disk.

The value does not depend on the path:  
by Green's theorem, the integral around a closed curve in the disk  
equals $\iint\left(u_{xx} + u_{yy}\right)dA = 0$.  
Ending the path with a short horizontal or vertical segment  
shows that $v_x = -u_y$ and $v_y = u_x$.  
The same construction works on every simply connected region.

On a region with a hole, a harmonic conjugate may fail to exist.  
The function $u = \ln|z| = \tfrac12\ln(x^2 + y^2)$  
is harmonic on $\mathbb C \setminus \{0\}$, and
$$ -u_y\,dx + u_x\,dy = \frac{-y\,dx + x\,dy}{x^2 + y^2}. $$

Around the unit circle $x = \cos t$, $y = \sin t$,  
this integrates to $\int_0^{2\pi}dt = 2\pi \ne 0$,  
so the integral defining $v$ depends on the path.  
Indeed, a harmonic conjugate would be $\arg z$ plus a constant,  
and no continuous branch of $\arg z$  
exists on a region that encircles the origin.

### Orthogonal level curves

Let $f = u + iv$ be holomorphic with $f'(z_0) \ne 0$.  
By the Cauchy–Riemann equations, the gradients at $z_0$ are
$$ \nabla u = (u_x, u_y),\qquad \nabla v = (v_x, v_y) = (-u_y, u_x). $$

So $\nabla u \cdot \nabla v = -u_x u_y + u_y u_x = 0$,  
and $|\nabla u|^2 = |\nabla v|^2 = u_x^2 + u_y^2 = |f'(z_0)|^2 \ne 0$.  
Gradients are perpendicular to level curves,  
so **the level curves $u = c_1$ and $v = c_2$ through $z_0$  
cross at a right angle.**

For $f(z) = z^2$, the red curves are the hyperbolas $x^2 - y^2 = c$,  
and the blue curves are the hyperbolas $2xy = c$.  
They cross at right angles  
everywhere except at the origin, where $f'(0) = 0$.  
There the curves $x^2 - y^2 = 0$, which are the lines $y = \pm x$,  
and $2xy = 0$, which are the axes, meet at $45^\circ$.

For $f(z) = z^3$, the real and imaginary parts are
$$ u = x^3 - 3xy^2,\qquad v = 3x^2y - y^3. $$

```python
levels = [-4, -2, -1, -0.3, 0, 0.3, 1, 2, 4]
G = contour_plot(x^3 - 3*x*y^2, (x, -1.6, 1.6), (y, -1.6, 1.6), contours=levels, plot_points=300, fill=False, cmap=['red'], linestyles='solid')
G += contour_plot(3*x^2*y - y^3, (x, -1.6, 1.6), (y, -1.6, 1.6), contours=levels, plot_points=300, fill=False, cmap=['blue'], linestyles='solid')
G.show(aspect_ratio=1)
```

Now $f'(0) = f''(0) = 0$,  
and the level curves through the origin are lines:  
$u = 0$ at the angles $30^\circ$, $90^\circ$, $150^\circ$,  
and $v = 0$ at the angles $0^\circ$, $60^\circ$, $120^\circ$.  
Neighboring lines meet at $30^\circ$.

In general, if $f^{(n)}(z_0)$ is the first nonzero derivative,  
then $f(z) - f(z_0)$ behaves like a constant times $(z - z_0)^n$.  
The map $w = z^n$ multiplies angles at the origin by $n$,  
so the level curves through $z_0$ meet at $90^\circ/n$.

In physics, the two families have names.  
If $u$ is an electrostatic potential,  
its level curves are the **equipotentials**,  
and the level curves of $v$ are the **field lines**.  
If $u$ is the velocity potential of an ideal flow,  
the level curves of $v$ are the **streamlines**.

For example, $f(z) = z^2$ in the first quadrant  
describes a flow into a right-angled corner:  
the fluid moves along the hyperbolas $xy = c$,  
and the walls $x = 0$ and $y = 0$ are themselves streamlines.

For $f(z) = 1/z$,
$$ u = \frac{x}{x^2 + y^2},\qquad v = -\frac{y}{x^2 + y^2}. $$

Likewise, the level curves of $v$ are circles through $0$ and the real axis.

Physically, $1/z$ is the complex potential of a **dipole**:

### A steady-state temperature problem

Find a function $u$ that is harmonic in the upper half-plane $y > 0$,  
with boundary values
$$ u(x, 0) = 0\quad(x > 0),\qquad u(x, 0) = 1\quad(x < 0). $$

Physically, $u$ is the steady temperature in a large plate  
whose straight edge is held  
at temperature $0$ on one side of the origin  
and at $1$ on the other side.

The function $\operatorname{Arg}z = \operatorname{Im}\operatorname{Log}z$  
is harmonic in the upper half-plane,  
and it tends to $0$ on the positive real axis  
and to $\pi$ on the negative real axis.

So
$$ u(x, y) = \frac1\pi\operatorname{Arg}(x + iy) $$
solves the problem.

Its level curves, the isotherms, are the rays from the origin.  
Heat flows along the level curves  
of the harmonic conjugate $-\frac1\pi\ln|z|$,  
the semicircles centered at $0$.

```python
T = contour_plot(atan2(y, x)/pi, (x, -2, 2), (y, 0, 2), contours=10, cmap='coolwarm', colorbar=True)
T.show(aspect_ratio=1)
```

As a surface of height $u$ over the half-plane,  
the solution is half of a spiral staircase:  
the isotherm $u = t$ is a horizontal segment at height $t$  
above the ray $\operatorname{Arg}z = \pi t$.  
All isotherms meet at the origin,  
where the boundary value jumps from $0$ to $1$.

```python
import numpy as np
import matplotlib.pyplot as plt
R, Theta = np.meshgrid(np.linspace(0, 2, 41), np.linspace(0, np.pi, 181))
X, Y, U = R*np.cos(Theta), R*np.sin(Theta), Theta/np.pi
ax = plt.figure(figsize=(7, 5)).add_subplot(projection='3d')
ax.plot_surface(X, Y, U, cmap='coolwarm', rstride=2, cstride=2, linewidth=0)
for t in np.linspace(0.1, 0.9, 9):  # the isotherm u = t lies over the ray Arg z = pi t
    ax.plot([0, 2*np.cos(np.pi*t)], [0, 2*np.sin(np.pi*t)], [t, t], color='black', linewidth=1.5)
ax.set_box_aspect((2, 1, 1))
ax.set_xticks([-2, -1, 0, 1, 2]); ax.set_yticks([0, 1, 2]); ax.set_zticks([0, 0.5, 1])
ax.view_init(elev=35, azim=-70)
ax.set_xlabel('$x$'); ax.set_ylabel('$y$'); ax.set_zlabel('$u$')
plt.show()
```

## The Maximum Modulus Principle

Recall that in 1002 we proved the mean value property,  
which gives $|f(a)| \le \max_{|z-a|=r}|f(z)|$:  
the modulus at the center of a circle  
is at most the maximum modulus on the circle.

The maximum modulus principle says much more:  
$|f|$ cannot have a maximum inside a region at all,  
unless $f$ is constant.

### The mean value property, revisited

The mean value property says that if $f$ is holomorphic  
on an open set containing the closed disk $|z - a| \le r$, then
$$ f(a) = \frac{1}{2\pi}\int_0^{2\pi}f(a + re^{i\theta})\,d\theta. $$

Taking real parts, $u = \operatorname{Re}f$ satisfies
$$ u(a) = \frac{1}{2\pi}\int_0^{2\pi}u(a + re^{i\theta})\,d\theta. $$

Every harmonic function has a harmonic conjugate on a disk,  
so **every harmonic function has this mean value property**  
on every closed disk in its domain.

### The maximum modulus principle

**Maximum modulus principle.**  
Let $f$ be holomorphic on a region $U$.  
If $|f|$ attains its maximum over $U$ at a point $a \in U$,  
then $f$ is constant on $U$.

Let $M = |f(a)|$.  
If $M = 0$, then $f = 0$.  
Otherwise, multiplying $f$ by the constant $\overline{f(a)}/M$,  
which has modulus $1$,  
we may assume that $f(a) = M > 0$.

Choose $\delta > 0$ such that the disk $|z - a| < \delta$ lies in $U$.  
For $0 < r < \delta$, the real part of the mean value property reads
$$ M = \frac{1}{2\pi}\int_0^{2\pi}\operatorname{Re}f(a + re^{i\theta})\,d\theta. $$

Since $\operatorname{Re}f \le |f| \le M$, the continuous function  
$\theta \mapsto M - \operatorname{Re}f(a + re^{i\theta})$ is nonnegative,  
and its average is $0$, so it vanishes for every $\theta$.  
Then $\operatorname{Re}f = M \ge |f|$ forces $\operatorname{Im}f = 0$,  
so $f = M$ on every circle $|z - a| = r$ with $0 < r < \delta$.

Hence $f = M$ on a disk around $a$.  
Recall that in 1016 we proved the identity theorem;  
it gives $f = M$ on all of $U$.

**Corollary.**  
Let $U$ be a bounded region,  
and let $f$ be holomorphic on $U$  
and continuous on its closure $\overline U$.  
Then $|f|$ attains its maximum over $\overline U$ on the boundary of $U$.

Indeed, $\overline U$ is closed and bounded,  
so the continuous function $|f|$ attains a maximum on $\overline U$.  
If it were attained only at points of $U$,  
then $f$ would be constant on $U$, hence on $\overline U$,  
and the maximum would be attained on the boundary after all.

For example, on the closed unit disk,  
$|z^2 + 1|$ is largest somewhere on the circle $z = e^{i\theta}$, where
$$ |e^{2i\theta} + 1| = |e^{i\theta}|\,|e^{i\theta} + e^{-i\theta}| = 2|\cos\theta|. $$
So the maximum is $2$, attained at $z = \pm 1$.

```python
import numpy as np
import matplotlib.pyplot as plt
X, Y = np.meshgrid(np.linspace(-1.5, 1.5, 121), np.linspace(-1.5, 1.5, 121))
Z = np.abs((X + 1j*Y)**2 + 1)
ax = plt.figure(figsize=(7, 5)).add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', rstride=2, cstride=2, linewidth=0)
ax.view_init(elev=30, azim=-20)
ax.set_xlabel('$x$'); ax.set_ylabel('$y$'); ax.set_zlabel('$|z^2 + 1|$')
plt.show()
```

The graph of $|z^2 + 1|$ over a square  
has two pits, at the zeros $\pm i$,  
and a saddle at $0$, where the derivative $2z$ vanishes.  
It has no peak: over any region, its maximum is on the edge.  
In domain-coloring plots, likewise,  
no bright spot can appear away from the poles.

### The minimum modulus principle

**Minimum modulus principle.**  
Let $f$ be holomorphic and **without zeros** on a region $U$.  
If $|f|$ attains its minimum over $U$ at a point of $U$,  
then $f$ is constant.

Indeed, $1/f$ is holomorphic on $U$,  
and $|1/f|$ attains its maximum at that point.

The hypothesis on zeros is needed:  
$f(z) = z$ on the unit disk  
attains its minimum modulus $0$ at the interior point $0$.  
In domain-coloring plots, dark spots occur only at zeros.

### The maximum principle for harmonic functions

**Maximum principle.**  
If $u$ is harmonic on a region $U$  
and attains its maximum or its minimum at a point of $U$,  
then $u$ is constant.

Suppose $u$ attains its maximum at $a$.  
On a disk $D$ around $a$, write $u = \operatorname{Re}f$ with $f$ holomorphic.  
Then $|e^f| = e^u$ attains its maximum over $D$ at $a$,  
so $e^f$ is constant on $D$, and so is $u = \ln|e^f|$.

So the set of points where $u$ equals its maximum is open;  
it is also closed in $U$, because $u$ is continuous; and it is not empty.  
Since $U$ is connected, it is all of $U$.  
For the minimum, apply the same argument to $-u$.

Physically, a steady temperature distribution  
has no hot spots and no cold spots inside the plate:  
the hottest and the coldest points lie on the boundary.  
In the half-plane problem above, the temperature $\frac1\pi\operatorname{Arg}z$  
stays strictly between its boundary values $0$ and $1$.

A **circle on the sphere** is the intersection of $S^2$ with a plane
$$ \alpha_1x_1 + \alpha_2x_2 + \alpha_3x_3 = \delta $$
that meets $S^2$ in more than one point.

```python
def circle_on_sphere(alpha, delta, n=400):
    alpha = np.array(alpha, dtype=float); a, d = alpha/np.linalg.norm(alpha), delta/np.linalg.norm(alpha)
    e1 = np.cross(a, [0.3, 0.1, 0.9]); e1 /= np.linalg.norm(e1); e2 = np.cross(a, e1)
    t = np.linspace(0, 2*np.pi, n)
    return d*a[:, None] + np.sqrt(1 - d**2)*(np.outer(e1, np.cos(t)) + np.outer(e2, np.sin(t)))

ax = sphere_axes()
for alpha, delta, col in [((0.5, 0.2, 1), 0.2, 'C0'), ((0.5, -1, 1), 1, 'C3')]:  # the red plane contains N
    C = circle_on_sphere(alpha, delta)
    z = sigma(C); z[(abs(z.real) > 2) | (abs(z.imag) > 2)] = np.nan
    ax.plot(*C, color=col); ax.plot(z.real, z.imag, 0*z.real, color=col, linestyle='--')
plt.show()
```

In the picture, the blue circle misses $N$ and projects to a circle,  
and the red circle passes through $N$ and projects to a line.

#### (k) Every rotation comes from $SU(2)$
