# ENGINEERING MATH - COMPLEX VARIABLE

## Why Learn Complex Analysis 2?

Complex numbers turn rotations and oscillations into algebra.  
Complex analysis connects derivatives, power series, contour integrals, and geometry.

We can use these connections to study  
signals, circuits, differential equations, and two-dimensional fields.

### From oscillations to engineering tools

Euler's formula packages a sine wave and a cosine wave into one exponential:
$$ e^{i\omega t} = \cos(\omega t) + i\sin(\omega t). $$
Differentiation becomes multiplication by $i\omega$:
$$ \frac{d}{dt}e^{i\omega t} = i\omega e^{i\omega t}. $$

- **Circuits and signals:** phasors simplify sinusoidal steady-state calculations in linear circuits.  
  Fourier and Laplace transforms organize signals by frequency and growth or decay.
- **Differential equations and control:** poles describe natural modes.  
  A mode $e^{st}$ decays when $\operatorname{Re}s < 0$ and grows when $\operatorname{Re}s > 0$.
- **Potential fields:** the real and imaginary parts of a holomorphic function satisfy Laplace's equation.  
  Conformal maps help solve suitable two-dimensional electrostatic and ideal-flow problems.
- **Integration:** residues can turn difficult real integrals and inverse-transform calculations  
  into computations at a few singularities.

How much follows from complex differentiability?

### Surprise 1: differentiability forces a power series

In real calculus, differentiability does not imply analyticity.  
Even infinitely many derivatives are not enough.

If $f$ is complex differentiable **at every point of an open set** $U$,  
we call $f$ **holomorphic** on $U$.  
It is automatically infinitely differentiable and **analytic**.  
Near every $a\in U$, it equals its Taylor series:
$$ f(z) = \sum_{n = 0}^{\infty}\frac{f^{(n)}(a)}{n!}(z - a)^n. $$

Complex differentiability at just one point is not enough.  
For example, $f(z) = |z|^2$ is complex differentiable at $0$, but nowhere else, so it is not holomorphic on any neighborhood of $0$.

Why is the complex derivative so restrictive?  
The limit
$$ f'(a) = \lim_{h\to 0}\frac{f(a + h) - f(a)}{h} $$
must have the same value as $h$ approaches zero from **every direction in the complex plane**.

#### A smooth real function that is not analytic

Consider
$$ g(x) = \begin{cases}
e^{-1/x^2}, & x\ne 0, \\
0, & x = 0.
\end{cases} $$
This function is infinitely differentiable on $\mathbb R$, and
$$ g^{(n)}(0) = 0\qquad \text{for every }n\ge 0. $$

$e^{-1/x^2}$ tends to zero faster than every power of $|x|$.  
Away from zero, each derivative is a polynomial in $1/x$  
times that exponential.

Its Taylor series at zero is therefore the zero series.  
Yet $g(x) > 0$ for every $x\ne 0$,  
so the Taylor series does not represent $g$ on any neighborhood of zero.

**Real smoothness permits behavior that holomorphicity rules out.**

A related smooth, nonanalytic example appears in Section 10.7:  
[John K. Hunter's introduction to analysis](https://www.math.ucdavis.edu/~hunter/intro_analysis_pdf/intro_analysis.pdf).

#### The mechanism: Cauchy's integral formula

Suppose $f$ is holomorphic on a neighborhood of the closed disk $|\zeta - a|\le r$.  
For $|z - a| < r$,
$$ f(z) = \frac{1}{2\pi i}\oint_{|\zeta - a| = r}\frac{f(\zeta)}{\zeta - z}\,d\zeta, $$
where the circle is traversed counterclockwise.

Expand the kernel as a geometric series:
$$ \frac{1}{\zeta - z} = \sum_{n = 0}^{\infty}\frac{(z - a)^n}{(\zeta - a)^{n + 1}}. $$

On every smaller disk, the series converges uniformly along the contour.  
We can integrate term by term.  
This produces the Taylor series and the derivative formula
$$ f^{(n)}(a) = \frac{n!}{2\pi i}\oint_{|\zeta - a| = r}\frac{f(\zeta)}{(\zeta - a)^{n + 1}}\,d\zeta. $$

**Values along a surrounding circle determine every value and every derivative inside.**  
This is why a local derivative condition gives such strong control.

### Complex singularities explain real approximation limits

The real function
$$ h(x) = \frac{1}{1 + x^2} $$
is smooth for every real $x$, but its Taylor series at zero is
$$ \frac{1}{1 + x^2} = \sum_{n = 0}^{\infty}(-1)^n x^{2n}, \qquad |x| < 1. $$

Why does the series stop converging at distance $1$?  
Nothing goes wrong on the real axis.

In the complex plane,
$$ 1 + z^2 = (z - i)(z + i), $$
so there are poles at $z = \pm i$, both distance $1$ from the origin.  
These complex singularities limit the Taylor series' radius of convergence.

**Complex analysis explains the limits of a real-variable approximation.**

### Surprise 2: generating functions

How many $n$-bit binary strings have a number of $1$s divisible by $3$?  
This is a purely combinatorial question: no complex numbers in sight.

The count is a sum of binomial coefficients,
$$ b_n = \sum_{3\mid k}\binom nk, $$
a sum over roughly a third of the terms of $(1 + x)^n = \sum_k\binom nk x^k$.

To isolate every third term, plug $x = 1, \omega, \omega^2$ into $(1 + x)^n$,  
where $\omega = e^{2\pi i/3}$ is a cube root of unity, and average:
$$ \boxed{b_n = \frac13\bigl[(1 + 1)^n + (1 + \omega)^n + (1 + \omega^2)^n\bigr]}. $$

For each exponent $k$, the average $\frac13(1 + \omega^k + \omega^{2k})$ equals $1$ when $3\mid k$  
and $0$ otherwise, because $1, \omega^k, \omega^{2k}$ then sum to zero.  
Every term with $3\nmid k$ cancels, leaving exactly $b_n$.

The formula for $b_n$ is manifestly a nonnegative integer,  
yet its cleanest closed form passes through the complex number $\omega$.

**Complex numbers can be the simplest route to a purely real, integer-valued answer.**

### Surprise 3: extra complex dimensions can remove singularities

*Optional preview beyond this course's main focus on one complex variable.*

In one complex variable, $1/z$ has a nonremovable singularity at $0$.  
In two or more complex variables, every isolated singularity of a holomorphic function is removable: this is a consequence of **Hartogs' extension theorem**.

More generally, let $A$ be a closed complex analytic subset   of a domain $U\subset \mathbb C^n$,  
with **complex codimension at least two**.  
Every holomorphic function on $U\setminus A$ extends holomorphically across $A$.

Here codimension means the number of missing **complex** dimensions.  
A point in $\mathbb C^2$ has complex codimension two (real codimension four).  
A point in $\mathbb C$ has real codimension two, and $1/z$ shows that such a singularity need not be removable.

See the Hartogs discussion in [Several Complex Variables, Section 3](https://math.hse.ru/data/2024/02/05/2090218179/lektsii-24_0402.pdf).

#### Use one variable for the contour and another to bypass the hole

Here is the idea for a missing point in $\mathbb C^2$.  
Suppose $f(z, w)$ is holomorphic on the bidisk
$$ \{(z, w) : |z| < R, \ |w| < R\}\setminus \{(0, 0)\}. $$

Choose $0 < r < R$.  
For $|z| < r$ and $|w| < R$, define
$$ F(z, w) = \frac{1}{2\pi i}\oint_{|\zeta| = r}\frac{f(\zeta, w)}{\zeta - z}\,d\zeta. $$

The contour never touches the missing point,  
even when $w = 0$, because $|\zeta| = r > 0$.  
The integrand depends holomorphically on $(z, w)$ near the contour.  
The integral therefore defines a holomorphic function,  
including at $(0, 0)$.

When $w\ne 0$, the entire $z$-disk is available.  
Cauchy's formula gives $F(z, w) = f(z, w)$.  
Continuity then gives equality for $w = 0$, $z\ne 0$, too.  
Thus $F$ fills in the missing value and agrees with $f$ wherever both are defined.

One complex variable runs the Cauchy integral.  
The other lets us move to slices without a hole.

This also explains why complex codimension one is different:  
$1/z$ in $\mathbb C^2$ is singular along the entire set $\{z = 0\}$.  
Moving $w$ never produces a slice without that singularity.

## Continuity

### Continuous functions: $\mathbb R\to\mathbb R$

A function $f : \mathbb R\to\mathbb R$ is **continuous** at $a$  
if for every $\varepsilon > 0$ there is a $\delta > 0$ such that
$$ |x - a| < \delta \quad \Longrightarrow \quad |f(x) - f(a)| < \varepsilon. $$
Equivalently, $\lim_{x\to a}f(x) = f(a)$:
nearby inputs must produce nearby outputs.

For example, $f(x) = x^2$ is continuous everywhere,  
while the sign function
$$ \operatorname{sgn}(x) = \begin{cases}
1, & x > 0, \\
0, & x = 0, \\
-1, & x < 0.
\end{cases} $$
is discontinuous at $0$:
no $\delta$ makes nearby inputs land close to a single output value.

### Continuous functions: $\mathbb R^2\to\mathbb R^2$

A function $f : \mathbb R^2\to\mathbb R^2$ is **continuous** at $a$  
if for every $\varepsilon > 0$ there is a $\delta > 0$ such that
$$ \|x - a\| < \delta \quad \Longrightarrow \quad \|f(x) - f(a)\| < \varepsilon, $$
where $\|\cdot\|$ is the Euclidean length of a vector in $\mathbb R^2$.  
The definition is identical to the $\mathbb R\to\mathbb R$ case,  
with absolute value replaced by length.

For example, $f(x, y) = (x^2 - y^2, 2xy)$ is continuous everywhere:
each coordinate is a polynomial in $x, y$, hence continuous,  
and a map into $\mathbb R^2$ is continuous exactly when both of its coordinate functions are.

### Continuous functions: $\mathbb C\to\mathbb C$

A function $f : \mathbb C\to\mathbb C$ is **continuous** at $a$  
if for every $\varepsilon > 0$ there is a $\delta > 0$ such that
$$ |z - a| < \delta \quad \Longrightarrow \quad |f(z) - f(a)| < \varepsilon, $$
where $|\cdot|$ is now the complex modulus.  
Writing $z = x + iy$, this is the same condition as continuity of $\mathbb R^2\to\mathbb R^2$,  
since $|z| = \|(x, y)\|$.

For example, $f(z) = z^2$ is continuous everywhere.  
Writing $z = x + iy$ gives $z^2 = (x^2 - y^2) + i(2xy)$,  
matching the $\mathbb R^2\to\mathbb R^2$ example above.

## Differentiability

### Differentiable functions: $\mathbb R\to\mathbb R$

A function $f : \mathbb R\to\mathbb R$ is **differentiable** at $a$  
if the limit
$$ f'(a) = \lim_{h\to 0}\frac{f(a + h) - f(a)}{h} $$
exists, where $h$ ranges over nonzero real numbers.  
Division by $h$ makes sense because $\mathbb R$ is a field.

For example, $f(x) = x^2$ is differentiable everywhere, with $f'(x) = 2x$:
$$ \frac{(x + h)^2 - x^2}{h} = 2x + h, $$
which tends to $2x$ as $h\to 0$.  
By contrast, $f(x) = |x|$ is not differentiable at $0$,  
since the quotient equals $1$ for $h > 0$ and $-1$ for $h < 0$.

### Differentiable functions: $\mathbb R^2\to\mathbb R^2$

For $f : \mathbb R^2\to\mathbb R^2$, dividing by a vector $h\in\mathbb R^2$ makes no sense,  
so we instead ask for a good **linear** approximation.  
$f$ is **differentiable** at $a$ if there is a linear map $L : \mathbb R^2\to\mathbb R^2$ such that
$$ \lim_{h\to 0}\frac{\|f(a + h) - f(a) - Lh\|}{\|h\|} = 0. $$
Such an $L$, when it exists, is unique, and is written $Df(a)$.

For example, $f(x, y) = (x^2 - y^2, 2xy)$ is differentiable everywhere,  
with $Df(a)$ given by the Jacobian matrix of partial derivatives:
$$ Df(x, y) = \begin{pmatrix} 2x & -2y \\ 2y & 2x \end{pmatrix}. $$
In general, $Df(a)$ can be **any** $2 \times 2$ real matrix,  
such as one that stretches the two axes by different amounts.

### Differentiable functions: $\mathbb C\to\mathbb C$

A function $f : \mathbb C\to\mathbb C$ is **differentiable** at $a$  
(or **complex differentiable**) if the limit
$$ f'(a) = \lim_{h\to 0}\frac{f(a + h) - f(a)}{h} $$
exists, where $h$ now ranges over nonzero complex numbers.  
Division by $h$ makes sense because $\mathbb C$ is a field,  
so the definition looks exactly like the $\mathbb R\to\mathbb R$ case.

For example, $f(z) = z^2$ is complex differentiable everywhere, with $f'(z) = 2z$:
the real computation above works unchanged, since it only uses field axioms.  
But now $h\to 0$ may approach from **any direction in the complex plane**,  
and the limit must be the same single complex number $f'(a)$ every time,  
a far stronger requirement than the $\mathbb R^2\to\mathbb R^2$ case,  
where $Df(a)$ was allowed to be any $2 \times 2$ real matrix.

### Real differentiability does not imply complex differentiability

Consider $f(z) = \bar z$, that is, $f(x, y) = (x, -y)$.  
As a map $\mathbb R^2\to\mathbb R^2$, $f$ is linear,  
so it is differentiable everywhere, with
$$ Df(x, y) = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}. $$

The complex difference quotient is
$$ \frac{f(a + h) - f(a)}{h} = \frac{\bar h}{h}. $$
For real $h$, this equals $1$;  
for purely imaginary $h$, this equals $-1$.  
The limit as $h\to 0$ depends on direction, so it does not exist:
$f$ is **not** complex differentiable anywhere.

**Real differentiability does not imply complex differentiability.**  
The next section pins down exactly what extra condition is missing.

### The Cauchy–Riemann equations

Write $f(z) = f(x + iy) = u(x, y) + iv(x, y)$,  
where $u, v : \mathbb R^2\to\mathbb R$ are the real and imaginary parts of $f$.  
Writing subscripts for partial derivatives, the Jacobian of $f$ as a map $\mathbb R^2\to\mathbb R^2$ is
$$ Df(x, y) = \begin{pmatrix} u_x & u_y \\ v_x & v_y \end{pmatrix}. $$

Recall that multiplying by a complex number $p + qi$ corresponds to the real matrix
$$ \begin{pmatrix} p & -q \\ q & p \end{pmatrix}. $$
$f$ is complex differentiable at $a$ exactly when $Df(a)$ takes this form:
$u_x = v_y$ and $u_y = -v_x$.

These are the **Cauchy–Riemann equations**:
$$ \boxed{u_x = v_y, \qquad u_y = -v_x}. $$
When they hold, the derivative is
$$ f'(a) = u_x + iv_x = v_y - iu_y. $$

Precisely: $f$ is complex differentiable at $a$ if and only if $f$ is $\mathbb R^2$-differentiable at $a$  
**and** $u, v$ satisfy the Cauchy–Riemann equations there.  
The equations alone, without real differentiability, are not enough.

For $f(z) = z^2$, that is, $u = x^2 - y^2$ and $v = 2xy$,
$$ u_x = 2x = v_y, \qquad u_y = -2y = -v_x, $$
so the Cauchy–Riemann equations hold everywhere,  
matching $f'(z) = 2z$ found earlier.

For $f(z) = \bar z$, that is, $u = x$ and $v = -y$,
$$ u_x = 1, \qquad v_y = -1, $$
so $u_x\ne v_y$ everywhere:
the Cauchy–Riemann equations fail, confirming that $f$ is nowhere complex differentiable.

## Topology of the Complex Plane

### Open sets

For $z_0\in\mathbb C$ and $r > 0$, the **open disk** of radius $r$ centered at $z_0$ is
$$ D(z_0, r) = \{z\in\mathbb C : |z - z_0| < r\}. $$
A set $U\subset \mathbb C$ is **open** if for every $z_0\in U$  
there is some $r > 0$ such that $D(z_0, r)\subset U$.

For example, every open disk $D(z_0, r)$ is open,  
and so are $\mathbb C$ itself and the empty set $\emptyset$ (vacuously).  
The half-plane $\{z : \operatorname{Re}z > 0\}$ is open,  
but $\{z : \operatorname{Re}z\ge 0\}$ is not:
no disk around a point with $\operatorname{Re}z = 0$ stays inside the set.

### Unions and intersections of open sets

Let $\{U_\alpha\}$ be any collection of open subsets of $\mathbb C$, finite or infinite.  
Their union $\bigcup_\alpha U_\alpha$ is open:
if $z_0$ lies in the union, then $z_0\in U_\alpha$ for some $\alpha$,  
and any disk $D(z_0, r)\subset U_\alpha$ also lies in the union.

Now let $U_1, \dotsc, U_n$ be **finitely many** open sets.  
Their intersection $\bigcap_{k = 1}^n U_k$ is open:
if $z_0$ lies in the intersection, each $U_k$ contains some disk $D(z_0, r_k)$,  
and $D(z_0, r)\subset \bigcap_k U_k$ for $r = \min(r_1, \dotsc, r_n)$,  
which is positive because there are only finitely many radii to take the minimum of.

This last step genuinely needs finitely many sets:
with infinitely many $U_n$, the infimum of the radii could be $0$,  
and the argument breaks down.

For example, let $U_n = D(0, 1/n)$ for $n = 1, 2, 3, \dotsc$  
Each $U_n$ is open, but
$$ \bigcap_{n = 1}^{\infty} U_n = \{0\}, $$
a single point, which is closed, not open.

### Closed sets

A set $F\subset \mathbb C$ is **closed** if its complement $\mathbb C\setminus F$ is open.  
Equivalently, $F$ contains every point that is a limit of points in $F$.

For example, every closed disk $\{z : |z - z_0|\le r\}$ is closed,  
and so are a single point $\{z_0\}$ and the whole plane $\mathbb C$.  
The set $\{z : 0\le \operatorname{Re}z < 1\}$ is neither open nor closed.  
The only subsets of $\mathbb C$ that are both open and closed are $\emptyset$ and $\mathbb C$.

### Bounded sets

A set $S\subset \mathbb C$ is **bounded** if there is some $R > 0$ such that
$$ S\subset D(0, R), $$
that is, $|z|\le R$ for every $z\in S$.  
Otherwise, $S$ is **unbounded**.

For example, every disk (open or closed) is bounded,  
while a half-plane, a line, and $\mathbb C$ itself are unbounded.  
Boundedness does not depend on being open or closed:
the punctured disk $\{z : 0 < |z| < 1\}$ is bounded and open,  
while $\{z : |z|\ge 1\}$ is unbounded and closed.

## Topology of the Riemann Sphere

### The point at infinity

The **Riemann sphere** adds a single point $\infty$ to the complex plane,
$$ \hat{\mathbb C} = \mathbb C\cup\{\infty\}. $$
Stereographic projection realizes $\hat{\mathbb C}$ as an actual sphere in $\mathbb R^3$,  
with $\infty$ corresponding to the north pole,  
but for topology we only need to say what a neighborhood of $\infty$ looks like.

For $R > 0$, a **neighborhood of $\infty$** is a set of the form
$$ \{z\in\mathbb C : |z| > R\}\cup\{\infty\}. $$
This plays the same role at $\infty$ that the disk $D(z_0, r)$ plays at a finite point $z_0$.

### Open sets on the sphere

A set $U \subset \hat{\mathbb C}$ is **open**  
if every point of $U$ has a neighborhood inside $U$:  
a disk $D(z_0, r) \subset U$ at each finite $z_0 \in U$,    
and a neighborhood of $\infty$ contained in $U$ if $\infty \in U$.

For example, $\{z : |z| > R\}\cup\{\infty\}$ is open,  
and so is $\hat{\mathbb C}$ itself.  
Every open subset of $\mathbb C$ is still open in $\hat{\mathbb C}$,  
since disks around finite points do not change.

### Closed sets on the sphers

As before, $F\subset \hat{\mathbb C}$ is **closed** if $\hat{\mathbb C} \setminus F$ is open.

For example, every closed disk is still closed,  
and so is the single point $\{\infty\}$.  
The only subsets of $\hat{\mathbb C}$ that are both open and closed are $\emptyset$ and $\hat{\mathbb C}$.

### "Bounded" sets on the sphere

**Every subset of $\hat{\mathbb C}$ is "bounded".**  
Adding the single point $\infty$ turns every previously unbounded set,  
such as a half-plane, a line, or $\mathbb C$ itself, into a "bounded" one:
each now sits inside the sphere alongside $\infty$, with no room left to escape to.

"Boundedness" is therefore no longer a meaningful distinction on $\hat{\mathbb C}$;  
only open and closed remain interesting.  
This is the topological price, or reward, of compactifying $\mathbb C$ by a single point.

## The Real Projective Plane

*This section is for contrast only: it will not appear on homework or exams.*

The Riemann sphere added a single point at infinity to $\mathbb C\cong\mathbb R^2$.  
A different way to compactify the plane gives a very different surface.

The **real projective plane** $\mathbb{RP}^2$ adds one point at infinity  
for every **direction** in $\mathbb R^2$, with opposite directions identified.  
Equivalently, $\mathbb{RP}^2$ is the set of lines through the origin in $\mathbb R^3$.

Directions in $\mathbb R^2$ form a circle,  
so the points at infinity form a whole circle, not a single point,  
with each point on that circle glued to the one opposite it.

Every line in $\mathbb R^2$ extends in two opposite directions,  
which are identified to the **same** point at infinity.  
Two parallel lines therefore meet at exactly one point at infinity,  
matching the classical picture from perspective drawing.

Adding a single point at infinity to $\mathbb R^2$ gives the sphere $S^2$, which is orientable.  
Adding a whole circle of directions, with antipodal points identified, gives $\mathbb{RP}^2$,  
which is **non-orientable** and cannot be embedded in $\mathbb R^3$ without self-intersection.

The square represents a disk whose boundary points are identified antipodally.  
Write $a = S/2$ for its half-width.  
Across the vertical edges, the local continuation is
$$ (a + t, y + v)\longmapsto(-a + t, -y - v). $$
It preserves horizontal motion and reflects the portrait from top to bottom.  
Across the horizontal edges, the local continuation is
$$ (x + u, a + t)\longmapsto(-x - u, -a + t). $$
It preserves vertical motion and reflects the portrait from left to right.  
After a second crossing, the reflection undoes itself.

```python
import urllib.request
from PIL import Image, ImageDraw

MONA_URL = "https://upload.wikimedia.org/wikipedia/commons/6/6a/Mona_Lisa.jpg"
MONA_PATH = "0918/mona.jpg"
HORIZONTAL_PATH = "0918/rp2_mona.gif"
VERTICAL_PATH = "0918/rp2_mona_vertical.gif"

S = 400
A = S / 2
MARGIN = 40
CANVAS = S + 2 * MARGIN
BORDER = 3
```

```python
try:
    with open(MONA_PATH, "rb") as file:
        raw = file.read()
except FileNotFoundError:
    request = urllib.request.Request(MONA_URL, headers={"User-Agent": "Mozilla/5.0"})
    raw = urllib.request.urlopen(request).read()
    with open(MONA_PATH, "wb") as file:
        file.write(raw)

mona = Image.open(MONA_PATH).convert("RGBA")
sprite_h = 150
sprite_w = round(mona.width * sprite_h / mona.height)
sprite = mona.resize((sprite_w, sprite_h), Image.Resampling.LANCZOS)
sprite_tb = sprite.transpose(Image.Transpose.FLIP_TOP_BOTTOM)
sprite_lr = sprite.transpose(Image.Transpose.FLIP_LEFT_RIGHT)
HALF_W, HALF_H = sprite_w / 2, sprite_h / 2
```

```python
def to_px(x, y):
    return MARGIN + x + A, MARGIN + A - y


square_clip = Image.new("L", (CANVAS, CANVAS), 0)
ImageDraw.Draw(square_clip).rectangle(
    [MARGIN, MARGIN, MARGIN + S, MARGIN + S], fill=255
)
```

```python
def arrow(draw, point, direction, color, size=14):
    px, py = point
    dx, dy = direction
    draw.line(
        [px - dx * size, py - dy * size, px + dx * size, py + dy * size],
        fill=color, width=3,
    )
    tip = px + dx * size, py + dy * size
    left = tip[0] - dy * 6 - dx * 8, tip[1] + dx * 6 - dy * 8
    right = tip[0] + dy * 6 - dx * 8, tip[1] - dx * 6 - dy * 8
    draw.polygon([tip, left, right], fill=color)
```

```python
def draw_fundamental_square(draw):
    draw.rectangle(
        [to_px(-A, A), to_px(A, -A)], outline="black", width=BORDER
    )
    red = (180, 0, 0, 255)
    green = (0, 140, 0, 255)
    arrow(draw, to_px(A, S / 4), (0, -1), red)
    arrow(draw, to_px(-A, -S / 4), (0, 1), red)
    arrow(draw, to_px(S / 4, A), (-1, 0), green)
    arrow(draw, to_px(-S / 4, -A), (1, 0), green)
```

```python
def paste_sprite(layer, center_x, center_y, image):
    px, py = to_px(center_x, center_y)
    layer.alpha_composite(
        image, (round(px - image.width / 2), round(py - image.height / 2))
    )


def finish_frame(layer):
    red, green, blue, alpha = layer.split()
    alpha = Image.composite(alpha, Image.new("L", alpha.size, 0), square_clip)
    layer = Image.merge("RGBA", (red, green, blue, alpha))
    canvas = Image.new("RGBA", (CANVAS, CANVAS), "white")
    canvas.alpha_composite(layer)
    draw_fundamental_square(ImageDraw.Draw(canvas))
    return canvas.convert("RGB")
```

#### Horizontal motion

```python
def render_horizontal(x, y, reflected):
    layer = Image.new("RGBA", (CANVAS, CANVAS), (0, 0, 0, 0))
    image = sprite_tb if reflected else sprite
    seam_image = sprite if reflected else sprite_tb
    paste_sprite(layer, x, y, image)

    if x + HALF_W > A:
        paste_sprite(layer, x - S, -y, seam_image)
    if x - HALF_W < -A:
        paste_sprite(layer, x + S, -y, seam_image)
    return finish_frame(layer)
```

```python
horizontal_frames = []
vx = 8
x, y = -A + vx, S / 4
reflected = False

for _ in range(100):
    horizontal_frames.append(render_horizontal(x, y, reflected))
    x += vx
    if x > A:
        x -= S
        y = -y
        reflected = not reflected

horizontal_frames[0].save(
    HORIZONTAL_PATH, save_all=True, append_images=horizontal_frames[1:],
    duration=45, loop=0, disposal=2,
)
```

#### Vertical motion

```python
def render_vertical(x, y, reflected):
    layer = Image.new("RGBA", (CANVAS, CANVAS), (0, 0, 0, 0))
    image = sprite_lr if reflected else sprite
    seam_image = sprite if reflected else sprite_lr
    paste_sprite(layer, x, y, image)

    if y + HALF_H > A:
        paste_sprite(layer, -x, y - S, seam_image)
    if y - HALF_H < -A:
        paste_sprite(layer, -x, y + S, seam_image)
    return finish_frame(layer)
```

```python
vertical_frames = []
vy = 8
x, y = S / 4, -A + vy
reflected = False

for _ in range(100):
    vertical_frames.append(render_vertical(x, y, reflected))
    y += vy
    if y > A:
        y -= S
        x = -x
        reflected = not reflected

vertical_frames[0].save(
    VERTICAL_PATH, save_all=True, append_images=vertical_frames[1:],
    duration=45, loop=0, disposal=2,
)
```
