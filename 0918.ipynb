### Three collinear points from a determinant

Let $a, b, c$ be three distinct points in $\mathbb C$.  
Show that they are collinear if and only if
$$ \boxed{\det\begin{pmatrix}
a & \bar a & 1 \\
b & \bar b & 1 \\
c & \bar c & 1
\end{pmatrix} = 0}. $$

First explain why $a, b, c$ are collinear if and only if
$$ \frac{c - a}{b - a}\in\mathbb R. $$
Use the fact that a complex number is real exactly when it equals its conjugate.  
Relate the resulting equation to the determinant above.

What happens to the determinant if two of the points coincide?

### A circumcircle from a determinant

Let $a, b, c$ be three noncollinear points in $\mathbb C$.  
The **circumcircle** is the circle passing through these three points.  
Its center is called the **circumcenter**.

We want to describe this circle using a determinant.

Let $o\in\mathbb C$ be the center of a circle and let $R > 0$ be its radius.  
Start from
$$ |z - o|^2 = R^2. $$
Expand this equation to write it as a linear combination of
$z\bar z$, $z$, $\bar z$, and $1$,  
wherein the coefficients can contain $o$ and $R$

Show that $z$ lies on the circumcircle of $a, b, c$ if and only if
$$ \det\begin{pmatrix}
z\bar z & z & \bar z & 1 \\
a\bar a & a & \bar a & 1 \\
b\bar b & b & \bar b & 1 \\
c\bar c & c & \bar c & 1
\end{pmatrix} = 0. $$
Explain why the determinant vanishes at the three given points.  
Then prove that its zero set is exactly their circumcircle.  
Checking only the three given points is not enough.

Where do you use the assumption that the points are noncollinear?
