### Arzelà–Ascoli by a diagonal argument

This problem proves the Arzelà–Ascoli theorem, the engine behind normal families.  
Let $K \subset \mathbb C$ be compact, and let $f_n : K \to \mathbb C$ be a sequence of functions.  
Assume that the sequence is

- **uniformly bounded:** $|f_n(z)| \le M$ for all $n$ and all $z \in K$; and
- **equicontinuous:** for every $\varepsilon > 0$ there is $\delta > 0$ such that $|z - w| < \delta$ implies $|f_n(z) - f_n(w)| < \varepsilon$ for **all** $n$ and all $z, w \in K$.

The goal is to find a subsequence that converges uniformly on $K$.

Remark:  
The uniformity in uniform boundedness is a "$\forall n$" quantifier.  
The "equi" part of equicontinuity is again a "$\forall n$" quantifier.  
Why can't we call it uniform continous?  
Becuase that uniform is a "$\forall z$" quantifier.

#### (a) A countable dense subset

Show that for every $\varepsilon > 0$ there is a finite set $F_\varepsilon \subset K$  
such that every point of $K$ lies within distance $\varepsilon$ of $F_\varepsilon$.  
Hint: if not, choose $z_1 \in K$, then $z_2 \in K$ with $|z_2 - z_1| \ge \varepsilon$,  
then $z_3 \in K$ at distance at least $\varepsilon$ from both, and so on.  
Show that this sequence has no convergent subsequence,  
contradicting the compactness problem from the previous lecture.  
Deduce that $S = \bigcup_{m \ge 1}F_{1/m}$ is countable  
and that every point of $K$ is a limit of points of $S$.

#### (b) The diagonal argument

Let $S = \{s_1, s_2, s_3, \ldots\}$ be a countable set,  
and suppose $|f_n(s)| \le M$ for all $n$ and all $s \in S$.  
Show that some subsequence $(f_{n_k})$ converges at **every** point of $S$.  
Hint: $(f_n(s_1))$ is a bounded sequence,  
so by part (b) of the compactness problem some subsequence of $(f_n)$ converges at $s_1$.  
From it extract a further subsequence converging at $s_2$.  
From it extract a further subsequence converging at $s_3$.  
From it extract a further subsequence converging at $s_4$.  
No single one of these subsequences works for all $s_j$.  
Instead, take the $k$th term of the $k$th subsequence,  
and explain why this **diagonal** sequence converges at every $s_j$.

#### (c) From countably many points to the whole set

Let $(g_k)$ be the diagonal subsequence from (b), with $S$ as in (a).  
Show that $(g_k)$ is **uniformly Cauchy** on $K$.  
Given $\varepsilon > 0$, choose $\delta$ by equicontinuity, then $m$ with $1/m < \delta$, and let $F = F_{1/m} \subset S$.  
Choose $N$ such that $|g_k(s) - g_l(s)| < \varepsilon$ for all $s \in F$ and all $k, l \ge N$.  
For an arbitrary $z \in K$, pick $s \in F$ with $|z - s| < \delta$ and estimate
$$ |g_k(z) - g_l(z)| \le |g_k(z) - g_k(s)| + |g_k(s) - g_l(s)| + |g_l(s) - g_l(z)|. $$
Conclude, using the completeness of $\mathbb C$, that $(g_k)$ converges uniformly on $K$.

#### (d) Holomorphic functions give equicontinuous families

Surprise: it's all about Cauchy Integral formulachallenge

Let each $f_n$ be holomorphic on an open set containing the closed disk $|z - a| \le 2r$, with $|f_n(z)| \le M$ there.  
Take for granted that Cauchy's formula for derivatives and Cauchy's estimate hold for such functions.  
If $|z - a| \le r$, then the circle $|\zeta - z| = r$ lies in the disk $|\zeta - a| \le 2r$,  
so Cauchy's estimate gives $|f_n'(z)| \le M/r$.  
Integrate $f_n'$ along the line segment from $w$ to $z$ to show
$$ |f_n(z) - f_n(w)| \le \frac{M}{r}|z - w| $$
whenever $|z - a| \le r$ and $|w - a| \le r$.  
Conclude that $(f_n)$ has a subsequence converging uniformly on $|z - a| \le r$.  
A family with this property on every compact subset is called **normal**.

### Numerical check (optional)

Check your answers for the circles numerically with the function below.  
It approximates $\oint_{|z - a| = r} f(z)\,dz$, counterclockwise, by the trapezoid rule.  
For example, the last cell checks $\oint_{|z| = 2}e^z/(z - 1)\,dz = 2\pi i\,e$.

````python
def tidy(w, tol=1e-9):
    x0, y0 = w.real(), w.imag()
    return CDF(0 if abs(x0) < tol else x0, 0 if abs(y0) < tol else y0)
````

````python
def circle_integral(f, a=0, r=1, n=2000):
    """Integral of f over |z - a| = r, counterclockwise (trapezoid rule)."""
    h = 2 * RDF(pi) / n
    total = CDF(0)
    for k in range(n):
        w = CDF(cos(k * h), sin(k * h))
        total += f(a + r * w) * (CDF(0, 1) * r * w) * h
    return tidy(total)
````

### Winding numbers

Let $\gamma : [t_0, t_1] \to \mathbb C$ be a closed, piecewise smooth path, and let $a$ be a point not on $\gamma$.  
Recall the winding number
$$ \operatorname{wind}(\gamma, a) = \frac{1}{2\pi i}\oint_\gamma\frac{dz}{z - a}. $$
This problem proves its basic properties and then uses it to count the roots of a polynomial.

#### (a) The winding number is an integer

For $t_0 \le t \le t_1$, let
$$ h(t) = \int_{t_0}^t\frac{\gamma'(s)}{\gamma(s) - a}\,ds, $$
so that $h(t_1) = 2\pi i\operatorname{wind}(\gamma, a)$.  
Show that $\bigl(\gamma(t) - a\bigr)e^{-h(t)}$ has derivative $0$ on each smooth piece of $\gamma$,  
so it is constant.  
Since $\gamma(t_1) = \gamma(t_0)$, conclude that $e^{h(t_1)} = 1$,  
and hence, by the description of all logarithms of $1$, that $\operatorname{wind}(\gamma, a)$ is an integer.

#### (b) Constant on each region

For points $a$ and $b$ not on $\gamma$, show that
$$ \operatorname{wind}(\gamma, a) - \operatorname{wind}(\gamma, b)
= \frac{a - b}{2\pi i}\oint_\gamma\frac{dz}{(z - a)(z - b)}. $$
Let $\delta > 0$ be the distance from $a$ to $\gamma$, and let $L$ be the length of $\gamma$.  
For $|b - a| < \delta/2$, use the ML-inequality to show
$$ |\operatorname{wind}(\gamma, a) - \operatorname{wind}(\gamma, b)| \le \frac{L|a - b|}{\pi\delta^2}. $$
Conclude that $\operatorname{wind}(\gamma, b) = \operatorname{wind}(\gamma, a)$ for all $b$ close enough to $a$,  
and explain why the winding number is constant along every path that does not meet $\gamma$.

Now suppose that $\gamma$ lies in the disk $|z| \le R$, and let $|a| > R$.

````python
var('t')

def plot_curve(gamma, **kwds):
    """Plot the curve gamma(t), 0 <= t <= 2 pi, in the complex plane."""
    return parametric_plot((real_part(gamma), imag_part(gamma)), (t, 0, 2 * pi),
                           aspect_ratio=1, **kwds)
````

````python
def tidy(w, tol=1e-9):
    x0, y0 = w.real(), w.imag()
    return CDF(0 if abs(x0) < tol else x0, 0 if abs(y0) < tol else y0)

def winding_number(gamma, a, n=2000):
    """Winding number of the closed curve gamma(t), 0 <= t <= 2 pi, around a."""
    g = fast_callable(gamma, vars=[t], domain=CDF)
    dg = fast_callable(diff(gamma, t), vars=[t], domain=CDF)
    h = 2 * RDF(pi) / n
    total = sum(dg(k * h) / (g(k * h) - a) for k in range(n)) * h
    return tidy(total / CDF(0, 2 * RDF(pi)))
````
