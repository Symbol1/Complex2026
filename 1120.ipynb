# ENGINEERING MATH - COMPLEX VARIABLE

Today we use the residue theorem from 1016  
to evaluate real integrals,  
many of which have no elementary antiderivative.

The idea is to recognize a real integral  
as a contour integral, or as a piece of one.  
We treat three kinds of integrals, where $F$ and $f$ are rational functions:

1. $\int_0^{2\pi}F(\cos\theta, \sin\theta)\,d\theta$, over the unit circle;
2. $\int_{-\infty}^\infty f(x)\,dx$, closed by a large semicircle;
3. $\int_{-\infty}^\infty f(x)\cos(ax)\,dx$ and $\int_{-\infty}^\infty f(x)\sin(ax)\,dx$, through $e^{iaz}$.

Poles on the real axis are handled at the end with indented contours.

## Recap: The Residue Theorem

Recall that in 1016 we defined the **residue** of $f$  
at an isolated singularity $a$  
as the coefficient $a_{-1}$ of $(z - a)^{-1}$ in its Laurent series at $a$.  
By Laurent's coefficient formula, for every small $\rho > 0$,
$$ \operatorname{Res}(f, a) = a_{-1} = \frac{1}{2\pi i}\oint_{|z-a|=\rho}f(z)\,dz. $$

**Residue theorem.**  
Let $C$ be a positively oriented simple closed curve,  
and let $f$ be holomorphic on and inside $C$  
except at finitely many points $a_1, \ldots, a_k$ inside $C$. Then
$$ \oint_C f(z)\,dz = 2\pi i\sum_{j=1}^k\operatorname{Res}(f, a_j). $$

![Blackboard: a closed curve that winds twice around one singular point and once around another, deformed into two loops](1120/curve-winding-twice-split-into-loops.jpeg)

Today all singularities are poles.  
At a simple pole, $f(z) = a_{-1}/(z - a) + a_0 + a_1(z - a) + \cdots$, so
$$ \operatorname{Res}(f, a) = \lim_{z\to a}(z - a)f(z). $$

Often $f = g/h$, where $g$ and $h$ are holomorphic at $a$  
and $h(a) = 0 \ne h'(a)$.  
Then $(z - a)f(z)$ is a $0/0$ form at $a$, and L'Hôpital's rule gives
$$ \lim_{z\to a}\frac{(z - a)g(z)}{h(z)} = \lim_{z\to a}\frac{g(z) + (z - a)g'(z)}{h'(z)} = \frac{g(a)}{h'(a)}. $$

So
$$ \operatorname{Res}\Bigl(\frac gh, a\Bigr) = \frac{g(a)}{h'(a)}. $$
Since this limit exists, $f$ has at most a simple pole at $a$;  
if $g(a) = 0$, the singularity is removable and both sides are $0$.

For example, at a root $\omega$ of $z^n + 1$,  
$\operatorname{Res}\bigl(1/(z^n + 1), \omega\bigr) = 1/n\omega^{n-1}$,  
with no need to factor $z^n + 1$.  
We use this for the sector below.

**Caution.** The limit of $(z - a)f(z)$ is the residue only at a simple pole.  
Suppose instead that $h$ has a zero of order $2$ at $a$ and $g(a) \ne 0$, so that
$$ f = \frac gh = \frac{a_{-2}}{(z - a)^2} + \frac{a_{-1}}{z - a} + a_0 + \cdots,\qquad a_{-2} \ne 0. $$
Now L'Hôpital's rule turns $(z - a)g(z)/h(z)$ into $g(a)/0$,  
and $(z - a)f(z) \to \infty$.

Multiplying by $(z - a)^2$ instead, L'Hôpital's rule applies twice;  
since $\bigl((z - a)^2 g\bigr)'' = 2g(a)$ at $a$,  
it gives the leading coefficient, not the residue:
$$ a_{-2} = \lim_{z\to a}\frac{(z - a)^2 g(z)}{h(z)} = \frac{2g(a)}{h''(a)}. $$

The residue comes only after we peel off this term:
$$ \begin{aligned}
a_{-1} &= \lim_{z\to a}\frac{(z - a)^2 f(z) - a_{-2}}{z - a} \\
&= \lim_{z\to a}\frac{(z - a)^2 g(z) - a_{-2}h(z)}{(z - a)h(z)}.
\end{aligned} $$

The numerator and the denominator both vanish to order $3$ at $a$,  
so we apply L'Hôpital's rule three times.  
At $a$, the third derivatives are  
$\bigl((z - a)^2 g\bigr)''' = 6g'(a)$ and $\bigl((z - a)h\bigr)''' = 3h''(a)$, so
$$ a_{-1} = \frac{6g'(a) - a_{-2}h'''(a)}{3h''(a)} = \frac{2g'(a)}{h''(a)} - \frac{2g(a)h'''(a)}{3h''(a)^2}. $$

For example, $1/\bigl(z^2(z - 1)\bigr)$ at $0$ has $g = 1$ and $h = z^3 - z^2$,  
with $h''(0) = -2$ and $h''' = 6$,  
so its residue at $0$ is $-2 \cdot 6/(3 \cdot 4) = -1$, as computed in 1016.

Limits see the most singular term $a_{-m}(z - a)^{-m}$ first,  
while the residue is the least singular term of the principal part;  
the two coincide only at a simple pole.  
Peeling off $a_{-m}, \ldots, a_{-2}$ one at a time is laborious,  
and the derivative formula does it all at once.

At a pole of order $m$, in general,
$$ \operatorname{Res}(f, a) = \frac{1}{(m - 1)!}\lim_{z\to a}\frac{d^{m-1}}{dz^{m-1}}\bigl((z - a)^m f(z)\bigr). $$

For example, inside $|z| = 2$ the function $z/\bigl((z - 1)^2(z + 3)\bigr)$  
has only the pole $1$, of order $2$, so
$$ \begin{aligned}
\oint_{|z|=2}\frac{z\,dz}{(z - 1)^2(z + 3)} &= 2\pi i\left.\frac{d}{dz}\frac{z}{z + 3}\right|_{z=1} \\
&= 2\pi i \cdot \frac{3}{16} = \frac{3\pi i}{8}.
\end{aligned} $$

## Trigonometric Integrals

### The substitution $z = e^{i\theta}$

As $\theta$ runs from $0$ to $2\pi$,  
$z = e^{i\theta}$ runs once counterclockwise around the unit circle, and
$$ dz = iz\,d\theta,\qquad \cos\theta = \frac{z + z^{-1}}{2},\qquad \sin\theta = \frac{z - z^{-1}}{2i}. $$

Therefore
$$ \begin{aligned}
&\int_0^{2\pi}F(\cos\theta, \sin\theta)\,d\theta \\
&\quad = \oint_{|z|=1}F\Bigl(\frac{z + z^{-1}}{2}, \frac{z - z^{-1}}{2i}\Bigr)\frac{dz}{iz}.
\end{aligned} $$

The new integrand is a rational function of $z$,  
so the residue theorem evaluates the integral  
once we know its poles inside the unit circle.

Recall from the 1009 homework that this substitution gives
$$ \int_0^{2\pi}\frac{d\theta}{2 + \cos\theta} = \frac{2\pi}{\sqrt3}. $$
There only one simple pole lay inside the unit circle,  
and Cauchy's integral formula was enough.  
The residue theorem handles any number of poles, of any order.

### A family of examples

Let $a > 1$.  
Substituting $z = e^{i\theta}$ gives
$$ \begin{aligned}
\int_0^{2\pi}\frac{d\theta}{a + \cos\theta} &= \oint_{|z|=1}\frac{1}{a + (z + z^{-1})/2} \cdot \frac{dz}{iz} \\
&= \frac2i\oint_{|z|=1}\frac{dz}{z^2 + 2az + 1}.
\end{aligned} $$

The roots of $z^2 + 2az + 1$ are
$$ z_\pm = -a \pm \sqrt{a^2 - 1}. $$
Both are real, their product is $1$, and $|z_-| > a > 1$,  
so $z_+$ lies inside the unit circle and $z_-$ lies outside.

The pole $z_+$ is simple,  
with residue $1/(z_+ - z_-) = 1/\bigl(2\sqrt{a^2 - 1}\bigr)$, so
$$ \int_0^{2\pi}\frac{d\theta}{a + \cos\theta} = \frac2i \cdot 2\pi i \cdot \frac{1}{2\sqrt{a^2 - 1}} = \frac{2\pi}{\sqrt{a^2 - 1}}. $$
For $a = 2$, this is the value from the 1009 homework.

**A pole of order $2$.**  
In the same way,
$$ \int_0^{2\pi}\frac{d\theta}{(a + \cos\theta)^2} = \frac4i\oint_{|z|=1}\frac{z\,dz}{(z - z_+)^2(z - z_-)^2}, $$
and the integrand has a pole of order $2$ at $z_+$.

Its residue there is
$$ \left.\frac{d}{dz}\frac{z}{(z - z_-)^2}\right|_{z=z_+} = -\frac{z_+ + z_-}{(z_+ - z_-)^3} = \frac{2a}{8(a^2 - 1)^{3/2}}. $$

So, by the residue theorem,
$$ \int_0^{2\pi}\frac{d\theta}{(a + \cos\theta)^2} = \frac4i \cdot 2\pi i \cdot \frac{a}{4(a^2 - 1)^{3/2}} = \frac{2\pi a}{(a^2 - 1)^{3/2}}. $$
Differentiating the previous formula with respect to $a$  
gives the same result.

## Improper Integrals of Rational Functions

### Convergence and the principal value

The improper integral $\int_{-\infty}^\infty f(x)\,dx$  
is defined by two independent limits:
$$ \int_{-\infty}^\infty f(x)\,dx = \lim_{r\to\infty}\int_{-r}^0 f(x)\,dx + \lim_{R\to\infty}\int_0^R f(x)\,dx. $$
It converges when both limits exist.

The **Cauchy principal value** takes a single symmetric limit:
$$ \operatorname{P.V.}\int_{-\infty}^\infty f(x)\,dx = \lim_{R\to\infty}\int_{-R}^R f(x)\,dx. $$
If the integral converges, its principal value exists and equals it.

The converse fails:  
$\int_{-\infty}^\infty x\,dx$ diverges,  
but $\int_{-R}^R x\,dx = 0$ for every $R$, so its principal value is $0$.  
The same happens for every odd function, such as $x/(x^2 + 1)$.

![Blackboard: the convergent integral of 1/(1 + x²) plus the principal value of x/(1 + x²) is a principal value, not a convergent integral](1120/convergent-plus-principal-value.jpeg)

Let $f = P/Q$ with polynomials $P$ and $Q$  
such that $\deg Q \ge \deg P + 2$,  
and suppose that $Q$ has no real roots.  
Then $|f(x)| \le K/x^2$ for large $|x|$,  
so $\int_{-\infty}^\infty f(x)\,dx$ converges absolutely,  
and we may compute it as a principal value.

### Closing the contour with a semicircle

Let $C_R$ be the upper semicircle $z = Re^{i\theta}$, $0 \le \theta \le \pi$,  
and close it with the segment $[-R, R]$.  
If $R$ exceeds the modulus of every pole of $f$,  
the closed contour encloses exactly the poles in the upper half-plane, so
$$ \int_{-R}^R f(x)\,dx + \int_{C_R}f(z)\,dz = 2\pi i\sum_{\operatorname{Im}a>0}\operatorname{Res}(f, a). $$

```python
R = 3
G = line([(-R, 0), (R, 0)], color='blue', thickness=2)
G += arc((0, 0), R, sector=(0, pi), color='blue', thickness=2)
G += arrow((1, 0), (1.6, 0), color='blue', width=2)
G += arrow((R*cos(0.9), R*sin(0.9)), (R*cos(1.1), R*sin(1.1)), color='blue', width=2)
G += point([(0, 1), (0, 2)], size=40, color='black', zorder=10)
G += point([(0, -1), (0, -2)], size=40, color='gray', zorder=10)
G += text('$C_R$', (2.5, 2.3), fontsize=14, color='blue')
G.show(aspect_ratio=1)
```

The picture shows $f(z) = 1/\bigl((z^2 + 1)(z^2 + 4)\bigr)$ and $R = 3$:  
the black poles $i$ and $2i$ lie inside the contour,  
and the gray poles $-i$ and $-2i$ do not.

### The integral over the big semicircle

**Lemma.**  
If $\deg Q \ge \deg P + 2$,  
then $\int_{C_R}P(z)/Q(z)\,dz \to 0$ as $R \to \infty$.

Recall from the 1002 homework that a polynomial is large far away:  
if $Q$ has degree $m$ and leading coefficient $q_m$,  
then $|Q(z)| \ge \frac12|q_m||z|^m$ for all large $|z|$.  
In the same way, $|P(z)| \le 2|p_n||z|^n$,  
where $n = \deg P$ and $p_n$ is the leading coefficient.  
So $|f(z)| \le K/|z|^2$ for all large $|z|$, with $K = 4|p_n|/|q_m|$.

The ML-inequality then gives
$$ \Bigl|\int_{C_R}f(z)\,dz\Bigr| \le \pi R \cdot \frac{K}{R^2} = \frac{\pi K}{R} \to 0. $$
Letting $R \to \infty$ in the contour identity proves the following theorem.

**Theorem.**  
Let $f = P/Q$ with $\deg Q \ge \deg P + 2$  
and no poles on the real axis. Then
$$ \boxed{\int_{-\infty}^\infty f(x)\,dx = 2\pi i\sum_{\operatorname{Im}a>0}\operatorname{Res}(f, a)}. $$

The segment $[-R, R]$ grows symmetrically,  
so the contour always computes the principal value at $\infty$,  
$\lim_{R\to\infty}\int_{-R}^R f(x)\,dx$.  
When $\deg Q \ge \deg P + 2$, the integral converges,  
so this principal value is the integral itself.

Closing the contour in the lower half-plane works as well,  
but the lower semicircle runs clockwise around the poles there,  
so the right side becomes $-2\pi i\sum_{\operatorname{Im}a<0}\operatorname{Res}(f, a)$.

### Examples

**Two simple poles.**  
For $f(z) = 1/\bigl((z^2 + 1)(z^2 + 4)\bigr)$,  
the poles in the upper half-plane are $i$ and $2i$, with
$$ \operatorname{Res}(f, i) = \frac{1}{2i \cdot 3} = \frac{1}{6i},\qquad \operatorname{Res}(f, 2i) = \frac{1}{(-3) \cdot 4i} = -\frac{1}{12i}. $$

Hence
$$ \int_{-\infty}^\infty\frac{dx}{(x^2 + 1)(x^2 + 4)} = 2\pi i\Bigl(\frac{1}{6i} - \frac{1}{12i}\Bigr) = \frac\pi6. $$
Partial fractions confirm it:  
the integrand is $\frac13\bigl(1/(x^2 + 1) - 1/(x^2 + 4)\bigr)$,  
and $\frac13(\pi - \pi/2) = \pi/6$.

**A sector instead of a semicircle.**  
Let $n \ge 2$.  
To evaluate $\int_0^\infty dx/(x^n + 1)$,  
integrate $f(z) = 1/(z^n + 1)$ around the boundary of the sector  
$0 \le |z| \le R$, $0 \le \arg z \le 2\pi/n$.

```python
n, R = 5, 1.6
G = line([(0, 0), (R, 0)], color='blue', thickness=2)
G += arc((0, 0), R, sector=(0, 2*pi/n), color='blue', thickness=2)
G += line([(R*cos(2*pi/n), R*sin(2*pi/n)), (0, 0)], color='blue', thickness=2)
G += circle((0, 0), 1, color='lightgray', linestyle='--')
G += point([(cos((2*k + 1)*pi/n), sin((2*k + 1)*pi/n)) for k in range(1, n)], size=40, color='gray', zorder=10)
G += point((cos(pi/n), sin(pi/n)), size=40, color='black', zorder=10)
G.show(aspect_ratio=1)
```

For $n = 5$, the poles of $f$ are the five fifth roots of $-1$ on the unit circle,  
and only $e^{i\pi/5}$ lies inside the sector.

On the far edge, $z = te^{2\pi i/n}$ with $t$ running from $R$ down to $0$,  
and $z^n = t^n$, so
$$ \int_{\text{edge}}f(z)\,dz = -e^{2\pi i/n}\int_0^R\frac{dt}{t^n + 1}. $$
The sector is chosen so that $z^n$ is real and positive on both straight edges.

![Blackboard: on the far edge of the sector, f(z) takes the same values as on the real axis and dz becomes ω dx, so the two edges give (1 − ω) times the integral from 0 to R](1120/sector-far-edge-gains-factor-omega.jpeg)

On the arc, $|f(z)| \le 1/(R^n - 1)$, and the arc has length $2\pi R/n$,  
so the integral over the arc tends to $0$.  
The only pole inside is $\omega = e^{i\pi/n}$, and by the $g/h$ rule above,
$$ \operatorname{Res}(f, \omega) = \frac{1}{n\omega^{n-1}} = \frac{\omega}{n\omega^n} = -\frac\omega n. $$

Let $I = \int_0^\infty dx/(x^n + 1)$.  
Letting $R \to \infty$, the residue theorem gives
$$ (1 - \omega^2)I = -\frac{2\pi i\omega}{n}. $$

Since $\omega - \omega^{-1} = 2i\sin(\pi/n)$,
$$ I = \frac{2\pi i}{n(\omega - \omega^{-1})} = \frac{\pi}{n\sin(\pi/n)}. $$
For $n = 2$ this is $\pi/2$, the limit of $\arctan R$.

## Fourier Integrals

For $a > 0$, the **Fourier integrals**
$$ \int_{-\infty}^\infty f(x)\cos(ax)\,dx\qquad\text{and}\qquad \int_{-\infty}^\infty f(x)\sin(ax)\,dx $$
are the real and imaginary parts of $\int_{-\infty}^\infty f(x)e^{iax}\,dx$  
when $f$ is real on the real axis.  
They give the Fourier transform of $f$ at the frequency $a$.

### Why $e^{iaz}$ and not $\cos(az)$

On the upper semicircle, $\cos(az)$ is useless:  
at $z = iy$ it equals $\cosh(ay)$, which grows exponentially.  
But
$$ |e^{iaz}| = |e^{ia(x+iy)}| = e^{-ay} \le 1\qquad(y \ge 0), $$
so $e^{iaz}$ is bounded in the upper half-plane.

So we integrate $f(z)e^{iaz}$ around the semicircle contour,  
and take real and imaginary parts at the end.  
If $\deg Q \ge \deg P + 2$, the lemma on the big semicircle still applies,  
because $|e^{iaz}| \le 1$ on $C_R$.

### Jordan's lemma

When $\deg Q = \deg P + 1$, as for $x\sin(x)/(x^2 + 1)$,  
the ML-inequality only gives the bound $\pi R \cdot K/R = \pi K$,  
which does not tend to $0$.  
We must use that $e^{iaz}$ decays away from the real axis.

**Jordan's inequality.**  
For $0 \le \theta \le \pi/2$,
$$ \sin\theta \ge \frac{2\theta}{\pi}, $$
because $\sin$ is concave on $[0, \pi/2]$,  
and the right side is its chord from $(0, 0)$ to $(\pi/2, 1)$.

```python
var('t')
G = plot(sin(t), (t, 0, pi), color='blue', legend_label=r'$\sin\theta$')
G += plot(2*t/pi, (t, 0, pi/2), color='red', legend_label=r'$2\theta/\pi$')
G += plot(2 - 2*t/pi, (t, pi/2, pi), color='red')
G.show(aspect_ratio=1)
```

By the symmetry $\sin(\pi - \theta) = \sin\theta$,
$$ \begin{aligned}
\int_0^\pi e^{-aR\sin\theta}\,d\theta &= 2\int_0^{\pi/2}e^{-aR\sin\theta}\,d\theta \\
&\le 2\int_0^{\pi/2}e^{-2aR\theta/\pi}\,d\theta < \frac{\pi}{aR}.
\end{aligned} $$

**Jordan's lemma.**  
Let $a > 0$, and suppose that $|f(z)| \le M_R$ on $C_R$,  
where $M_R \to 0$ as $R \to \infty$.  
Since $|e^{iaz}| = e^{-aR\sin\theta}$ and $|dz| = R\,d\theta$ on $C_R$,
$$ \Bigl|\int_{C_R}f(z)e^{iaz}\,dz\Bigr| \le \int_0^\pi M_R e^{-aR\sin\theta}R\,d\theta < \frac{\pi M_R}{a} \to 0. $$

For $f = P/Q$ with $\deg Q \ge \deg P + 1$, we may take $M_R = K/R$.  
Hence, if $f$ has no poles on the real axis,
$$ \operatorname{P.V.}\int_{-\infty}^\infty f(x)e^{iax}\,dx = 2\pi i\sum_{\operatorname{Im}z_k>0}\operatorname{Res}\bigl(f(z)e^{iaz}, z_k\bigr). $$

When $\deg Q = \deg P + 1$, the integral converges only conditionally.  
One can show by integrating by parts that it does converge,  
so the principal value is its value.

![Blackboard: x/(x² + 1) decays like 1/x, so its integral diverges, while the oscillation of x sin(x)/(x² + 1) makes its integral converge conditionally](1120/slow-tail-versus-oscillating-tail.jpeg)

### Examples

**One simple pole.**  
Let $a \ge 0$ and $b > 0$.  
The function $e^{iaz}/(z^2 + b^2)$ has one pole $ib$ in the upper half-plane,  
with residue $e^{-ab}/(2ib)$, so
$$ \int_{-\infty}^\infty\frac{\cos(ax)}{x^2 + b^2}\,dx = \operatorname{Re}\Bigl(2\pi i \cdot \frac{e^{-ab}}{2ib}\Bigr) = \frac{\pi}{b}e^{-ab}. $$

The integral with $\sin(ax)$ is $0$, as it must be for an odd integrand.  
The value decays exponentially in the frequency $a$:  
$1/(z^2 + b^2)$ is holomorphic in the strip $|\operatorname{Im}z| < b$,  
and the wider such a strip is, the faster the Fourier transform decays.

**Jordan's lemma at work.**  
For $a > 0$ and $b > 0$, the function $ze^{iaz}/(z^2 + b^2)$  
has residue $ibe^{-ab}/(2ib) = e^{-ab}/2$ at $ib$.  
Here $\deg Q = \deg P + 1$, so by Jordan's lemma
$$ \int_{-\infty}^\infty\frac{x\sin(ax)}{x^2 + b^2}\,dx = \operatorname{Im}\Bigl(2\pi i \cdot \frac{e^{-ab}}{2}\Bigr) = \pi e^{-ab}. $$

As $b \to 0$, the integrand tends to $\sin(ax)/x$, and the value tends to $\pi$.  
This suggests that $\int_{-\infty}^\infty\sin(ax)/x\,dx = \pi$,  
but $e^{iaz}/z$ has its pole on the real axis,  
and the semicircle contour no longer applies.

## Indented Contours

### Principal value at a pole

If $f$ has a pole at a real point $c$, the integral of $f$ across $c$ usually diverges.  
Its **principal value** removes a symmetric interval around $c$:
$$ \operatorname{P.V.}\int_{-\infty}^\infty f(x)\,dx = \lim_{r\to0,\ R\to\infty}\Bigl(\int_{-R}^{c-r} + \int_{c+r}^R\Bigr)f(x)\,dx. $$
For example, $\operatorname{P.V.}\int_{-1}^1 dx/x = 0$, although $\int_0^1 dx/x$ diverges.

![Blackboard: the principal value of the integral of 1/x³ over [−1, 1] removes the interval (−ε, ε)](1120/principal-value-removes-symmetric-interval.jpeg)

To compute it, replace the segment $[c - r, c + r]$  
by a small semicircle above $c$, which detours around the pole.  
We say that the contour is **indented** at $c$.

```python
R, r = 3, 0.6
G = line([(-R, 0), (-r, 0)], color='blue', thickness=2) + line([(r, 0), (R, 0)], color='blue', thickness=2)
G += arc((0, 0), R, sector=(0, pi), color='blue', thickness=2)
G += arc((0, 0), r, sector=(0, pi), color='red', thickness=2)
G += point((0, 0), size=40, color='black', zorder=10)
G += text('$C_R$', (2.5, 2.3), fontsize=14, color='blue')
G += text('$C_r$', (0.75, 0.65), fontsize=14, color='red')
G.show(aspect_ratio=1)
```

Let $C_r$ be the semicircle $z = c + re^{i\theta}$, $0 \le \theta \le \pi$,  
shown in red with $c = 0$.  
As a curve, $C_r$ runs counterclockwise from $c + r$ to $c - r$;  
the indented contour runs along it in the opposite direction.

### Small arcs around a simple pole

Let $f$ have a simple pole at $c$.

**Half-residue lemma.**  
If $c$ is real and $C_r$ is the semicircle above, then
$$ \lim_{r\to0}\int_{C_r}f(z)\,dz = \pi i\operatorname{Res}(f, c). $$

**Small-arc lemma.**  
For $0 < \alpha \le 2\pi$, let $A_r$ be the arc $z = c + re^{i\theta}$  
with $\theta_0 \le \theta \le \theta_0 + \alpha$. Then
$$ \lim_{r\to0}\int_{A_r}f(z)\,dz = i\alpha\operatorname{Res}(f, c). $$

It suffices to prove the small-arc lemma, since $\alpha = \pi$ gives the other.  
Near $c$, write $f(z) = a_{-1}/(z - c) + g(z)$,  
where $a_{-1} = \operatorname{Res}(f, c)$ and $g$ is holomorphic, hence bounded by some $M$.

Then
$$ \int_{A_r}\frac{a_{-1}}{z - c}\,dz = \int_{\theta_0}^{\theta_0+\alpha}\frac{a_{-1}}{re^{i\theta}}\,ire^{i\theta}\,d\theta = i\alpha a_{-1}, $$
while $\bigl|\int_{A_r}g(z)\,dz\bigr| \le \alpha rM \to 0$.

Only the angle of the arc matters, not where it starts.  
The full circle, $\alpha = 2\pi$, gives $2\pi i\operatorname{Res}(f, c)$, as in the residue theorem.

Likewise, suppose a positively oriented contour  
has a right-angle corner at a simple pole $c$.  
Then indenting it by a quarter circle that keeps $c$ outside  
contributes $-\frac{\pi}{2}i\operatorname{Res}(f, c)$ in the limit,  
since the quarter circle runs clockwise around $c$.

Both lemmas fail for poles of higher order:  
for $f(z) = 1/z^2$ and $c = 0$,  
the integral over $C_r$ is $\bigl[-1/z\bigr]_r^{-r} = 2/r$, which blows up.

### Poles on the axis and above it

**Theorem.**  
Let $f = P/Q$ with $\deg Q \ge \deg P + 2$,  
and suppose that every pole of $f$ on the real axis is simple. Then
$$ \begin{aligned}
&\operatorname{P.V.}\int_{-\infty}^\infty f(x)\,dx \\
&\quad = 2\pi i\sum_{\operatorname{Im}a>0}\operatorname{Res}(f, a) + \pi i\sum_{\operatorname{Im}a=0}\operatorname{Res}(f, a).
\end{aligned} $$

Indent the contour at each real pole $c$.  
The indented contour encloses exactly the poles in the upper half-plane,  
and it runs along each small semicircle clockwise,  
which contributes $-\pi i\operatorname{Res}(f, c)$ in the limit.  
Moving these terms to the other side gives the theorem.

By Jordan's lemma, the same holds for $f(z)e^{iaz}$ with $a > 0$  
whenever $\deg Q \ge \deg P + 1$.

### Examples

**A pole on the axis.**  
For $f(z) = 1/\bigl((z - 1)(z^2 + 1)\bigr)$,
$$ \operatorname{Res}(f, i) = \frac{1}{(i - 1) \cdot 2i} = \frac{-1 + i}{4},\qquad \operatorname{Res}(f, 1) = \frac12. $$

Therefore
$$ \operatorname{P.V.}\int_{-\infty}^\infty\frac{dx}{(x - 1)(x^2 + 1)} = 2\pi i \cdot \frac{-1 + i}{4} + \pi i \cdot \frac12 = -\frac\pi2. $$

Partial fractions confirm it: the integrand is
$$ \frac12\Bigl(\frac{1}{x - 1} - \frac{x}{x^2 + 1} - \frac{1}{x^2 + 1}\Bigr), $$
whose first two terms have principal value $0$,  
and whose last term integrates to $-\pi/2$.

**The Dirichlet integral.**  
The function $e^{iz}/z$ has a simple pole at $0$ with residue $1$,  
and no poles in the upper half-plane.  
By the theorem, with Jordan's lemma for $M_R = 1/R$,
$$ \operatorname{P.V.}\int_{-\infty}^\infty\frac{e^{ix}}{x}\,dx = \pi i. $$

Since $\cos(x)/x$ is odd and $\sin(x)/x$ is even, the imaginary part gives
$$ \int_0^\infty\frac{\sin x}{x}\,dx = \frac\pi2. $$

Here no principal value is needed:  
$\sin(x)/x$ extends continuously to $x = 0$,  
and the integral converges at $\infty$, though only conditionally.  
For every $a > 0$, substituting $x = t/a$ gives  
$\int_{-\infty}^\infty\sin(ax)/x\,dx = \pi$, as guessed above.

### Summary

| Integral | Contour | Value |
|---|---|---|
| $\int_0^{2\pi}F(\cos\theta, \sin\theta)\,d\theta$ | unit circle, $z = e^{i\theta}$ | $2\pi i$ times the residues inside $\lvert z\rvert = 1$ |
| $\int_{-\infty}^\infty f(x)\,dx$, $\deg Q \ge \deg P + 2$ | upper semicircle | $2\pi i\sum_{\operatorname{Im}z_k>0}\operatorname{Res}(f, z_k)$ |
| $\int_{-\infty}^\infty f(x)e^{iax}\,dx$, $a > 0$, $\deg Q \ge \deg P + 1$ | upper semicircle, Jordan's lemma | $2\pi i\sum_{\operatorname{Im}z_k>0}\operatorname{Res}\bigl(f(z)e^{iaz}, z_k\bigr)$ |
| simple poles on the real axis | indented contour | add $\pi i\operatorname{Res}$ at each of them |

```python
import numpy as np
import matplotlib.pyplot as plt

def sphere_axes(lim=2):
    ax = plt.figure(figsize=(7, 5)).add_subplot(projection='3d')
    u, v = np.meshgrid(np.linspace(0, 2*np.pi, 37), np.linspace(0, np.pi, 19))
    ax.plot_wireframe(np.cos(u)*np.sin(v), np.sin(u)*np.sin(v), np.cos(v), color='gray', linewidth=0.3)
    X, Y = np.meshgrid([-lim, lim], [-lim, lim])
    ax.plot_surface(X, Y, 0*X, color='lightblue', alpha=0.2)  # the plane x_3 = 0
    ax.scatter(0, 0, 1, color='black', s=20); ax.text(0.05, 0, 1.1, '$N$')
    ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim); ax.set_zlim(-1, 1)
    ax.set_box_aspect((1, 1, 0.5), zoom=1.3); ax.set_axis_off(); ax.view_init(elev=18, azim=-60)
    return ax
```

```python
def sigma(p):
    return (p[0] + 1j*p[1]) / (1 - p[2])

def sigma_inv(z):
    return np.array([2*z.real, 2*z.imag, abs(z)**2 - 1]) / (abs(z)**2 + 1)
```
