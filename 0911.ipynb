## Why Were Complex Numbers Worth Studying?

### A Historical Reason: Solving Cubic Equations

During the sixteenth century,  
Italian mathematicians developed formulas for solving cubic equations.  
Work by del Ferro and Tartaglia led to the formula published by Cardano in *Ars Magna* (1545).  
Bombelli's *Algebra* (1572) developed rules for calculating with complex quantities  
and showed how they could yield real solutions of cubics.

The striking motivation was this:  
**a formula for a real equation could pass through square roots of negative numbers**  
**and still produce a real answer.**

https://mathshistory.st-andrews.ac.uk/HistTopics/Quadratic_etc_equations/.

### Cardano's Formula

For the **depressed cubic** (a cubic with no quadratic term)
$$ x^3 + px + q = 0, $$
set
$$ \Delta = \left(\frac q2\right)^2 + \left(\frac p3\right)^3. $$

One solution is
$$\boxed{x = \sqrt[3]{-\frac q2 + \sqrt\Delta} + \sqrt[3]{-\frac q2 - \sqrt\Delta}}.$$

A general cubic $Ax^3 + Bx^2 + Cx + D = 0$, $A \ne 0$,  
reduces to this form by substituting $x = t - B/3A$.  
We will use the formula today without deriving it.

### Example 1: $\Delta > 0$

Consider
$$ x^3 + 6x - 7 = 0. $$
Cardano's formula gives
$$ p = 6, \qquad q = -7, \qquad \Delta = \frac{49}{4} + 8 = \frac{81}{4} > 0. $$

Substitute into the formula:
$$ x = \sqrt[3]{\frac72 + \frac92} + \sqrt[3]{\frac72 - \frac92} = \sqrt[3]{8} + \sqrt[3]{-1} = 2 - 1 = \boxed{1}. $$
All the intermediate calculations use real numbers.  
In parituclar, the cube roots are real.

We can check the answer directly:
$$ 1^3 + 6\cdot 1 - 7 = 0. $$


Dividing the original equation by $x - 1$ gives
$$ x^3 + 6x - 7 = (x - 1)(x^2 + x + 7). $$
The quadratic factor has discriminant $1 - 28 = -27$.  
So $1$ is the only real root.

Nowadays we can use quadratic formula to find the complex roots.  
But back then, when $\Delta > 0$,  
people just accept that there is only one root.  
this is **not** the motivation for complex numbers.

### Example 2: $\Delta = 0$

Construct a cubic with integer roots $2, -1, -1$:
$$ (x - 2)(x + 1)^2 = x^3 - 3x - 2. $$

Suppose we are given only the expanded equation $x^3 - 3x - 2 = 0$.  
Cardano's formula gives
$$ p = -3, \qquad q = -2, \qquad \Delta = 1 - 1 = 0, $$
so
$$ x = \sqrt[3]{1} + \sqrt[3]{1} = 2. $$

Divide by $x - 2$ to get
$$ x^2 + 2x + 1 = (x + 1)^2. $$
The roots, counted with multiplicity, are
$$ \boxed{2, \ -1, \ -1}. $$

The formula finds a root from the coefficients,  
without first guessing a factor.  
This gives us some confidence for the usefulness of Cardano's formula.

### Example 3: $\Delta < 0$

Construct a cubic with three distinct integer roots $4, 1, -5$:
$$ (x - 4)(x - 1)(x + 5) = x^3 - 21x + 20. $$

Apply the same formula to the expanded equation:
$$ p = -21, \qquad q = 20, \qquad \Delta = 10^2 + (-7)^3 = -243. $$
It asks us to calculate
$$ x = \sqrt[3]{-10 + 9\sqrt 3\,i} + \sqrt[3]{-10 - 9\sqrt 3\,i}, \qquad i^2 = -1. $$

Use the ordinary algebraic rules and $i^2 = -1$.  
We can check that
$$ (2 + i\sqrt 3)^3 = -10 + 9i\sqrt 3, \qquad (2 - i\sqrt 3)^3 = -10 - 9i\sqrt 3. $$

The two cube roots are compatible:
$$ (2 + i\sqrt 3)(2 - i\sqrt 3) = 7 = -p/3. $$
Therefore
$$ x = (2 + i\sqrt 3) + (2 - i\sqrt 3) = \boxed{4}. $$

Dividing the cubic by $x - 4$ gives $x^2 + 4x - 5 = (x - 1)(x + 5)$,  
so the other roots are $1$ and $-5$.

**The complex terms cancel, leaving a real solution.**  
All the answers are integers,  
but the formula takes us through complex numbers.

## Complex Numbers: Algebra and Geometry

### The Form $a + bi$

A complex number has the form
$$ z = a + bi, \qquad a, b\in \mathbb R, \qquad i^2 = -1. $$
We write $\mathbb C = \{a + bi : a, b\in \mathbb R\}$.  
The representation is unique:
$$ a + bi = c + di\quad \Longleftrightarrow \quad a = c\text{ and }b = d. $$

The **real part** is $\operatorname{Re}z = a$.  
The **imaginary part** is $\operatorname{Im}z = b$.  
The imaginary part is a real number, not $bi$.

Real numbers correspond to $b = 0$.  
Purely imaginary numbers have $a = 0$.

### Arithmetic and Complex Conjugation

For $z = a + bi$ and $w = c + di$,
$$ z + w = (a + c) + (b + d)i, \qquad z - w = (a - c) + (b - d)i, $$
$$ zw = (ac - bd) + (ad + bc)i. $$

The **complex conjugate** of $z$ is $\bar z = a - bi$.  
It satisfies
$$ z\bar z = a^2 + b^2, \qquad \overline{z + w} = \bar z + \bar w, \qquad \overline{zw} = \bar z\,\bar w. $$

For $w\ne 0$, multiply numerator and denominator by $\bar w$:
$$ \frac zw = \frac{z\bar w}{w\bar w} = \frac{ac + bd}{c^2 + d^2} + \frac{bc - ad}{c^2 + d^2}i. $$

For example,
$$ (1 + 2i)(3 - i) = 5 + 5i, \qquad \frac{1 + 2i}{3 - i} = \frac{1 + 7i}{10}. $$
(Powers of $i$ repeat every four steps: $1, i, -1, -i, 1, \ldots$.)

### A Representation by Real Matrices

Associate $z = a + bi$ with
$$ M(z) = \begin{pmatrix}a & -b \\ b & a\end{pmatrix}. $$
Then
$$ M(z + w) = M(z) + M(w), \qquad M(zw) = M(z)M(w), \qquad M(1) = I. $$

In particular,
$$ M(i) = J = \begin{pmatrix}0 & -1 \\ 1 & 0\end{pmatrix}, \qquad J^2 = -I. $$
Thus the rule $i^2 = -1$ has a concrete realization using real matrices.

For $z = a + bi$ and $w = c + di$, multiply their matrix representations:
$$ M(z)M(w) = \begin{pmatrix}a & -b \\ b & a\end{pmatrix}\begin{pmatrix}c & -d \\ d & c\end{pmatrix} = \begin{pmatrix}ac - bd & -(ad + bc) \\ ad + bc & ac - bd\end{pmatrix}. $$
Since $zw = (ac - bd) + (ad + bc)i$,  
this is exactly $M(zw)$.

Also, $\det M(z) = a^2 + b^2$, so $M(z)$ is invertible exactly when $z\ne 0$.

### What Happens If We Diagonalize This Matrix?

The characteristic polynomial is
$$ \det(\lambda I - M(z)) = (\lambda - a)^2 + b^2. $$
Its eigenvalues are $a + bi = z$ and $a - bi = \bar z$.  
If $b\ne 0$, we cannot diagonalize over $\mathbb R$.  
We can diagonalize over $\mathbb C$:

$$ P = \begin{pmatrix}1 & 1 \\ -i & i\end{pmatrix}, \qquad P^{-1} = \frac 12\begin{pmatrix}1 & i \\ 1 & -i\end{pmatrix}, $$
$$ \boxed{P^{-1}M(z)P = \begin{pmatrix}z & 0 \\ 0 & \bar z\end{pmatrix}}. $$
The same $P$ works for every $z$.  
If $b = 0$, then $M(z) = aI$ is already diagonal.

The same change of basis diagonalizes both $M(z)$ and $M(w)$:
$$ P^{-1}M(z)P = \begin{pmatrix}z & 0 \\ 0 & \bar z\end{pmatrix}, \qquad P^{-1}M(w)P = \begin{pmatrix}w & 0 \\ 0 & \bar w\end{pmatrix}. $$
Multiply the two diagonal matrices:
$$ \begin{pmatrix}z & 0 \\ 0 & \bar z\end{pmatrix}\begin{pmatrix}w & 0 \\ 0 & \bar w\end{pmatrix} = \begin{pmatrix}zw & 0 \\ 0 & \bar z\bar w\end{pmatrix} = \begin{pmatrix}zw & 0 \\ 0 & \overline{zw}\end{pmatrix}. $$
This is why
$$ (P^{-1}M(z)P)(P^{-1}M(w)P) = P^{-1}M(z)M(w)P = P^{-1}M(zw)P. $$

Two worlds run in parallel:

- **Primary world:** multiply $z$ and $w$ to get $zw$.
- **Conjugate world:** multiply $\bar z$ and $\bar w$ to get $\overline{zw}$.

Every multiplication in the primary world has a matching multiplication in the conjugate world.  
The diagonal form keeps the two calculations separate.

### A Vector Space over $\mathbb R$

As a real vector space, $\mathbb C$ is two-dimensional with basis $\{1, i\}$:
$$ a + bi\ \longleftrightarrow \ \begin{pmatrix}a \\ b \end{pmatrix}\in \mathbb R^2. $$
Addition is vector addition.  
Multiplication by a real number scales both coordinates.  
Thus $\mathbb C\cong \mathbb R^2$ **as real vector spaces**.

Complex multiplication gives us additional structure.  
It is not componentwise multiplication of vectors.  
Multiplication by a fixed $z$ is the real linear map $M(z)$ above.

The dimension depends on the scalar field:
$$ \dim_{\mathbb R}\mathbb C = 2, \qquad \dim_{\mathbb C}\mathbb C = 1. $$
As a complex vector space, $\mathbb C$ has basis $\{1\}$.

### Modulus and Distance

The **modulus** of $z = a + bi$ is its Euclidean length:
$$ |z| = \sqrt{a^2 + b^2} = \sqrt{z\bar z}. $$
The distance between $z$ and $w$ is $|z - w|$.  
For example, $|3 + 4i| = 5$.

Useful identities are
$$ |z|\ge 0, \qquad |z| = 0\Longleftrightarrow z = 0, \qquad |\bar z| = |z|, $$
$$ |zw| = |z||w|, \qquad \left|\frac zw\right| = \frac{|z|}{|w|}\quad (w\ne 0). $$

For instance,
$$ |zw|^2 = (zw)\overline{zw} = (z\bar z)(w\bar w) = |z|^2|w|^2. $$
Taking nonnegative square roots gives the multiplication identity.

### The Triangle Inequality

For all complex numbers $z, w$,
$$ \boxed{|z + w|\le |z| + |w|}. $$
The direct distance is no longer than taking two successive steps.

![alt text](image.png)

Algebraically, use $\operatorname{Re}(z\bar w)\le |z\bar w| = |z||w|$:
$$ \begin{aligned}
|z + w|^2 & = |z|^2 + 2\operatorname{Re}(z\bar w) + |w|^2 \\
& \le |z|^2 + 2|z||w| + |w|^2 = (|z| + |w|)^2.
\end{aligned} $$
Equality holds when either number is zero,  
or when the two nonzero numbers point in the same direction.

![alt text](image-1.png)`

Apply the triangle inequality to $z = (z - w) + w$,  
then interchange $z, w$.  
This gives the **reverse triangle inequality**:
$$ \bigl||z|-|w|\bigr|\le |z - w|. $$
These inequalities let us estimate complex expressions without calculating their exact values.

### Subsets of the complex plane

Write $z = x + iy$ and let $c = a + bi$ be a fixed point.  
We can specify a set by an equation or an inequality:
$$ \{ z\in\mathbb C : \text{a condition on }z \}. $$
Read this as "all complex numbers $z$ satisfying the condition."

Remember:
$$ |z - c| = \sqrt{(x - a)^2 + (y - b)^2}, \qquad \operatorname{Re}z = x, \qquad \operatorname{Im}z = y. $$
Modulus measures distance.  
Real and imaginary parts give coordinates.

#### Circles

A **circle** consists of the points at a fixed distance $R > 0$ from $c$:
$$ \boxed{\{ z\in\mathbb C : |z - c| = R \}}. $$
In Cartesian coordinates,
$$ (x - a)^2 + (y - b)^2 = R^2. $$
The circle is the boundary curve.  
It does not include the region inside.

For example,
$$ |z - (1 + 2i)| = 3 $$
is the circle with center $1 + 2i$ and radius $3$.

Be careful with signs:
$$ |z - 1 + 2i| = |z - (1 - 2i)|. $$
Here the center is $1 - 2i$.

#### Disks

An **open disk** contains the points whose distance from $c$ is less than $R$:
$$ \boxed{D(c, R) = \{ z\in\mathbb C : |z - c| < R \}}, \qquad R > 0. $$
A **closed disk** includes the boundary circle:
$$ \boxed{\overline{D(c, R)} = \{ z\in\mathbb C : |z - c| \le R \}}. $$
The bar here denotes the closure of a set.  
It is not complex conjugation.

For example,
$$ |z| < 1 $$
is the open unit disk, while
$$ |z| \le 1 $$
is the closed unit disk.

The unit circle is $|z| = 1$.  
A circle and a disk are different sets.

#### Annulus (= Ring)

An **annulus** is the region between two concentric circles.  
For $0 < r < R$, the open annulus is
$$ \boxed{\{ z\in\mathbb C : r < |z - c| < R \}}. $$
Both boundary circles are excluded.

The closed annulus is
$$ \{ z\in\mathbb C : r \le |z - c| \le R \}. $$
We can also include just one boundary, for example $r < |z - c| \le R$.

For example,
$$ 1 < |z + i| < 2 $$
is a ring centered at $-i$,  
with inner radius $1$ and outer radius $2$.

If the inner radius is zero, the condition
$$ 0 < |z - c| < R $$
describes a **punctured disk**: an open disk with its center removed.

#### Half-planes

A line divides the plane into two **half-planes**.

The upper half-plane is
$$ \{ z\in\mathbb C : \operatorname{Im}z > 0 \}. $$
The right half-plane is
$$ \{ z\in\mathbb C : \operatorname{Re}z > 0 \}. $$
These are open half-planes.  
Replacing $>$ by $\ge$ includes the boundary line and gives a closed half-plane.

The boundary can be any line.  
For real $A, B, C$, with $(A, B)\ne(0, 0)$,
$$ Ax + By = C \quad\Longleftrightarrow\quad \operatorname{Re}((A - iB)z) = C. $$
Therefore one side of the line is
$$ \boxed{\{ z\in\mathbb C : \operatorname{Re}((A - iB)z) > C \}}. $$
Changing $>$ to $<$ selects the other side.

For example,
$$ \operatorname{Re}((1 - i)z) > 1 \quad\Longleftrightarrow\quad x + y > 1. $$
The boundary is $x + y = 1$.  
The point $z = 2$ belongs to this half-plane; $z = 0$ does not.

We compare real quantities such as $|z|$, $\operatorname{Re}z$, and $\operatorname{Im}z$.  
An inequality such as $z > 1$ has no standard meaning for a general complex number.

#### Strips

A **strip** is the region between two parallel lines.

A horizontal open strip is
$$ \boxed{\{ z\in\mathbb C : a < \operatorname{Im}z < b \}}, \qquad a < b. $$
Its width is $b - a$.

A vertical open strip is
$$ \{ z\in\mathbb C : a < \operatorname{Re}z < b \}. $$
For example, $-1 < \operatorname{Re}z < 2$ is the vertical strip  
between $x = -1$ and $x = 2$, excluding both boundary lines.

For an inclined strip, use the same linear expression for both boundaries:
$$ \{ z\in\mathbb C : c_1 < \operatorname{Re}((A - iB)z) < c_2 \}, $$
where $A, B, c_1, c_2$ are real, $(A, B)\ne(0, 0)$, and $c_1 < c_2$.

For example,
$$ 1 < \operatorname{Re}((1 - i)z) < 3 \quad\Longleftrightarrow\quad 1 < x + y < 3. $$
The two boundary lines are $x + y = 1$ and $x + y = 3$.

Use $\le$ to include a boundary line.  
A strip is the intersection of two half-planes.

### Argument (an angle)

For $z = a + bi\ne 0$,  
an **argument** is any angle $\theta$ satisfying
$$ \cos \theta = \frac a{|z|}, \qquad \sin \theta = \frac b{|z|}. $$

If $\theta$ is an argument,  
so is $\theta + 2\pi k$ for every integer $k$.  
We use the principal argument convention
$$ \operatorname{Arg}z\in (-\pi, \pi], \qquad \arg z = \{\operatorname{Arg}z + 2\pi k : k\in \mathbb Z\}. $$

Examples:
$$ \operatorname{Arg}(1 + i) = \frac \pi 4, \quad \operatorname{Arg}(-1 + i) = \frac{3\pi}4, \quad \operatorname{Arg}(-1) = \pi. $$
Zero has no argument.  
The principal argument jumps across the negative real axis.  
Choosing a single angle introduces a discontinuity.

### Cartesian and Polar Coordinates

Cartesian coordinates use $(a, b)$.  
Polar coordinates use $(r, \theta)$:
$$ z = a + bi = r(\cos \theta + i\sin \theta), \qquad r = |z|. $$
The conversions are
$$ a = r\cos \theta, \qquad b = r\sin \theta, \qquad r = \sqrt{a^2 + b^2}. $$

For $z\ne 0$, use both coordinates to choose the correct quadrant.  
$\arctan(b/a)$ loses the quadrant information  
and is undefined when $a = 0$.  
A function such as $\operatorname{atan2}(b, a)$ keeps track of the quadrant.

For example,
$$ -1 + i = \sqrt 2\left(\cos \frac{3\pi}4 + i\sin \frac{3\pi}4\right). $$

Euler's formula is $e^{i\theta} = \cos \theta + i\sin \theta$.  
It lets us write $z = re^{i\theta}$.  
We will justify the complex exponential later.  
For now, the sine-and-cosine form gives the geometry.

### Multiplication Scales Lengths and Adds Angles

Let $z = r(\cos \alpha + i\sin \alpha)$  
and $w = s(\cos \beta + i\sin \beta)$.  
Expanding their product gives
$$ zw = rs\bigl[(\cos \alpha \cos \beta - \sin \alpha \sin \beta) + i(\sin \alpha \cos \beta + \cos \alpha \sin \beta)\bigr]. $$

The angle addition identities turn this into
$$ \boxed{zw = rs\bigl[\cos(\alpha + \beta) + i\sin(\alpha + \beta)\bigr]}. $$
Moduli multiply, and arguments add **modulo $2\pi$**.  
Division divides moduli and subtracts arguments.

Equivalently,
$$ M(z) = r\begin{pmatrix}\cos \alpha & -\sin \alpha \\ \sin \alpha & \cos \alpha \end{pmatrix}. $$
Multiplication by $z\ne 0$ is a rotation through $\alpha$ followed by scaling by $r$.  
In particular, multiplying by $i$ rotates a vector counterclockwise by $90^\circ$.

### Complex Multiplication and the Tangent Addition Formula

When $\tan \alpha$ and $\tan \beta$ are defined,
$$ (1 + i\tan \alpha)(1 + i\tan \beta) = (1 - \tan \alpha \tan \beta) + i(\tan \alpha + \tan \beta). $$

Divide the imaginary part by the real part:
$$ \boxed{\tan(\alpha + \beta) = \frac{\tan \alpha + \tan \beta}{1 - \tan \alpha \tan \beta}}, $$
provided $\cos \alpha \ne 0$, $\cos \beta \ne 0$, and $\cos(\alpha + \beta)\ne 0$.

For example,
$$ (1 + i)(1 + 2i) = -1 + 3i. $$
Thus, for $\alpha = \pi/4$ and $\beta = \arctan(2)$, the tangent of the sum is $(1 + 2)/(1 - 2) = -3$.  
The product lies in quadrant II, so the angle sum is $\pi - \arctan(3)$, not a negative angle.

**Tangent records a slope, so it determines an angle only modulo $\pi$.**  
The signs of both coordinates retain the quadrant information.

### Powers and Geometric Sequences

Repeated multiplication gives **de Moivre's formula**:
$$ \bigl[r(\cos \theta + i\sin \theta)\bigr]^n = r^n\bigl[\cos(n\theta) + i\sin(n\theta)\bigr], \qquad n\ge 1. $$

A complex geometric sequence has the form
$$ a_n = a_0q^n, \qquad n = 0, 1, 2, \ldots. $$

Suppose $a_0\ne 0$ and $q = \rho e^{i\theta}\ne 0$.  
Each step multiplies the length by $\rho$  
and rotates by $\theta$:

- $|q| < 1$: $a_n\to 0$.
- $|q| > 1$: $|a_n|\to \infty$.
- $|q| = 1$: the points stay on a circle.  
They are periodic exactly when $\theta/(2\pi)$ is rational.

For $a_0 = 1$, the ratio $q = i$ gives $1, i, -1, -i, 1, \ldots$.  
The ratio $q = i/2$ gives $1, i/2, -1/4, -i/8, \ldots$, rotating while shrinking toward zero.

If $a_0 = 0$, the sequence is identically zero.  
If $q = 0$, every term after the initial term is zero.

### Finite and Infinite Geometric Sums

For $S_N = 1 + q + q^2 + \cdots + q^N$, subtraction gives
$$ (1 - q)S_N = 1 - q^{N + 1}. $$
Therefore
$$ \boxed{\sum_{n = 0}^Nq^n = \frac{1 - q^{N + 1}}{1 - q}}\quad (q\ne 1). $$

If $q = 1$, the sum is $N + 1$.  
If $|q| < 1$, then $q^{N + 1}\to 0$, so
$$ \boxed{\sum_{n = 0}^{\infty}q^n = \frac 1{1 - q}}. $$
For $|q|\ge 1$, the terms $q^n$ do not tend to zero, so the infinite series diverges.

For example,
$$ 1 + \frac i2 - \frac 14 - \frac i8 + \cdots = \frac 1{1 - i/2} = \frac 45 + \frac 25i. $$
This same geometric-series identity will later expand the kernel in Cauchy's integral formula and produce Taylor series.

## Why Learn Complex Analysis 2?

Complex numbers turn rotations and oscillations into algebra.  
Complex analysis connects derivatives, power series, contour integrals, and geometry.

We can use these connections to study  
signals, circuits, differential equations, and two-dimensional fields.

### From Oscillations to Engineering Tools

Euler's formula packages a sine wave and a cosine wave into one exponential:
$$ e^{i\omega t} = \cos(\omega t) + i\sin(\omega t). $$
Differentiation becomes multiplication by $i\omega$:
$$ \frac{d}{dt}e^{i\omega t} = i\omega e^{i\omega t}. $$

- **Circuits and signals:** phasors simplify sinusoidal steady-state calculations in linear circuits.  
  Fourier and Laplace transforms organize signals by frequency and growth or decay.
- **Differential equations and control:** poles describe natural modes.  
  A mode $e^{st}$ decays when $\operatorname{Re}s < 0$ and grows when $\operatorname{Re}s > 0$.
- **Potential fields:** the real and imaginary parts of a holomorphic function satisfy Laplace's equation.  
  Conformal maps help solve suitable two-dimensional electrostatic and ideal-flow problems.
- **Integration:** residues can turn difficult real integrals and inverse-transform calculations  
  into computations at a few singularities.

How much follows from complex differentiability?

### Surprise 1: Differentiability Forces a Power Series

In real calculus, differentiability does not imply analyticity.  
Even infinitely many derivatives are not enough.

If $f$ is complex differentiable **at every point of an open set** $U$,  
we call $f$ **holomorphic** on $U$.  
It is automatically infinitely differentiable and **analytic**.  
Near every $a\in U$, it equals its Taylor series:
$$ f(z) = \sum_{n = 0}^{\infty}\frac{f^{(n)}(a)}{n!}(z - a)^n. $$

Complex differentiability at just one point is not enough.  
For example, $f(z) = |z|^2$ is complex differentiable at $0$, but nowhere else, so it is not holomorphic on any neighborhood of $0$.

Why is the complex derivative so restrictive?  
The limit
$$ f'(a) = \lim_{h\to 0}\frac{f(a + h) - f(a)}{h} $$
must have the same value as $h$ approaches zero from **every direction in the complex plane**.

#### A Smooth Real Function That Is Not Analytic

Consider
$$ g(x) = \begin{cases}
e^{-1/x^2}, & x\ne 0, \\
0, & x = 0.
\end{cases} $$
This function is infinitely differentiable on $\mathbb R$, and
$$ g^{(n)}(0) = 0\qquad \text{for every }n\ge 0. $$

$e^{-1/x^2}$ tends to zero faster than every power of $|x|$.  
Away from zero, each derivative is a polynomial in $1/x$  
times that exponential.

Its Taylor series at zero is therefore the zero series.  
Yet $g(x) > 0$ for every $x\ne 0$, so the Taylor series does not represent $g$ on any neighborhood of zero.

**Real smoothness permits behavior that holomorphicity rules out.**

A related smooth, nonanalytic example appears in Section 10.7:  
[John K. Hunter's introduction to analysis](https://www.math.ucdavis.edu/~hunter/intro_analysis_pdf/intro_analysis.pdf).

#### The Mechanism: Cauchy's Integral Formula

Suppose $f$ is holomorphic on a neighborhood of the closed disk $|\zeta - a|\le r$.  
For $|z - a| < r$,
$$ f(z) = \frac{1}{2\pi i}\oint_{|\zeta - a| = r}\frac{f(\zeta)}{\zeta - z}\,d\zeta, $$
where the circle is traversed counterclockwise.

Expand the kernel as a geometric series:
$$ \frac{1}{\zeta - z} = \sum_{n = 0}^{\infty}\frac{(z - a)^n}{(\zeta - a)^{n + 1}}. $$

On every smaller disk, the series converges uniformly along the contour.  
We can integrate term by term.  
This produces the Taylor series and the derivative formula
$$ f^{(n)}(a) = \frac{n!}{2\pi i}\oint_{|\zeta - a| = r}\frac{f(\zeta)}{(\zeta - a)^{n + 1}}\,d\zeta. $$

**Values along a surrounding circle determine every value and every derivative inside.**  
This is why a local derivative condition gives such strong control.

### Complex Singularities Explain Real Approximation Limits

The real function
$$ h(x) = \frac{1}{1 + x^2} $$
is smooth for every real $x$, but its Taylor series at zero is
$$ \frac{1}{1 + x^2} = \sum_{n = 0}^{\infty}(-1)^n x^{2n}, \qquad |x| < 1. $$

Why does the series stop converging at distance $1$?  
Nothing goes wrong on the real axis.

In the complex plane,
$$ 1 + z^2 = (z - i)(z + i), $$
so there are poles at $z = \pm i$, both distance $1$ from the origin.  
These complex singularities limit the Taylor series' radius of convergence.

**Complex analysis explains the limits of a real-variable approximation.**

### Surprise 2: Extra Complex Dimensions Can Remove Singularities

*Optional preview beyond this course's main focus on one complex variable.*

In one complex variable, $1/z$ has a nonremovable singularity at $0$.  
In two or more complex variables, every isolated singularity of a holomorphic function is removable: this is a consequence of **Hartogs' extension theorem**.

More generally, let $A$ be a closed complex analytic subset  
of a domain $U\subset \mathbb C^n$,  
with **complex codimension at least two**.  
Every holomorphic function on $U\setminus A$ extends holomorphically across $A$.

Here codimension means the number of missing **complex** dimensions.  
A point in $\mathbb C^2$ has complex codimension two (real codimension four).  
A point in $\mathbb C$ has real codimension two, and $1/z$ shows that such a singularity need not be removable.

See the Hartogs discussion in [Several Complex Variables, Section 3](https://math.hse.ru/data/2024/02/05/2090218179/lektsii-24_0402.pdf).

#### Use One Variable for the Contour and Another to Bypass the Hole

Here is the idea for a missing point in $\mathbb C^2$.  
Suppose $f(z, w)$ is holomorphic on the bidisk
$$ \{(z, w) : |z| < R, \ |w| < R\}\setminus \{(0, 0)\}. $$

Choose $0 < r < R$.  
For $|z| < r$ and $|w| < R$, define
$$ F(z, w) = \frac{1}{2\pi i}\oint_{|\zeta| = r}\frac{f(\zeta, w)}{\zeta - z}\,d\zeta. $$

The contour never touches the missing point,  
even when $w = 0$, because $|\zeta| = r > 0$.  
The integrand depends holomorphically on $(z, w)$ near the contour.  
The integral therefore defines a holomorphic function,  
including at $(0, 0)$.

When $w\ne 0$, the entire $z$-disk is available.  
Cauchy's formula gives $F(z, w) = f(z, w)$.  
Continuity then gives equality for $w = 0$, $z\ne 0$, too.  
Thus $F$ fills in the missing value and agrees with $f$ wherever both are defined.

One complex variable runs the Cauchy integral.  
The other lets us move to slices without a hole.

This also explains why complex codimension one is different: $1/z$ in $\mathbb C^2$ is singular along the entire set $\{z = 0\}$.  
Moving $w$ never produces a slice without that singularity.
