## Why Were Complex Numbers Worth Studying?

### A Historical Reason: Solving Cubic Equations

During the sixteenth century, Italian mathematicians developed formulas for solving cubic equations. Work by del Ferro and Tartaglia led to the formula published by Cardano in *Ars Magna* (1545). Bombelli's *Algebra* (1572) developed rules for calculating with complex quantities and showed how they could yield real solutions of cubics.

The striking motivation was this: **a formula for a real equation could pass through square roots of negative numbers and still produce a real answer.**

Historical background: [MacTutor, Quadratic, cubic and quartic equations](https://mathshistory.st-andrews.ac.uk/HistTopics/Quadratic_etc_equations/).

The two examples below are constructed for this lecture so that all their roots are integers.

### Cardano's Formula — Stated Without Derivation

For the **depressed cubic** (a cubic with no quadratic term)

$$x^3+px+q=0,$$

set

$$\Delta=\left(\frac q2\right)^2+\left(\frac p3\right)^3.$$

One solution is

$$\boxed{x=\sqrt[3]{-\frac q2+\sqrt\Delta}+\sqrt[3]{-\frac q2-\sqrt\Delta}}.$$

**The cube roots must be chosen together.** If the two terms are $u$ and $v$, choose them so that $uv=-p/3$. For real $p,q$ and $\Delta\ge0$, the real cube roots satisfy this condition. For complex cube roots, arbitrary independent choices may fail.

A general cubic $Ax^3+Bx^2+Cx+D=0$, $A\ne0$, reduces to this form by substituting $x=t-B/(3A)$.

We will use the formula today without deriving it.

### Example 1: The Formula Finds an Integer Root

Construct a cubic with integer roots $2,-1,-1$:

$$(x-2)(x+1)^2=x^3-3x-2.$$

Now imagine we are given only the expanded equation $x^3-3x-2=0$. Cardano's formula gives

$$p=-3,\qquad q=-2,\qquad\Delta=1-1=0,$$

so

$$x=\sqrt[3]{1}+\sqrt[3]{1}=2.$$

Once the formula finds $2$, division by $x-2$ leaves

$$x^2+2x+1=(x+1)^2.$$

Thus the complete list of roots, counted with multiplicity, is

$$\boxed{2,\ -1,\ -1}.$$

The formula gives a root directly from the coefficients, without first guessing a factor. This example has a repeated root deliberately: a real cubic with three distinct real roots has $\Delta<0$.

### Example 2: Complex Intermediate Values, Integer Answers

Construct a cubic with three distinct integer roots $4,1,-5$:

$$(x-4)(x-1)(x+5)=x^3-21x+20.$$

Apply the same formula to the expanded equation:

$$p=-21,\qquad q=20,\qquad\Delta=10^2+(-7)^3=-243.$$

It asks us to calculate

$$x=\sqrt[3]{-10+9\sqrt3\,i}+\sqrt[3]{-10-9\sqrt3\,i},\qquad i^2=-1.$$

Using the ordinary algebraic rules together with $i^2=-1$, we can check that

$$(2+i\sqrt3)^3=-10+9i\sqrt3,\qquad(2-i\sqrt3)^3=-10-9i\sqrt3.$$

These are compatible cube-root choices, since

$$(2+i\sqrt3)(2-i\sqrt3)=7=-p/3.$$

Therefore

$$x=(2+i\sqrt3)+(2-i\sqrt3)=\boxed{4}.$$

Dividing the cubic by $x-4$ gives $x^2+4x-5=(x-1)(x+5)$, so the other roots are $1$ and $-5$.

**The complex terms cancel, leaving a real solution.** Even a problem whose answers are all integers gives us a reason to learn complex arithmetic.

## Complex Numbers: Algebra and Geometry

### The Form $a+bi$

A complex number has the form

$$z=a+bi,\qquad a,b\in\mathbb R,\qquad i^2=-1.$$

We write $\mathbb C=\{a+bi:a,b\in\mathbb R\}$. The representation is unique:

$$a+bi=c+di\quad\Longleftrightarrow\quad a=c\text{ and }b=d.$$

The **real part** is $\operatorname{Re}z=a$ and the **imaginary part** is $\operatorname{Im}z=b$. The imaginary part is a real number, not $bi$.

Real numbers correspond to $b=0$. Purely imaginary numbers have $a=0$.

### Arithmetic and Complex Conjugation

For $z=a+bi$ and $w=c+di$,

$$z+w=(a+c)+(b+d)i,\qquad z-w=(a-c)+(b-d)i,$$

$$zw=(ac-bd)+(ad+bc)i.$$

The **complex conjugate** of $z$ is $\bar z=a-bi$. It satisfies

$$z\bar z=a^2+b^2,\qquad\overline{z+w}=\bar z+\bar w,\qquad\overline{zw}=\bar z\,\bar w.$$

For $w\ne0$, multiply numerator and denominator by $\bar w$:

$$\frac zw=\frac{z\bar w}{w\bar w}=\frac{ac+bd}{c^2+d^2}+\frac{bc-ad}{c^2+d^2}i.$$

For example,

$$(1+2i)(3-i)=5+5i,\qquad\frac{1+2i}{3-i}=\frac{1+7i}{10}.$$

Powers of $i$ repeat every four steps: $1,i,-1,-i,1,\ldots$.

### A Representation by Real Matrices

Associate $z=a+bi$ with

$$M(z)=\begin{pmatrix}a&-b\\b&a\end{pmatrix}.$$

Then

$$M(z+w)=M(z)+M(w),\qquad M(zw)=M(z)M(w),\qquad M(1)=I.$$

In particular,

$$M(i)=J=\begin{pmatrix}0&-1\\1&0\end{pmatrix},\qquad J^2=-I.$$

Thus the rule $i^2=-1$ has a concrete realization using real matrices.

Multiplying $w=c+di$ by $z$ is exactly the real linear transformation

$$\begin{pmatrix}a&-b\\b&a\end{pmatrix}\begin{pmatrix}c\\d\end{pmatrix}=\begin{pmatrix}ac-bd\\bc+ad\end{pmatrix}.$$

Also, $\det M(z)=a^2+b^2$, so $M(z)$ is invertible exactly when $z\ne0$.

### What Happens If We Diagonalize This Matrix?

The characteristic polynomial is

$$\det(\lambda I-M(z))=(\lambda-a)^2+b^2.$$

Its eigenvalues are $a+bi=z$ and $a-bi=\bar z$. If $b\ne0$, the matrix cannot be diagonalized over $\mathbb R$, but it can be diagonalized over $\mathbb C$:

$$P=\begin{pmatrix}1&1\\-i&i\end{pmatrix},\qquad P^{-1}=\frac12\begin{pmatrix}1&i\\1&-i\end{pmatrix},$$

$$\boxed{P^{-1}M(z)P=\begin{pmatrix}z&0\\0&\bar z\end{pmatrix}}.$$

The same $P$ works for every $z$. If $b=0$, then $M(z)=aI$ is already diagonal.

For a real vector representing $w=c+di$,

$$P^{-1}\begin{pmatrix}c\\d\end{pmatrix}=\frac12\begin{pmatrix}w\\\bar w\end{pmatrix}.$$

In these complex coordinates, multiplication acts separately by $z$ and $\bar z$. This explains why a real matrix can encode complex multiplication and why conjugate pairs appear naturally.

### A Vector Space over $\mathbb R$

As a real vector space, $\mathbb C$ is two-dimensional with basis $\{1,i\}$:

$$a+bi\ \longleftrightarrow\ \begin{pmatrix}a\\b\end{pmatrix}\in\mathbb R^2.$$

Addition is vector addition, and multiplication by a real number scales both coordinates. Thus $\mathbb C\cong\mathbb R^2$ **as real vector spaces**.

Complex multiplication is additional structure: it is not componentwise multiplication of vectors. Multiplication by a fixed $z$ is the real linear map $M(z)$ above.

The dimension depends on the scalar field:

$$\dim_{\mathbb R}\mathbb C=2,\qquad\dim_{\mathbb C}\mathbb C=1.$$

As a complex vector space, $\mathbb C$ has basis $\{1\}$.

### Modulus and Distance

The **modulus** of $z=a+bi$ is its Euclidean length:

$$|z|=\sqrt{a^2+b^2}=\sqrt{z\bar z}.$$

The distance between $z$ and $w$ is $|z-w|$. For example, $|3+4i|=5$.

Useful identities are

$$|z|\ge0,\qquad |z|=0\Longleftrightarrow z=0,\qquad |\bar z|=|z|,$$

$$|zw|=|z||w|,\qquad\left|\frac zw\right|=\frac{|z|}{|w|}\quad(w\ne0).$$

For instance,

$$|zw|^2=(zw)\overline{zw}=(z\bar z)(w\bar w)=|z|^2|w|^2.$$

Taking nonnegative square roots gives the multiplication identity.

### The Triangle Inequality

For all complex numbers $z,w$,

$$\boxed{|z+w|\le |z|+|w|}.$$

Geometrically, the direct distance is no longer than the path taking two successive steps.

Algebraically, use $\operatorname{Re}(z\bar w)\le|z\bar w|=|z||w|$:

$$\begin{aligned}
|z+w|^2&=|z|^2+2\operatorname{Re}(z\bar w)+|w|^2\\
&\le |z|^2+2|z||w|+|w|^2=(|z|+|w|)^2.
\end{aligned}$$

Equality holds when either number is zero or the two nonzero numbers point in the same direction.

Applying the triangle inequality to $z=(z-w)+w$ and then interchanging $z,w$ gives the **reverse triangle inequality**:

$$\bigl||z|-|w|\bigr|\le |z-w|.$$

These inequalities let us estimate complex expressions without calculating their exact values.

### Argument: An Angle Defined Modulo $2\pi$

For a nonzero number $z=a+bi$, an **argument** is any angle $\theta$ satisfying

$$\cos\theta=\frac a{|z|},\qquad\sin\theta=\frac b{|z|}.$$

If $\theta$ is an argument, so is $\theta+2\pi k$ for every integer $k$. We use the principal argument convention

$$\operatorname{Arg}z\in(-\pi,\pi],\qquad\arg z=\{\operatorname{Arg}z+2\pi k:k\in\mathbb Z\}.$$

Examples:

$$\operatorname{Arg}(1+i)=\frac\pi4,\quad\operatorname{Arg}(-1+i)=\frac{3\pi}4,\quad\operatorname{Arg}(-1)=\pi.$$

Zero has no argument. The principal argument jumps across the negative real axis; choosing a single angle introduces a discontinuity.

### Cartesian and Polar Coordinates

Cartesian coordinates use $(a,b)$; polar coordinates use $(r,\theta)$:

$$z=a+bi=r(\cos\theta+i\sin\theta),\qquad r=|z|.$$

The conversions are

$$a=r\cos\theta,\qquad b=r\sin\theta,\qquad r=\sqrt{a^2+b^2}.$$

For $z\ne0$, choose $\theta$ using both coordinates and the correct quadrant. The expression $\arctan(b/a)$ alone loses quadrant information and is undefined when $a=0$; a quadrant-aware function such as $\operatorname{atan2}(b,a)$ is useful.

For example,

$$-1+i=\sqrt2\left(\cos\frac{3\pi}4+i\sin\frac{3\pi}4\right).$$

Euler's formula, $e^{i\theta}=\cos\theta+i\sin\theta$, gives the compact notation $z=re^{i\theta}$. We will justify the complex exponential later; for now the sine-and-cosine form contains all the geometry.

### Multiplication Scales Lengths and Adds Angles

Let $z=r(\cos\alpha+i\sin\alpha)$ and $w=s(\cos\beta+i\sin\beta)$. Expanding their product gives

$$zw=rs\bigl[(\cos\alpha\cos\beta-\sin\alpha\sin\beta)+i(\sin\alpha\cos\beta+\cos\alpha\sin\beta)\bigr].$$

The angle addition identities turn this into

$$\boxed{zw=rs\bigl[\cos(\alpha+\beta)+i\sin(\alpha+\beta)\bigr]}.$$

Moduli multiply, and arguments add **modulo $2\pi$**. Division divides moduli and subtracts arguments.

Equivalently,

$$M(z)=r\begin{pmatrix}\cos\alpha&-\sin\alpha\\\sin\alpha&\cos\alpha\end{pmatrix}.$$

Multiplication by $z\ne0$ is a rotation through $\alpha$ followed by scaling by $r$. In particular, multiplying by $i$ rotates a vector counterclockwise by $90^\circ$.

### Complex Multiplication and the Tangent Addition Formula

When $\tan\alpha$ and $\tan\beta$ are defined,

$$(1+i\tan\alpha)(1+i\tan\beta)=(1-\tan\alpha\tan\beta)+i(\tan\alpha+\tan\beta).$$

Taking the ratio of imaginary to real parts of the product yields

$$\boxed{\tan(\alpha+\beta)=\frac{\tan\alpha+\tan\beta}{1-\tan\alpha\tan\beta}},$$

provided $\cos\alpha\ne0$, $\cos\beta\ne0$, and $\cos(\alpha+\beta)\ne0$.

For example,

$$(1+i)(1+2i)=-1+3i.$$

Thus, for $\alpha=\pi/4$ and $\beta=\arctan(2)$, the tangent of the sum is $(1+2)/(1-2)=-3$. The product lies in quadrant II, so the angle sum is $\pi-\arctan(3)$, not a negative angle.

**Tangent records a slope, so it determines an angle only modulo $\pi$.** The signs of both coordinates retain the quadrant information.

### Powers and Geometric Sequences

Repeated multiplication gives **de Moivre's formula**:

$$\bigl[r(\cos\theta+i\sin\theta)\bigr]^n=r^n\bigl[\cos(n\theta)+i\sin(n\theta)\bigr],\qquad n\ge1.$$

A complex geometric sequence has the form

$$a_n=a_0q^n,\qquad n=0,1,2,\ldots.$$

If $a_0\ne0$ and $q=\rho e^{i\theta}\ne0$, each step multiplies the length by $\rho$ and rotates by $\theta$:

- $|q|<1$: $a_n\to0$.
- $|q|>1$: $|a_n|\to\infty$.
- $|q|=1$: the points stay on a circle. They are periodic exactly when $\theta/(2\pi)$ is rational.

For $a_0=1$, the ratio $q=i$ gives $1,i,-1,-i,1,\ldots$. The ratio $q=i/2$ gives $1,i/2,-1/4,-i/8,\ldots$, rotating while shrinking toward zero.

If $a_0=0$, the sequence is identically zero. If $q=0$, every term after the initial term is zero.

### Finite and Infinite Geometric Sums

For $S_N=1+q+q^2+\cdots+q^N$, subtraction gives

$$(1-q)S_N=1-q^{N+1}.$$

Therefore

$$\boxed{\sum_{n=0}^Nq^n=\frac{1-q^{N+1}}{1-q}}\quad(q\ne1).$$

If $q=1$, the sum is $N+1$. If $|q|<1$, then $q^{N+1}\to0$, so

$$\boxed{\sum_{n=0}^{\infty}q^n=\frac1{1-q}}.$$

For $|q|\ge1$, the terms $q^n$ do not tend to zero, so the infinite series diverges.

For example,

$$1+\frac i2-\frac14-\frac i8+\cdots=\frac1{1-i/2}=\frac45+\frac25i.$$

This same geometric-series identity will later expand the kernel in Cauchy's integral formula and produce Taylor series.

## Why Learn Complex Analysis?

Complex numbers turn rotations and oscillations into algebra. Complex analysis goes further: it connects derivatives, power series, contour integrals, and geometry in ways that real calculus alone cannot.

For engineers, these connections provide tools for understanding signals, circuits, differential equations, and two-dimensional fields.

### From Oscillations to Engineering Tools

Euler's formula packages a sine wave and a cosine wave into one exponential:

$$e^{i\omega t}=\cos(\omega t)+i\sin(\omega t).$$

Differentiation becomes multiplication by $i\omega$:

$$\frac{d}{dt}e^{i\omega t}=i\omega e^{i\omega t}.$$

- **Circuits and signals:** phasors simplify sinusoidal steady-state calculations in linear circuits; Fourier and Laplace transforms organize signals by frequency and growth or decay.
- **Differential equations and control:** poles describe natural modes. A mode $e^{st}$ decays when $\operatorname{Re}s<0$ and grows when $\operatorname{Re}s>0$.
- **Potential fields:** the real and imaginary parts of a holomorphic function satisfy Laplace's equation. Conformal maps help solve suitable two-dimensional electrostatic and ideal-flow problems.
- **Integration:** residues can turn difficult real integrals and inverse-transform calculations into computations at a few singularities.

The surprising part is how much follows from a single condition: complex differentiability.

### Surprise 1: Differentiability Forces a Power Series

In real calculus, differentiability does not imply analyticity. Even infinitely many derivatives are not enough.

In complex analysis, if $f$ is complex differentiable **at every point of an open set** $U$, then $f$ is called **holomorphic** on $U$. It is automatically infinitely differentiable and **analytic**: near every $a\in U$, it equals its Taylor series,

$$f(z)=\sum_{n=0}^{\infty}\frac{f^{(n)}(a)}{n!}(z-a)^n.$$

Complex differentiability at just one point is not enough. For example, $f(z)=|z|^2$ is complex differentiable at $0$, but nowhere else, so it is not holomorphic on any neighborhood of $0$.

Why is the complex derivative so restrictive? The limit

$$f'(a)=\lim_{h\to0}\frac{f(a+h)-f(a)}{h}$$

must have the same value as $h$ approaches zero from **every direction in the complex plane**.

#### A Smooth Real Function That Is Not Analytic

Consider

$$
g(x)=\begin{cases}
e^{-1/x^2},&x\ne0,\\
0,&x=0.
\end{cases}
$$

This function is infinitely differentiable on $\mathbb R$, and

$$g^{(n)}(0)=0\qquad\text{for every }n\ge0.$$

The reason is that $e^{-1/x^2}$ tends to zero faster than every power of $|x|$; away from zero, each derivative is a polynomial in $1/x$ times that exponential.

Its Taylor series at zero is therefore the zero series. Yet $g(x)>0$ for every $x\ne0$, so the Taylor series does not represent $g$ on any neighborhood of zero.

**Real smoothness permits behavior that holomorphicity rules out.**

For a related smooth, nonanalytic example, see Section 10.7 of [John K. Hunter's introduction to analysis](https://www.math.ucdavis.edu/~hunter/intro_analysis_pdf/intro_analysis.pdf).

#### The Mechanism: Cauchy's Integral Formula

Suppose $f$ is holomorphic on a neighborhood of the closed disk $|\zeta-a|\le r$. For $|z-a|<r$,

$$f(z)=\frac{1}{2\pi i}\oint_{|\zeta-a|=r}\frac{f(\zeta)}{\zeta-z}\,d\zeta,$$

where the circle is traversed counterclockwise.

Expand the kernel as a geometric series:

$$\frac{1}{\zeta-z}=\sum_{n=0}^{\infty}\frac{(z-a)^n}{(\zeta-a)^{n+1}}.$$

On every smaller disk, this series converges uniformly along the contour, so we can integrate term by term. This produces the Taylor series and the derivative formula

$$f^{(n)}(a)=\frac{n!}{2\pi i}\oint_{|\zeta-a|=r}\frac{f(\zeta)}{(\zeta-a)^{n+1}}\,d\zeta.$$

**Values along a surrounding circle determine every value and every derivative inside.** This is why a local derivative condition gives such strong control.

### Complex Singularities Explain Real Approximation Limits

The real function

$$h(x)=\frac{1}{1+x^2}$$

is smooth for every real $x$, but its Taylor series at zero is

$$\frac{1}{1+x^2}=\sum_{n=0}^{\infty}(-1)^n x^{2n},\qquad |x|<1.$$

Why does the series stop converging at distance $1$ when nothing goes wrong on the real axis?

In the complex plane,

$$1+z^2=(z-i)(z+i),$$

so there are poles at $z=\pm i$, both distance $1$ from the origin. These complex singularities limit the Taylor series' radius of convergence.

**Complex analysis explains the limits of a real-variable approximation.**

### Surprise 2: Extra Complex Dimensions Can Remove Singularities

*Optional preview beyond this course's main focus on one complex variable.*

In one complex variable, $1/z$ has a nonremovable singularity at $0$. In two or more complex variables, every isolated singularity of a holomorphic function is removable: this is a consequence of **Hartogs' extension theorem**.

More generally, if $A$ is a closed complex analytic subset of a domain $U\subset\mathbb C^n$, with **complex codimension at least two**, every holomorphic function on $U\setminus A$ extends holomorphically across $A$.

Here codimension means the number of missing **complex** dimensions. A point in $\mathbb C^2$ has complex codimension two (real codimension four). A point in $\mathbb C$ has real codimension two, and $1/z$ shows that such a singularity need not be removable.

See the Hartogs discussion in [Several Complex Variables, Section 3](https://math.hse.ru/data/2024/02/05/2090218179/lektsii-24_0402.pdf).

#### Use One Variable for the Contour and Another to Bypass the Hole

Here is the idea for a missing point in $\mathbb C^2$. Suppose $f(z,w)$ is holomorphic on the bidisk

$$\{(z,w):|z|<R,\ |w|<R\}\setminus\{(0,0)\}.$$

Choose $0<r<R$. For $|z|<r$ and $|w|<R$, define

$$F(z,w)=\frac{1}{2\pi i}\oint_{|\zeta|=r}\frac{f(\zeta,w)}{\zeta-z}\,d\zeta.$$

The contour never touches the missing point, even when $w=0$, because $|\zeta|=r>0$. The integrand depends holomorphically on $(z,w)$ near the contour, so the integral defines a holomorphic function there, including at $(0,0)$.

When $w\ne0$, the entire $z$-disk is available. Cauchy's formula gives $F(z,w)=f(z,w)$. Continuity then gives equality for $w=0$, $z\ne0$, too. Thus $F$ fills in the missing value and agrees with $f$ wherever both are defined.

**One complex variable runs the Cauchy integral; the other lets us move to slices without a hole.**

This also explains why complex codimension one is different: $1/z$ in $\mathbb C^2$ is singular along the entire set $\{z=0\}$. Moving $w$ never produces a slice without that singularity.
